# Matplotlib — Python's Foundation for Data Visualization

## What is Matplotlib?

Matplotlib is Python's **original plotting library** — the one that everything else is built on. It gives you complete control over every pixel of your chart. Think of it like Microsoft Paint vs. a fully equipped art studio: Paint is limited, but once you know the studio, you can create anything.

**Real-world analogy**: Matplotlib is the **HTML of data visualization**. Just as HTML gives you raw control over a web page (but requires more code than a website builder), Matplotlib gives you raw control over every aspect of a plot. Higher-level libraries like Seaborn are like CSS frameworks — they hide the complexity but use HTML (Matplotlib) underneath.

## Why Learn Matplotlib First?

1. **It's everywhere** — Seaborn, Pandas `.plot()`, Scikit-learn's `plot_tree()` all produce Matplotlib figures
2. **Full control** — when Seaborn doesn't do exactly what you want, you drop into Matplotlib
3. **Publication-quality figures** — researchers use Matplotlib for journal papers
4. **Career essential** — virtually every data science job asks about it

## Prerequisites
- Python basics (functions, loops)
- NumPy arrays (most plot data is NumPy arrays)

## 2. The Two APIs: `plt` (MATLAB-style) vs Object-Oriented

Matplotlib has two ways to create plots. **Both work**, but they're used differently:

| Style | When to Use | Example |
|-------|------------|--------|
| **pyplot (`plt`)** | Quick, exploratory plots | `plt.plot(x, y); plt.show()` |
| **Object-Oriented (OO)** | Multiple subplots, complex customization | `fig, ax = plt.subplots(); ax.plot(x, y)` |

**Analogy**: `plt` is like talking to a restaurant — you just say "I want pasta" and they figure out the rest. The OO API is like being the chef — you control every step.

> **Best practice**: Use the OO API for anything beyond a single quick plot. Most professional code uses `fig, ax = plt.subplots()`.

````python
x = np.linspace(0, 2 * np.pi, 100)

# ── Method 1: pyplot (MATLAB) style ──────────────────────────────────────────
plt.figure(figsize=(8, 3))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('MATLAB-style API (plt)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()  # <-- required to display in scripts; optional in Jupyter

# ── Method 2: Object-Oriented API (preferred) ─────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 3))  # fig = the whole canvas, ax = one plot area
ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')
ax.set_title('Object-Oriented API (ax)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Key difference: OO uses ax.set_title() instead of plt.title()")
print("OO is preferred for multi-panel figures and when passing plots to functions.")
````

## 3. Basic Plot Types

### When to use which chart?

| Chart Type | Best For | Example |
|-----------|----------|--------|
| **Line** | Trends over time | Stock prices, temperature over a day |
| **Scatter** | Relationship between two variables | Height vs weight |
| **Bar** | Comparing categories | Sales by region |
| **Histogram** | Distribution of one variable | Age distribution of users |
| **Box** | Distribution + outliers | Salary ranges by department |
| **Pie** | Part-to-whole (use sparingly!) | Market share |
| **Heatmap (imshow)** | Matrix data | Correlation matrix, image |

````python
np.random.seed(42)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# ── Line Plot: Stock Price ─────────────────────────────────────────────────────
days = np.arange(1, 31)
price = 100 + np.cumsum(np.random.randn(30))  # random walk
axes[0, 0].plot(days, price, color='steelblue', linewidth=2)
axes[0, 0].fill_between(days, price, 100, alpha=0.3, color='steelblue')
axes[0, 0].set_title('Line Plot: Stock Price')
axes[0, 0].set_xlabel('Day')
axes[0, 0].set_ylabel('Price ($)')
axes[0, 0].axhline(100, color='red', linestyle='--', alpha=0.5, label='Start')
axes[0, 0].legend()

# ── Scatter Plot: Height vs Weight ────────────────────────────────────────────
heights = np.random.normal(170, 10, 150)
weights = 0.5 * heights - 10 + np.random.normal(0, 5, 150)
colors  = np.where(heights > 170, 'blue', 'orange')  # color by height group
axes[0, 1].scatter(heights, weights, c=colors, alpha=0.6, s=30, edgecolors='white', linewidths=0.5)
axes[0, 1].set_title('Scatter Plot: Height vs Weight')
axes[0, 1].set_xlabel('Height (cm)')
axes[0, 1].set_ylabel('Weight (kg)')

# ── Bar Chart: Sales by Region ────────────────────────────────────────────────
regions = ['North', 'South', 'East', 'West', 'Central']
sales   = [45, 62, 38, 71, 55]
bar_colors = ['#2ecc71' if s >= 60 else '#e74c3c' for s in sales]  # green if good
bars = axes[0, 2].bar(regions, sales, color=bar_colors, edgecolor='white', linewidth=0.5)
axes[0, 2].bar_label(bars, fmt='%d', padding=3)
axes[0, 2].set_title('Bar Chart: Sales by Region')
axes[0, 2].set_ylabel('Sales (units)')
axes[0, 2].axhline(np.mean(sales), color='navy', linestyle='--', label=f'Avg: {np.mean(sales):.0f}')
axes[0, 2].legend()

# ── Histogram: Age Distribution ────────────────────────────────────────────────
ages = np.concatenate([np.random.normal(28, 5, 200), np.random.normal(50, 8, 100)])
axes[1, 0].hist(ages, bins=30, color='mediumpurple', edgecolor='white', linewidth=0.5)
axes[1, 0].axvline(np.mean(ages), color='red', linestyle='--', label=f'Mean: {np.mean(ages):.1f}')
axes[1, 0].set_title('Histogram: User Ages')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Count')
axes[1, 0].legend()

# ── Box Plot: Salary by Department ────────────────────────────────────────────
departments = ['Eng', 'Sales', 'Marketing', 'HR', 'Finance']
salary_data = [np.random.normal(m, s, 50) 
               for m, s in [(95, 15), (70, 20), (75, 12), (60, 10), (85, 18)]]
bp = axes[1, 1].boxplot(salary_data, labels=departments, patch_artist=True,
                         medianprops=dict(color='red', linewidth=2))
colors_box = ['lightblue', 'lightgreen', 'lightyellow', 'lightpink', 'lightgray']
for patch, color in zip(bp['boxes'], colors_box):
    patch.set_facecolor(color)
axes[1, 1].set_title('Box Plot: Salary by Department (K$)')
axes[1, 1].set_ylabel('Salary (K$)')

# ── Pie Chart: Market Share ────────────────────────────────────────────────────
companies = ['Company A', 'Company B', 'Company C', 'Others']
shares    = [35, 28, 22, 15]
explode   = [0.05, 0.05, 0.05, 0.05]
axes[1, 2].pie(shares, labels=companies, autopct='%1.1f%%', explode=explode,
                startangle=90, colors=['#3498db', '#e74c3c', '#2ecc71', '#95a5a6'])
axes[1, 2].set_title('Pie Chart: Market Share')

plt.suptitle('Matplotlib: Core Chart Types', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 4. Figure Anatomy: Figure, Axes, Artists

Understanding this hierarchy is the key to mastering Matplotlib:

```
Figure  ──── The entire canvas (the window/image)
  │
  ├── Axes  ── One plot area (has x-axis, y-axis, title, etc.)
  │    ├── Axis (XAxis, YAxis)  ── The number lines
  │    │    ├── Tick, TickLabel  ── The tick marks and numbers
  │    │    └── Label            ── "x-axis label"
  │    ├── Title                 ── Axes title
  │    ├── Line2D                ── Lines from plt.plot()
  │    ├── PathCollection        ── Points from plt.scatter()
  │    └── Rectangle             ── Bars from plt.bar()
  └── FigureText                 ── fig.suptitle()
```

**Analogy**: Figure is the canvas. Axes are the frames. Everything inside a frame (lines, dots, bars) is an Artist.

````python
# Demonstrate the anatomy with labeled arrows
fig, ax = plt.subplots(figsize=(10, 5))

x = np.linspace(0, 10, 100)
line, = ax.plot(x, np.sin(x), 'b-', linewidth=2, label='sin(x)')  # Line2D artist

# Title
ax.set_title('Matplotlib Figure Anatomy', fontsize=14, fontweight='bold')

# Axis labels
ax.set_xlabel('X Axis (time, index, etc.)', fontsize=12)
ax.set_ylabel('Y Axis (values)', fontsize=12)

# Legend
ax.legend(fontsize=11)

# Grid
ax.grid(True, linestyle='--', alpha=0.5)

# Custom tick marks
ax.set_xticks([0, np.pi, 2*np.pi, 3*np.pi])  # Where ticks appear
ax.set_xticklabels(['0', 'π', '2π', '3π'])   # Labels for ticks

# Axis limits
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

# Add a text annotation with an arrow
ax.annotate('Peak at π/2',
            xy=(np.pi/2, 1), xytext=(2, 1.3),
            arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=10, color='red')

# Add a shaded region
ax.axvspan(0, np.pi, alpha=0.1, color='green', label='First period')

# Report on the objects
print(f"Figure type:  {type(fig)}")
print(f"Axes type:    {type(ax)}")
print(f"Line type:    {type(line)}")
print(f"All artists:  {[type(a).__name__ for a in ax.get_children()[:6]]}...")

plt.tight_layout()
plt.show()
````

## 5. Subplots: Multiple Plots in One Figure

Subplots let you place multiple charts side-by-side in one figure.

```python
fig, axes = plt.subplots(nrows, ncols)   # returns a 2D array of Axes
fig, axes = plt.subplots(1, 3)           # 1 row, 3 columns
fig, axes = plt.subplots(2, 2)           # 2x2 grid
```

````python
np.random.seed(0)

# ── 2x2 grid of subplots ──────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Top-left: Multiple lines
x = np.linspace(0, 2*np.pi, 200)
for n in range(1, 5):
    axes[0, 0].plot(x, np.sin(n * x) / n, label=f'sin({n}x)/{n}')
axes[0, 0].set_title('Fourier Components')
axes[0, 0].legend(fontsize=8)
axes[0, 0].grid(True, alpha=0.3)

# Top-right: Scatter with color mapped to a third variable
x2 = np.random.randn(200)
y2 = x2 * 2 + np.random.randn(200)
z  = x2**2 + y2**2  # distance from origin → color
sc = axes[0, 1].scatter(x2, y2, c=z, cmap='viridis', alpha=0.7, s=40)
plt.colorbar(sc, ax=axes[0, 1], label='Distance from origin')
axes[0, 1].set_title('Scatter with Colormap')

# Bottom-left: Horizontal bar chart
categories = ['Python', 'JavaScript', 'Java', 'C++', 'R', 'Go']
popularity = [75, 65, 55, 40, 35, 25]
axes[1, 0].barh(categories, popularity, color='coral', edgecolor='white')
axes[1, 0].set_title('Horizontal Bar: Language Popularity')
axes[1, 0].set_xlabel('Popularity Index')

# Bottom-right: Step plot (good for inventory, events)
time = np.arange(0, 20)
inventory = np.array([100, 100, 85, 85, 85, 60, 60, 120, 120, 95, 95, 80, 80, 80, 50, 50, 90, 90, 70, 70])
axes[1, 1].step(time, inventory, where='post', color='purple', linewidth=2)
axes[1, 1].fill_between(time, inventory, step='post', alpha=0.2, color='purple')
axes[1, 1].set_title('Step Plot: Inventory Level')
axes[1, 1].set_xlabel('Day')
axes[1, 1].set_ylabel('Units')
axes[1, 1].axhline(70, color='red', linestyle='--', label='Reorder point')
axes[1, 1].legend()

fig.suptitle('Subplots: Multiple Charts in One Figure', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ── Irregular subplot layout with GridSpec ─────────────────────────────────────
fig = plt.figure(figsize=(12, 4))
gs  = fig.add_gridspec(1, 3, width_ratios=[2, 1, 1], wspace=0.3)

ax_wide  = fig.add_subplot(gs[0])   # Takes 2/4 of width
ax_mid   = fig.add_subplot(gs[1])   # Takes 1/4
ax_right = fig.add_subplot(gs[2])   # Takes 1/4

ax_wide.plot(x, np.sin(x)); ax_wide.set_title('Wide (2/4)')
ax_mid.bar(['A','B'], [3, 5]); ax_mid.set_title('Narrow (1/4)')
ax_right.hist(np.random.randn(100), bins=15); ax_right.set_title('Narrow (1/4)')

fig.suptitle('GridSpec: Irregular Layout', fontsize=12)
plt.show()
````

## 6. Styling: Colors, Markers, Linestyles, Themes

### Color Formats
- Named colors: `'red'`, `'steelblue'`, `'tomato'`, `'lime'`
- Hex: `'#FF5733'`
- RGB tuple: `(0.2, 0.4, 0.8)`
- Colormap letter: `'C0'`, `'C1'` (cycles through default colors)

### Line Styles
`'-'` solid, `'--'` dashed, `'-.'` dash-dot, `':'` dotted

### Markers
`'o'` circle, `'s'` square, `'^'` triangle, `'*'` star, `'+'` plus, `'D'` diamond

### Themes
Use `plt.style.use()` to change the entire look of all plots.

````python
# ── Available styles ─────────────────────────────────────────────────────────
print("Available styles:", plt.style.available[:10], '...')

# ── Compare popular styles ────────────────────────────────────────────────────
x = np.linspace(0, 10, 50)
y = np.sin(x)

styles = ['default', 'ggplot', 'seaborn-v0_8', 'fivethirtyeight']
fig, axes = plt.subplots(1, 4, figsize=(16, 3))

for ax, style in zip(axes, styles):
    with plt.style.context(style):  # Apply style temporarily
        ax.plot(x, y, label='sin(x)')
        ax.set_title(style, fontsize=9)
        ax.legend(fontsize=8)

plt.suptitle('Style Comparison', fontsize=12)
plt.tight_layout()
plt.show()

# ── Line style and marker reference ────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 4))

line_styles = ['-', '--', '-.', ':']
markers     = ['o', 's', '^', 'D', '*']
colors      = ['#e74c3c', '#3498db', '#2ecc71', '#f39c12', '#9b59b6']

x_small = np.linspace(0, 2*np.pi, 15)
for i, (ls, marker, color) in enumerate(zip(line_styles * 2, markers, colors)):
    ax.plot(x_small, np.sin(x_small) + i * 0.5, 
            linestyle=ls, marker=marker, color=color,
            markersize=7, linewidth=1.5,
            label=f'ls={ls!r}, marker={marker!r}')

ax.set_title('Line Styles & Markers Reference')
ax.legend(bbox_to_anchor=(1, 1), loc='upper left', fontsize=9)  # Legend outside
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 7. Annotations & Text

Annotations turn raw charts into stories. They highlight key moments, label important points, or explain context.

**Analogy**: Think of annotations as sticky notes on a chart — they draw attention to what matters most.

````python
np.random.seed(42)

# Simulate monthly revenue with a notable spike
months = np.arange(1, 13)
revenue = [120, 132, 128, 145, 155, 162, 200, 175, 168, 180, 195, 220]  # July spike

fig, ax = plt.subplots(figsize=(11, 5))

# Main line
ax.plot(months, revenue, 'o-', color='steelblue', linewidth=2.5, markersize=8, label='Monthly Revenue')

# Fill below the line
ax.fill_between(months, revenue, min(revenue), alpha=0.15, color='steelblue')

# ── Annotation with arrow ──────────────────────────────────────────────────────
ax.annotate('Summer Sale\nCampaign (+24%)',
            xy=(7, 200),           # Point being annotated (arrow tip)
            xytext=(8.5, 185),     # Where the text is placed
            fontsize=10, color='darkred',
            arrowprops=dict(arrowstyle='->', color='darkred', lw=1.5),
            bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', edgecolor='darkred'))

# ── Horizontal reference line with text ───────────────────────────────────────
target = 160
ax.axhline(target, color='green', linestyle='--', linewidth=1.5, alpha=0.8)
ax.text(0.5, target + 2, f'Target: ${target}K', fontsize=9, color='green')

# ── Mark specific months ───────────────────────────────────────────────────────
max_month = months[np.argmax(revenue)]
max_rev   = max(revenue)
ax.scatter([max_month], [max_rev], color='gold', s=200, zorder=5, edgecolors='darkorange', linewidths=2)
ax.text(max_month, max_rev + 4, f'Peak: ${max_rev}K', ha='center', fontsize=9, fontweight='bold')

# ── Figure-level text ─────────────────────────────────────────────────────────
ax.text(0.98, 0.05, 'FY 2024', transform=ax.transAxes,  # transAxes: 0-1 coordinates
        fontsize=12, ha='right', va='bottom', color='gray', alpha=0.5)

# Formatting
month_names = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
ax.set_xticks(months)
ax.set_xticklabels(month_names)
ax.set_title('Monthly Revenue with Annotations', fontsize=14, fontweight='bold')
ax.set_ylabel('Revenue ($K)')
ax.legend()
ax.grid(True, axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
````

## 8. Axes Customization

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

x = np.logspace(0, 4, 100)  # 1 to 10,000
y = x ** 0.5

# ── 1. Log scale ───────────────────────────────────────────────────────────────
axes[0].plot(x, y, 'b-')
axes[0].set_xscale('log')  # Log scale on x-axis
axes[0].set_title('Log Scale (x-axis)')
axes[0].set_xlabel('x (log scale)')
axes[0].set_ylabel('√x')
axes[0].grid(True, which='both', alpha=0.3)  # Grid on both major/minor ticks

# ── 2. Twin axes (two y-axes) ──────────────────────────────────────────────────
t = np.linspace(0, 10, 100)
temp = 20 + 10 * np.sin(t)
pressure = 1013 + 15 * np.cos(t)

ax2 = axes[1]
ax2_twin = ax2.twinx()  # Create a second y-axis sharing the same x-axis
line1, = ax2.plot(t, temp, 'r-', label='Temperature (°C)')
line2, = ax2_twin.plot(t, pressure, 'b-', label='Pressure (hPa)', alpha=0.7)
ax2.set_ylabel('Temperature (°C)', color='red')
ax2_twin.set_ylabel('Pressure (hPa)', color='blue')
ax2.tick_params(axis='y', labelcolor='red')
ax2_twin.tick_params(axis='y', labelcolor='blue')
ax2.set_title('Twin Axes: Two Y Scales')
ax2.legend(handles=[line1, line2], loc='upper right', fontsize=8)

# ── 3. Inset / Zoomed-in subplot ───────────────────────────────────────────────
x_full = np.linspace(0, 2*np.pi, 300)
axes[2].plot(x_full, np.sin(x_full), 'b-')
axes[2].set_title('Inset Axes (Zoom-in)')

# Add an inset axis showing a zoomed-in region
axins = axes[2].inset_axes([0.55, 0.4, 0.4, 0.4])  # [x, y, width, height] in axes coords
axins.plot(x_full, np.sin(x_full), 'b-')
axins.set_xlim(1.0, 2.5)  # Zoom in on this region
axins.set_ylim(0.7, 1.1)
axins.set_title('Zoomed', fontsize=8)
axes[2].indicate_inset_zoom(axins, edgecolor='red')  # Draw the zoom box

plt.suptitle('Axes Customization: Log Scale, Twin Axes, Inset', fontsize=12)
plt.tight_layout()
plt.show()
````

## 9. Advanced: Colormaps, 2D plots, Heatmaps

Colormaps map a number to a color. They're essential for heatmaps, scatter plots with a third dimension, contour plots, and images.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# ── 1. Heatmap (imshow) ───────────────────────────────────────────────────────
# Correlation matrix for 5 variables
np.random.seed(0)
data  = np.random.randn(5, 100)
corr  = np.corrcoef(data)
var_names = ['Sales', 'Ads', 'Price', 'Time', 'Quality']

im = axes[0].imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')
plt.colorbar(im, ax=axes[0], label='Correlation')
axes[0].set_xticks(range(5)); axes[0].set_yticks(range(5))
axes[0].set_xticklabels(var_names, rotation=45, ha='right', fontsize=9)
axes[0].set_yticklabels(var_names, fontsize=9)
# Add correlation values as text
for i in range(5):
    for j in range(5):
        axes[0].text(j, i, f'{corr[i,j]:.2f}', ha='center', va='center', 
                     fontsize=8, color='white' if abs(corr[i,j]) > 0.5 else 'black')
axes[0].set_title('Heatmap: Correlation Matrix')

# ── 2. Contour Plot ────────────────────────────────────────────────────────────
x_grid = np.linspace(-3, 3, 100)
y_grid = np.linspace(-3, 3, 100)
X, Y   = np.meshgrid(x_grid, y_grid)
Z      = np.exp(-(X**2 + Y**2))  # Gaussian bump

contour = axes[1].contourf(X, Y, Z, levels=20, cmap='hot')
axes[1].contour(X, Y, Z, levels=5, colors='white', alpha=0.5, linewidths=0.5)
plt.colorbar(contour, ax=axes[1], label='Probability Density')
axes[1].set_title('Contour: 2D Gaussian')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')

# ── 3. 2D Histogram ───────────────────────────────────────────────────────────
x_samples = np.random.randn(5000)
y_samples = x_samples * 0.7 + np.random.randn(5000) * 0.5
h, xedges, yedges, img = axes[2].hist2d(x_samples, y_samples, bins=40, cmap='Blues')
plt.colorbar(img, ax=axes[2], label='Count')
axes[2].set_title('2D Histogram: Joint Distribution')
axes[2].set_xlabel('x')
axes[2].set_ylabel('y')

plt.suptitle('Advanced: Heatmaps, Contours, 2D Histograms', fontsize=12)
plt.tight_layout()
plt.show()
````

## 10. Saving Figures

Always save figures programmatically — don't screenshot! Use `fig.savefig()` to export.

| Format | Use Case |
|--------|----------|
| PNG | Web, presentations (raster, not scalable) |
| PDF | Papers, printing (vector, scalable) |
| SVG | Web graphics, further editing in Illustrator |
| EPS | Some academic journals require this |

````python
import os

fig, ax = plt.subplots(figsize=(8, 4))
x = np.linspace(0, 2*np.pi, 100)
ax.plot(x, np.sin(x), 'b-', linewidth=2)
ax.set_title('Saved Figure Example')
ax.set_xlabel('x'); ax.set_ylabel('sin(x)')
ax.grid(True, alpha=0.3)

# Save in multiple formats
fig.savefig('/tmp/sine_wave.png', dpi=150, bbox_inches='tight')  # PNG at 150 DPI
fig.savefig('/tmp/sine_wave.pdf', bbox_inches='tight')            # PDF (vector)
fig.savefig('/tmp/sine_wave.svg', bbox_inches='tight')            # SVG (vector)

plt.show()

for ext in ['png', 'pdf', 'svg']:
    path = f'/tmp/sine_wave.{ext}'
    size = os.path.getsize(path)
    print(f"{ext.upper():4}: {path}  ({size:,} bytes)")

print("\nTips:")
print("  bbox_inches='tight'  → removes unnecessary whitespace around the figure")
print("  dpi=300              → high resolution for print (use 150 for screen)")
print("  transparent=True     → transparent background (useful for presentations)")
````

## 11. Common Pitfalls & Best Practices

| Pitfall | Problem | Fix |
|---------|---------|-----|
| `plt.show()` before saving | Shows the figure then clears it; save saves a blank | Always `fig.savefig()` BEFORE `plt.show()` |
| Not calling `plt.close()` | Memory leak with many figures in a loop | `plt.close(fig)` after saving |
| Mixing `plt.` and `ax.` | Confusing state; hard to debug | Pick one style per project |
| Not using `tight_layout()` | Labels overlap or get clipped | Add `plt.tight_layout()` before show/save |
| Default DPI (72) for print | Blurry in papers | Use `dpi=300` for publication |
| Pie charts with many slices | Impossible to read | Use bar chart for > 5 categories |
| `plt.cla()` vs `plt.clf()` vs `plt.close()` | Confusion | `cla()` clears current axes; `clf()` clears figure; `close()` closes the window |

## 12. Mini Project: Sales Performance Dashboard

**Scenario**: You're a data analyst at a retail company. You need to build a **single-page dashboard** showing:
1. Monthly revenue trends (with target comparison)
2. Revenue breakdown by product category
3. Geographic sales distribution (simulated)
4. Customer acquisition vs. churn

**Goal**: Produce a professional, publication-quality dashboard in pure Matplotlib.

````python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

np.random.seed(42)

# ── Generate synthetic business data ─────────────────────────────────────────
months     = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
revenue    = [185, 192, 205, 198, 220, 235, 285, 260, 248, 265, 290, 320]  # K$
target     = [200, 205, 210, 215, 225, 235, 250, 260, 265, 270, 280, 300]  # K$

categories = ['Electronics', 'Clothing', 'Home & Garden', 'Sports', 'Books']
cat_rev    = [420, 285, 190, 165, 95]  # Annual revenue by category
cat_growth = [12, -3, 8, 15, -8]       # YoY growth %

# Customer data (monthly)
new_customers  = np.random.randint(150, 350, 12)
churned        = np.random.randint(50, 120, 12)
net_growth     = new_customers - churned

# Simulated geographic data (region, lat, lon, sales)
regions = {
    'New York':  (-74.0, 40.7, 85),
    'LA':        (-118.2, 34.1, 72),
    'Chicago':   (-87.6, 41.9, 55),
    'Houston':   (-95.4, 29.8, 48),
    'Phoenix':   (-112.1, 33.4, 38),
    'Seattle':   (-122.3, 47.6, 44),
    'Miami':     (-80.2, 25.8, 40),
    'Denver':    (-104.9, 39.7, 31),
}

# ── Build the Dashboard ───────────────────────────────────────────────────────
fig = plt.figure(figsize=(16, 10), facecolor='#f8f9fa')
gs  = gridspec.GridSpec(2, 3, figure=fig, hspace=0.35, wspace=0.3,
                         left=0.06, right=0.97, top=0.88, bottom=0.08)

# Dashboard header
fig.suptitle('Sales Performance Dashboard  |  FY 2024', 
              fontsize=16, fontweight='bold', color='#2c3e50', y=0.95)

# ── Panel 1 (wide): Monthly Revenue vs Target ─────────────────────────────────
ax1 = fig.add_subplot(gs[0, :2])  # Spans first 2 columns
ax1.set_facecolor('white')
x_idx = range(12)
bars  = ax1.bar(x_idx, revenue, color=['#27ae60' if r >= t else '#e74c3c' 
                                         for r, t in zip(revenue, target)],
                alpha=0.8, label='Actual Revenue')
ax1.plot(x_idx, target, 'o--', color='#2c3e50', linewidth=2, markersize=5, label='Target')
ax1.fill_between(x_idx, revenue, target, 
                  where=[r < t for r, t in zip(revenue, target)],
                  alpha=0.2, color='red', label='Miss')
ax1.fill_between(x_idx, revenue, target, 
                  where=[r >= t for r, t in zip(revenue, target)],
                  alpha=0.2, color='green', label='Beat')
ax1.set_xticks(x_idx)
ax1.set_xticklabels(months, fontsize=9)
ax1.set_ylabel('Revenue ($K)', fontsize=10)
ax1.set_title('Monthly Revenue vs Target', fontsize=11, fontweight='bold')
ax1.legend(fontsize=9, loc='upper left')
ax1.grid(True, axis='y', alpha=0.3)
ax1.spines[['top', 'right']].set_visible(False)

# Add total at the top of each bar
for i, rev in enumerate(revenue):
    ax1.text(i, rev + 3, f'{rev}', ha='center', va='bottom', fontsize=7, color='#2c3e50')

# ── Panel 2 (narrow): Category Revenue ───────────────────────────────────────
ax2 = fig.add_subplot(gs[0, 2])
ax2.set_facecolor('white')
colors_cat = ['#3498db', '#2ecc71', '#e67e22', '#9b59b6', '#e74c3c']
bars2 = ax2.barh(categories, cat_rev, color=colors_cat, alpha=0.85)
ax2.bar_label(bars2, fmt='$%dK', padding=3, fontsize=9)
# Color the category labels by growth direction
for i, (label, growth) in enumerate(zip(ax2.get_yticklabels(), cat_growth)):
    label.set_color('#27ae60' if growth > 0 else '#e74c3c')
ax2.set_title('Category Revenue\n(green=growing, red=declining)', fontsize=10, fontweight='bold')
ax2.spines[['top', 'right']].set_visible(False)
ax2.set_xlim(0, max(cat_rev) * 1.25)

# ── Panel 3: Customer Net Growth ──────────────────────────────────────────────
ax3 = fig.add_subplot(gs[1, :2])
ax3.set_facecolor('white')
ax3.bar(x_idx, new_customers, color='#3498db', alpha=0.7, label='New Customers')
ax3.bar(x_idx, -churned, color='#e74c3c', alpha=0.7, label='Churned')
ax3.plot(x_idx, net_growth, 'k-o', linewidth=2, markersize=4, label='Net Growth')
ax3.axhline(0, color='black', linewidth=0.8)
ax3.set_xticks(x_idx)
ax3.set_xticklabels(months, fontsize=9)
ax3.set_title('Customer Acquisition vs Churn', fontsize=11, fontweight='bold')
ax3.set_ylabel('Customers')
ax3.legend(fontsize=9)
ax3.grid(True, axis='y', alpha=0.3)
ax3.spines[['top', 'right']].set_visible(False)

# ── Panel 4: KPI Summary ──────────────────────────────────────────────────────
ax4 = fig.add_subplot(gs[1, 2])
ax4.set_facecolor('white')
ax4.axis('off')

# Display KPIs as a text table
total_rev       = sum(revenue)
total_target    = sum(target)
vs_target_pct   = (total_rev - total_target) / total_target * 100
best_month      = months[np.argmax(revenue)]
total_new       = new_customers.sum()
total_churned   = churned.sum()
avg_net_growth  = net_growth.mean()

kpi_text = [
    ['KPI', 'Value', 'Status'],
    ['Total Revenue', f'${total_rev}K', '✓ On Track'],
    ['vs Target', f'{vs_target_pct:+.1f}%', '✓ Beat' if vs_target_pct > 0 else '✗ Miss'],
    ['Best Month', best_month, '★'],
    ['New Customers', f'{total_new:,}', '↑'],
    ['Churned', f'{total_churned:,}', '↓'],
    ['Avg Net Growth', f'+{avg_net_growth:.0f}/mo', '↑'],
]

y_pos = 0.95
for row in kpi_text:
    is_header = row[0] == 'KPI'
    weight = 'bold' if is_header else 'normal'
    color  = '#2c3e50' if is_header else '#555'
    ax4.text(0.0,  y_pos, row[0], transform=ax4.transAxes, fontsize=10, fontweight=weight, color=color)
    ax4.text(0.52, y_pos, row[1], transform=ax4.transAxes, fontsize=10, fontweight=weight, color=color)
    if len(row) > 2:
        status_color = '#27ae60' if '✓' in row[2] or '↑' in row[2] or '★' in row[2] else '#e74c3c'
        ax4.text(0.82, y_pos, row[2], transform=ax4.transAxes, fontsize=10, 
                  color=status_color if not is_header else color)
    y_pos -= 0.13
    if is_header:
        ax4.axhline(y_pos + 0.06, xmin=0, xmax=1, transform=ax4.transAxes, 
                     color='#bdc3c7', linewidth=0.8)

ax4.set_title('KPI Summary', fontsize=11, fontweight='bold')

plt.savefig('/tmp/sales_dashboard.png', dpi=120, bbox_inches='tight', facecolor=fig.get_facecolor())
print("Dashboard saved to /tmp/sales_dashboard.png")
plt.show()
````

## 13. Interview Q&A

**Q1: What is the difference between Figure and Axes in Matplotlib?**  
**A**: `Figure` is the entire canvas/window — it can contain multiple plots. `Axes` is a single plot area within the Figure — it has its own x-axis, y-axis, title, and data. One Figure can have many Axes (e.g., `plt.subplots(2,3)` creates 6 Axes inside one Figure).

---
**Q2: What is the difference between the pyplot API and the OO API?**  
**A**: The `pyplot` API (`plt.plot()`, `plt.title()`) works on the "current" figure/axes — convenient for quick single plots but hard to control multiple subplots. The OO API (`fig, ax = plt.subplots(); ax.plot()`) gives you explicit references to each Figure and Axes object — preferred for professional code, functions, and multi-panel figures.

---
**Q3: How do you add a second y-axis to a plot?**  
**A**: `ax2 = ax.twinx()` creates a twin Axes sharing the same x-axis but with an independent y-axis. Use `ax.plot()` for the left y-axis and `ax2.plot()` for the right y-axis.

---
**Q4: How do you save a figure without it appearing on screen?**  
**A**: Use `plt.ioff()` to turn off interactive mode (or use `matplotlib.use('Agg')` as the backend). Then `fig.savefig('file.png')` saves without showing. Important: call `plt.close(fig)` afterward to free memory.

---
**Q5: What does `bbox_inches='tight'` do in `savefig`?**  
**A**: It crops the figure tightly around all visible elements, removing unnecessary whitespace. Without it, axis labels and titles can sometimes be cut off at the edges.

---
**Q6: When would you use Matplotlib directly vs Seaborn?**  
**A**: Use Seaborn for statistical visualizations on DataFrames — it handles grouping, color palettes, and themes automatically. Use Matplotlib directly when you need: custom layouts (GridSpec), mixing chart types in unusual ways, precise annotation control, or when customizing a Seaborn chart that doesn't expose what you need via its API. The two work together: Seaborn uses Matplotlib under the hood.

---
**Q7: How do you change the default color cycle?**  
**A**: `plt.rcParams['axes.prop_cycle'] = plt.cycler(color=['#e41a1c','#377eb8','#4daf4a'])`. Or use `plt.style.use('seaborn-v0_8-colorblind')` to switch to a colorblind-friendly palette.
