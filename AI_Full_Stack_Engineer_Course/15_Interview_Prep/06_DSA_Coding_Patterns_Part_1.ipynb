## 🎯 How This Round Works

In a **data structures and algorithms (DSA) coding round** you get a small, precisely stated puzzle — "given an array of numbers, return…" — and 45 minutes to design an efficient algorithm, code it in a plain editor, test it, and explain its cost. The problems are rarely about ML, but AI/ML engineer candidates at most large companies go through the same coding bar as software engineers.

### What interviewers score

| Dimension | What "strong" looks like | Red flags |
|---|---|---|
| **Communication** | Restates the problem, asks about constraints, explains the plan *before* typing, narrates trade-offs | Silent coding; or talking for 15 minutes without writing anything |

**Constraints tell you the target complexity.** Python does roughly 10–50 million simple steps per second (you'll measure it below), and judges usually allow about a second:

| Input size n | Target complexity | Typical pattern |
|---|---|---|
| n ≤ 12 | O(n!) | permutations, brute-force search |
| n ≤ 25 | O(2ⁿ) | subsets, backtracking |
| n ≤ 500 | O(n³) | triple loops, some DP |
| n ≤ 5,000 | O(n²) | double loops, 2-D DP |
| n ≤ 10⁶ | O(n log n) or O(n) | sorting, hashing, two pointers, sliding window, heaps |
| n up to 10⁹ or more | O(log n) or O(1) | binary search, math |

## ⚙️ Setup

Run this cell first. The problems need only Python's standard library; NumPy and Matplotlib are used for one cross-check and two plots.

It also defines the helpers every exercise uses:
- **`check(name, fn, cases, hint, same, large)`** runs your function on every test case and prints ✅ / ⏳ / ❌. Inputs are copied before each call, so a function that modifies its input can't break the next test.
- **`CountingList`** is a read-only list that counts how many elements your code reads — the binary-search checkers use it to tell O(log n) from O(n) without relying on timing.
- **`any_order`, `any_order_nested`, `close`** are comparison helpers for "return the answer in any order" and floating-point answers.
- **`replay(cls, ops, args)`** runs a LeetCode-style sequence of method calls on a class you design (Min Stack, Time Map…).

````python
# Nothing to install: the course environment already has everything (standard library + numpy + matplotlib).
import bisect
import copy
import heapq
import math
import random
import signal
import sys
import threading
import time
from collections import Counter, defaultdict, deque

import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(f"Python {sys.version.split()[0]} | NumPy {np.__version__} | Matplotlib {matplotlib.__version__}")
random.seed(0)


class CountingList:
    """A read-only list that counts how many elements are read (index, slice, or iteration)."""

    def __init__(self, values, counter=None):
        self._values = values
        self.counter = counter if counter is not None else [0]     # a shared one-item list, so nested rows can share it

    def __len__(self):
        return len(self._values)

    def __getitem__(self, index):
        if isinstance(index, slice):
            part = self._values[index]
            self.counter[0] += len(part)
            return part
        self.counter[0] += 1
        return self._values[index]

    def __iter__(self):
        for value in self._values:
            self.counter[0] += 1
            yield value

    def __contains__(self, target):
        return any(value == target for value in self)          # reads until found, like list's `in`

    def index(self, target):
        for i, value in enumerate(self):
            if value == target:
                return i
        raise ValueError(f"{target!r} is not in list")

    def count(self, target):
        return sum(1 for value in self if value == target)

    def __repr__(self):
        return f"CountingList(n={len(self._values):,})"


class _TooSlow(BaseException):
    """Raised by the timer inside check(). A BaseException, so `except Exception` in your code can't swallow it."""


def _run_with_time_limit(fn, args, seconds):
    use_alarm = hasattr(signal, "SIGALRM") and threading.current_thread() is threading.main_thread()
    running = [True]

    def on_alarm(signum, frame):
        if running[0]:
            raise _TooSlow()

    if use_alarm:                                   # Linux/macOS: interrupt a slow solution instead of waiting forever
        previous = signal.signal(signal.SIGALRM, on_alarm)
        signal.setitimer(signal.ITIMER_REAL, seconds)
    start = time.perf_counter()
    try:
        result = fn(*args)
        running[0] = False
    finally:
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, 0)
            signal.signal(signal.SIGALRM, previous)
    elapsed = time.perf_counter() - start
    if elapsed > seconds:                           # Windows has no SIGALRM: we can only check afterwards
        raise _TooSlow()
    return result, elapsed


def _fresh(args):
    """Copy the inputs so a function that mutates them can't affect the next test (fast path for flat lists)."""
    out = []
    for arg in args:
        if isinstance(arg, list) and all(isinstance(x, (int, float, str, bool, type(None))) for x in arg):
            out.append(arg[:])
        else:
            out.append(copy.deepcopy(arg))
    return tuple(out)


def _short(value, limit=110):
    text = repr(value)
    if len(text) <= limit:
        return text
    size = f" … ({len(value):,} items)" if hasattr(value, "__len__") else " …"
    return text[:limit] + size


def _same(same, got, expected):
    try:
        return bool(same(got, expected))
    except Exception:                               # e.g. comparing None with a list
        return False


def check(name, fn, cases, hint="", same=None, large=None):
    """Run fn(*args) for every (args, expected) in cases and print ✅ / ⏳ / ❌.

    same  : optional same(got, expected) -> bool; default is got == expected.
    large : optional dict(args=..., expected=..., seconds=2.0, label="n = ...", max_reads=None).
            The large input must finish within `seconds` (an O(n²) answer is stopped). With max_reads,
            args[0] is a CountingList and your code may read at most max_reads of its elements.
    """
    same = same or (lambda got, expected: got == expected)
    for number, (args, expected) in enumerate(cases, start=1):
        try:
            got = fn(*_fresh(args))
        except NotImplementedError:
            print(f"⏳ {name}: not attempted yet — replace `raise NotImplementedError` with your code.")
            return
        except Exception as error:
            raise AssertionError(f"❌ {name}: test {number} crashed with {type(error).__name__}: {error}\n"
                                 f"   input   : {_short(args)}\n   hint    : {hint}") from None
        if not _same(same, got, expected):
            raise AssertionError(f"❌ {name}: test {number} of {len(cases)} failed\n"
                                 f"   input   : {_short(args)}\n   expected: {_short(expected)}\n"
                                 f"   got     : {_short(got)}\n   hint    : {hint}")
    summary = f"all {len(cases)} tests pass"
    if large is not None:
        seconds, label = large.get("seconds", 2.0), large.get("label", "large input")
        big_args = large["args"]
        if large.get("max_reads") is not None:
            big_args[0].counter[0] = 0
        try:
            got, elapsed = _run_with_time_limit(fn, big_args, seconds)
        except _TooSlow:
            raise AssertionError(f"❌ {name}: the small tests pass, but the large input ({label}) did not finish within "
                                 f"{seconds} s — that usually means a nested loop (O(n²)) where O(n) or O(n log n) is possible.\n"
                                 f"   hint    : {hint}") from None
        except Exception as error:
            raise AssertionError(f"❌ {name}: the large input ({label}) crashed with {type(error).__name__}: {error}\n"
                                 f"   hint    : {hint}") from None
        if not _same(same, got, large["expected"]):
            raise AssertionError(f"❌ {name}: wrong answer on the large input ({label})\n"
                                 f"   expected: {_short(large['expected'])}\n   got     : {_short(got)}\n   hint    : {hint}")
        summary += f" + large input ({label}) in {elapsed:.2f} s"
        if large.get("max_reads") is not None:
            reads = big_args[0].counter[0]
            if reads > large["max_reads"]:
                raise AssertionError(f"❌ {name}: correct, but your code read {reads:,} elements (limit {large['max_reads']}) "
                                     f"— that's a linear scan, not a binary search.\n   hint    : {hint}")
            summary += f", reading only {reads} elements"
    print(f"✅ {name}: {summary}")


def any_order(got, expected):
    """Same items, order ignored."""
    return sorted(got) == sorted(expected)


def any_order_nested(got, expected):
    """Same groups, ignoring both the order of the groups and the order inside each group."""
    return sorted(sorted(group) for group in got) == sorted(sorted(group) for group in expected)


def close(got, expected, tol=1e-6):
    """Numbers (or lists of numbers) equal up to floating-point rounding."""
    if isinstance(expected, (list, tuple)):
        return isinstance(got, (list, tuple)) and len(got) == len(expected) and all(close(g, e, tol) for g, e in zip(got, expected))
    return math.isclose(got, expected, rel_tol=tol, abs_tol=tol)


def replay(cls, ops, args):
    """LeetCode-style design test: ops[0] builds cls(*args[0]); every later op calls that method. Returns all outputs."""
    obj = cls(*args[0])
    outputs = [None]
    for op, op_args in zip(ops[1:], args[1:]):
        outputs.append(getattr(obj, op)(*op_args))
    return outputs


def best_time(fn, *args, repeat=3):
    """Fastest of several runs, in seconds (the fastest run is the least disturbed by other programs)."""
    best = float("inf")
    for _ in range(repeat):
        start = time.perf_counter()
        fn(*args)
        best = min(best, time.perf_counter() - start)
    return best


print("helpers ready: check · CountingList · any_order · any_order_nested · close · replay · best_time")
````

## 0. Big-O and Python's Toolbox 🟢

**Big-O notation** describes how the work an algorithm does *grows* as the input grows. We ignore constant factors and small terms and keep only the part that dominates for large inputs.

Think of looking for a friend's name:
- **O(1)** — you know their seat number and walk straight to it. Twice the audience, same effort.
- **O(log n)** — the guest list is alphabetical; you open it in the middle and halve the search each time. Twice the list, **one** more step.
- **O(n)** — you check every seat once. Twice the audience, twice the effort.
- **O(n log n)** — you first sort the guest list, then search it. A little more than twice the effort.
- **O(n²)** — you introduce every guest to every other guest. Twice the audience, **four times** the effort.

**Time complexity** counts steps; **space complexity** counts the *extra* memory the algorithm needs (not counting the input). Unless stated otherwise, interviewers mean the **worst case**.

The best way to believe this is to measure it: run each function on n items and on 2n items and look at the ratio of the times.

````python
def get_first(nums):              # O(1): one step, however long the list is
    return nums[0]


def total(nums):                  # O(n): touches every element once
    running = 0
    for x in nums:
        running += x
    return running


def sort_copy(nums):              # O(n log n): Python's sort
    return sorted(nums)


def count_equal_pairs(nums):      # O(n²): compares every pair
    count = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                count += 1
    return count


rng = random.Random(0)
print(f"{'function':<22}{'n':>9}{'time(n)':>12}{'time(2n)':>12}{'ratio':>8}")
for label, fn, n in [("O(1)       get_first", get_first, 400_000), ("O(n)       total", total, 400_000),
                     ("O(n log n) sort_copy", sort_copy, 400_000), ("O(n²)      pairs", count_equal_pairs, 1_000)]:
    small = [rng.randint(0, 1000) for _ in range(n)]
    big = [rng.randint(0, 1000) for _ in range(2 * n)]
    t_small, t_big = best_time(fn, small), best_time(fn, big)
    print(f"{label:<22}{n:>9,}{t_small * 1e3:>10.3f}ms{t_big * 1e3:>10.3f}ms{t_big / t_small:>8.2f}")
````

Reading the table: when n doubles, an O(n) function takes about **2×** as long, O(n log n) a bit more than 2×, and O(n²) about **4×**. The O(1) row is a few nanoseconds either way, so its ratio is just timer noise. (Exact numbers vary from run to run and machine to machine; the *ratios* are what matter.)

Now the same functions as a picture. Notice the quadratic curve shoot up while the others hug the floor — and that happens at only a few thousand elements.

````python
sizes = [250, 500, 1000, 1500, 2000]
curves = {"O(n) total": total, "O(n log n) sort_copy": sort_copy, "O(n²) count_equal_pairs": count_equal_pairs}
fig, ax = plt.subplots(figsize=(7, 4))
for label, fn in curves.items():
    times_ms = [best_time(fn, [rng.randint(0, 1000) for _ in range(n)], repeat=2) * 1e3 for n in sizes]
    ax.plot(sizes, times_ms, marker="o", label=label)
ax.set_xlabel("input size n")
ax.set_ylabel("time (ms)")
ax.set_title("Measured running time as n grows")
ax.legend()
plt.show()

steps = 3_000_000
seconds = best_time(total, list(range(steps)), repeat=2)
print(f"This laptop's Python: about {steps / seconds / 1e6:.0f} million simple loop steps per second "
      f"→ an O(n²) loop over n = 100,000 (10¹⁰ steps) would take roughly {1e10 / (steps / seconds) / 60:.0f} minutes.")
````

### Python's built-in data structures and what they cost

Before using a data structure in an interview you should be able to say what it *is* and what its operations cost:

| Structure | What it is (plain English) | Fast | Slow |
|---|---|---|---|
| `list` | A **dynamic array**: items stored side by side in memory, with spare room at the end | index `a[i]`, `append`, `pop()` from the end — O(1) | `insert(0, x)`, `pop(0)`, `x in a`, `remove(x)` — O(n) (everything shifts, or every item is checked) |
| `dict` | A **hash table**: a key is turned into a number (its *hash*) that says where to store the value | `d[k]`, `d[k] = v`, `k in d`, `del d[k]` — O(1) on average | iterating is O(n); keys must be *hashable* (immutable: str, int, tuple — not list) |
| `set` | A hash table with keys only | `add`, `x in s`, `remove` — O(1) average | no indexing, no order |
| `collections.deque` | A **double-ended queue**: fast at both ends | `append`, `appendleft`, `pop`, `popleft` — O(1) | `dq[i]` in the middle — O(n) |
| `heapq` on a list | A **binary min-heap**: a list arranged so `heap[0]` is always the smallest item | peek smallest O(1); `heappush`/`heappop` O(log n); `heapify` O(n) | finding anything except the minimum — O(n) |
| `bisect` on a sorted list | **Binary search** helpers | `bisect_left`/`bisect_right` — O(log n) | `insort` is O(n): finding the spot is fast, shifting items is not |
| `str` | An **immutable** sequence of characters | index, `len`, `in` for one char | building a string with `+=` in a loop can copy repeatedly — collect pieces in a list and `"".join(pieces)` (O(total length)) |
| `collections.Counter` | A dict that counts things | build from n items O(n); `most_common(k)` O(n log k) | — |
| `sorted(x, key=...)` | **Timsort**: stable, O(n log n) worst case, close to O(n) on nearly sorted data | the `key` function is called once per item | — |

Let's measure two of the most common interview mistakes: using a list as a queue, and using a list for membership tests.

````python
def drain_list(n):
    queue = list(range(n))
    while queue:
        queue.pop(0)                 # O(n): every remaining item shifts one slot left


def drain_deque(n):
    queue = deque(range(n))
    while queue:
        queue.popleft()              # O(1)


for n in (20_000, 40_000):
    print(f"n = {n:>6,}: list.pop(0) {best_time(drain_list, n, repeat=1) * 1e3:7.1f} ms | "
          f"deque.popleft() {best_time(drain_deque, n, repeat=1) * 1e3:5.1f} ms")

haystack_list = list(range(100_000))
haystack_set = set(haystack_list)
needles = [rng.randrange(100_000, 200_000) for _ in range(200)]        # all missing → the worst case for a list
t_list = best_time(lambda: sum(x in haystack_list for x in needles), repeat=2)
t_set = best_time(lambda: sum(x in haystack_set for x in needles), repeat=2)
print(f"200 lookups in 100,000 items: list {t_list * 1e3:.1f} ms | set {t_set * 1e3:.3f} ms → set is ~{t_list / t_set:,.0f}× faster")
````

Doubling n roughly **quadruples** the `pop(0)` time (n operations × O(n) shifting = O(n²)) while `popleft` just doubles. The membership test is the difference between O(n) per lookup and O(1).

### The toolbox you'll use in every section

### 💡 Interview angle

**Q1. What's the cost of `x in my_list`, `x in my_set`, and `my_list.pop(0)` — and what would you use for a queue?**

<details><summary>Show answer</summary>

- **30-second answer:** `x in list` is O(n) (it compares one by one), `x in set` is O(1) on average (a hash lookup), and `list.pop(0)` is O(n) because every remaining element shifts left. For a FIFO queue use `collections.deque` — `append` and `popleft` are both O(1). We measured roughly quadratic vs linear total time above.
- **Go deeper:** A `list` is a dynamic array; `append` is *amortized* O(1) because CPython over-allocates (about 12.5% extra) when it grows, so expensive resizes are rare. A `deque` is a doubly linked list of fixed-size blocks — O(1) at both ends but O(n) to index the middle. For thread-safe producer/consumer queues use `queue.Queue`.
- **❌ Common wrong answer:** "`pop` is O(1), so `pop(0)` is O(1) too."

</details>

**Q2. Is dictionary lookup really O(1)?**

<details><summary>Show answer</summary>

- **30-second answer:** On average, yes: the key's hash tells Python where to look. In the worst case — many keys colliding into the same slots — it degrades to O(n). Python randomizes string hashes per process (`PYTHONHASHSEED`) so attackers can't easily craft colliding keys.
- **Go deeper:** Hashing itself isn't free: hashing a string is O(length) the first time (CPython caches a `str` object's hash), and a tuple key rehashes all its elements on every lookup. Keys must be hashable and must not change while in the dict. If you need ordered keys or range queries, a hash map is the wrong tool — use a sorted list with `bisect` or a balanced tree.
- **❌ Common wrong answer:** "Always O(1), no exceptions" — or "O(log n), like a tree map".

</details>

**Q3. Your recursive solution passes the examples but crashes on the hidden tests. What happened, and what do you do?**

<details><summary>Show answer</summary>

- **30-second answer:** Almost certainly `RecursionError`: Python allows about 1,000 nested calls by default and has no tail-call optimization, so a recursion as deep as n fails on large n (we triggered it at depth 5,000 above). Rewrite it with a loop or an explicit stack.
- **Go deeper:** `sys.setrecursionlimit` can raise the limit, and in Python 3.12+ pure-Python recursion is much less likely to crash the interpreter, but deep recursion is still slow and memory-hungry; older versions could segfault. Recursion depth that is O(log n) (binary search, balanced trees, merge sort) is always fine.
- **❌ Common wrong answer:** "Just call `sys.setrecursionlimit(10**9)`."

</details>

## 1. Arrays & Hashing 🟢

### 🔍 Recognize it

Reach for a **hash map** (`dict`) or **hash set** (`set`) when the problem says or implies:
- "have we **seen** this value (or its partner) before?" — pairs that sum to a target, duplicates,
- "**count** how many times…", "most frequent…",
- "**group** items that share a property" (same letters, same length, same user),
- "**O(n)**" is required and the input is **not sorted** (sorting would cost O(n log n)).

The core trick: trade **memory** for **time**. A set/dict remembers what you've seen so you never re-scan the array.

### 🧰 The template

```python

    partner = target - x           # what would complete an answer with x?
    if partner in seen:            # O(1) average lookup instead of an O(n) scan
        return seen[partner], i
    seen[x] = i                    # remember x AFTER checking, so x can't pair with itself

groups = defaultdict(list)         # grouping: key → items that share it
for item in items:
    groups[key_of(item)].append(item)

counts = Counter(items)            # counting: item → how many times
```

### 🧪 Worked example — Two Sum (LeetCode 1)

**Problem:** given a list of integers and a target, return the indices of the two numbers that add up to the target. Exactly one answer exists and you can't use the same element twice.

**Step 1 — brute force.** Try every pair: O(n²) time, O(1) extra space.

````python
def two_sum_brute(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):           # j > i: never pair an element with itself
            if nums[i] + nums[j] == target:
                return [i, j]
    return []


def two_sum(nums, target):
    seen = {}                                        # value → index where we saw it
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
    return []


print(two_sum_brute([2, 7, 11, 15], 9), two_sum([2, 7, 11, 15], 9))
````

**Step 2 — find the bottleneck.** For each `x` the brute force *searches* the rest of the array for `target − x`. Searching is O(n); a hash map makes it O(1). **Step 3 — optimize:** walk once, and for each number ask "have I already seen its partner?"

Here is the one-pass version traced on a small example — watch the dictionary fill up until the partner is found:

````python
nums, target = [3, 8, 11, 2, 7], 9
seen = {}
print(f"{'i':>2} {'x':>3} {'need':>5}  {'need in seen?':<14} seen (after this step)")
for i, x in enumerate(nums):
    need = target - x
    found = need in seen
    if not found:
        seen[x] = i
    print(f"{i:>2} {x:>3} {need:>5}  {str(found):<14} {seen}")
    if found:
        print(f"→ answer: indices [{seen[need]}, {i}] because {nums[seen[need]]} + {x} = {target}")
        break

n = 3_000
worst_case = list(range(n))                       # the only answer is the last two elements
print(f"\nn = {n:,}: brute force {best_time(two_sum_brute, worst_case, 2 * n - 3, repeat=1) * 1e3:.0f} ms | "
      f"hash map {best_time(two_sum, worst_case, 2 * n - 3) * 1e3:.2f} ms")
print("duplicates [3, 3] → 6:", two_sum([3, 3], 6), "| negatives:", two_sum([-4, 1, 9, -2], -6), "| no answer:", two_sum([1, 2], 7))
````

**Complexity:** O(n) time, O(n) extra space (the dictionary).

**Edge cases to say out loud:** duplicates like `[3, 3]` (that's why we check *before* storing), negative numbers, no answer (clarify what to return), and "indices, not values".

```python
    right_product = 1
    for i in range(n - 1, -1, -1):              # multiply in everything RIGHT of i
        answer[i] *= right_product
        right_product *= nums[i]
    return answer


check("Product of Array Except Self", product_except_self, PRODUCT_TESTS, large=PRODUCT_LARGE)
```

**Complexity:** O(n) time; O(1) extra space besides the output (two running products).

**Why not divide the total product by nums[i]?** It fails with zeros (one zero: only that position is non-zero; two zeros: everything is 0), and it's explicitly banned in the interview version.

## 2. Two Pointers 🟢

### 🔍 Recognize it

- The input is **sorted** (or you're allowed to sort it) and you need a **pair / triplet** meeting a condition.
- You compare elements from **both ends** — palindromes, "container" areas, reversing.
- You must work **in place** with O(1) extra space — remove duplicates, move zeros, partition.

Two indices walk through the array, and each step **throws away** candidates that can't be part of an answer. That turns an O(n²) search over pairs into O(n).

### 🧰 The templates

```python
# Opposite ends — sorted input, pairs
left, right = 0, len(arr) - 1
while left < right:
    s = arr[left] + arr[right]
    if s == target:  ...                 # found
    elif s < target: left += 1           # need a bigger sum → move the small side up
    else:            right -= 1          # need a smaller sum → move the big side down

# Same direction — "slow" marks where the next kept element goes, "fast" scans
slow = 0
for fast in range(len(arr)):
    if keep(arr[fast]):
        arr[slow] = arr[fast]
        slow += 1
```

### 🧪 Worked example — Two Sum II: Input Array Is Sorted (LeetCode 167)

**Problem:** the list is sorted ascending; return the **1-based** indices of the two numbers that add up to the target, using O(1) extra space.

- **Brute force:** all pairs, O(n²) time.
- **Hash map (section 1):** O(n) time but O(n) space — the interviewer asked for O(1).
- **Two pointers:** O(n) time, O(1) space.

````python
def two_sum_sorted(numbers, target):
    left, right = 0, len(numbers) - 1
    while left < right:
        s = numbers[left] + numbers[right]
        if s == target:
            return [left + 1, right + 1]         # the problem wants 1-based indices
        if s < target:
            left += 1                            # numbers[left] can't reach target even with the largest partner
        else:
            right -= 1                           # numbers[right] overshoots even with the smallest partner
    return []


numbers, target = [1, 3, 4, 6, 8, 11], 10
left, right = 0, len(numbers) - 1
print(f"{'left':>4} {'right':>5} {'sum':>4}  decision")
while left < right:
    s = numbers[left] + numbers[right]
    decision = "found!" if s == target else ("too small → left += 1" if s < target else "too big → right -= 1")
    print(f"{left:>4} {right:>5} {s:>4}  {numbers[left]} + {numbers[right]}: {decision}")
    if s == target:
        break
    left, right = (left + 1, right) if s < target else (left, right - 1)
print("two_sum_sorted →", two_sum_sorted(numbers, target))
````

**Why it never misses the answer:** if `numbers[left] + numbers[right]` is too small, then `numbers[left]` paired with *any* remaining element is also too small (`numbers[right]` is the biggest one left) — so `left` can safely be discarded. The same argument discards `right` when the sum is too big. Every step removes one candidate, so there are at most n steps.

**Edge cases to say out loud:** duplicates (`[2, 2]`, target 4 works because `left < right`), negative numbers, no answer.

The same-direction template in action — remove duplicates from a sorted list **in place**:

````python
def remove_duplicates(nums):
    """Keep one copy of each value at the front of nums (in place); return how many remain."""
    if not nums:
        return 0
    slow = 1                                   # nums[:slow] holds the unique values found so far
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow - 1]:       # a new value → write it at slow
            nums[slow] = nums[fast]
            slow += 1
    return slow


data = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = remove_duplicates(data)
print(k, data[:k])
````

## 3. Sliding Window 🟡

### 🔍 Recognize it

- "**contiguous** subarray / substring" (the elements must be next to each other),
- "**longest / shortest** … such that …", "**at most k** distinct / replacements / zeros",
- "every window of size **k**", rolling averages, "in the last N minutes".

A **window** is a range `[left, right]` of the array. Instead of re-examining every range (O(n²)), you slide it: extend `right` one step at a time and shrink `left` only when the window breaks the rule. Both pointers only move forward, so the total work is O(n).

It works when the rule is **monotonic**: if a window is invalid, every larger window containing it is invalid too (e.g. "no repeated characters"). With negative numbers, "sum ≤ k" is *not* monotonic — that's section 4's job.

### 🧰 The templates

```python
# Variable-size window: longest valid window
left = 0
state = ...                               # what you need to know about nums[left:right+1]: a Counter, a sum, a set
best = 0
for right, x in enumerate(nums):
    add x to state
    while window is invalid:              # shrink until valid again
        remove nums[left] from state
        left += 1
    best = max(best, right - left + 1)

# Fixed-size window of k: add the new element, remove the one that fell out
window_sum = sum(nums[:k])
for right in range(k, len(nums)):
    window_sum += nums[right] - nums[right - k]
```

### 🧪 Worked example — Longest Substring Without Repeating Characters (LeetCode 3)

**Problem:** return the length of the longest substring with no repeated characters.

**Brute force:** for every start, extend while characters stay unique — O(n²) window checks (O(n³) if you re-check each substring from scratch).

**Sliding window:** remember the last index of each character. When `ch` repeats *inside* the current window, jump `left` just past its previous position.

````python
def longest_unique_brute(s):
    best = 0
    for start in range(len(s)):
        seen = set()
        for end in range(start, len(s)):
            if s[end] in seen:
                break
            seen.add(s[end])
        best = max(best, len(seen))
    return best


def length_of_longest_substring(s):
    last_seen = {}                                     # character → last index where it appeared
    left = best = 0
    for right, ch in enumerate(s):
        if ch in last_seen and last_seen[ch] >= left:  # repeat INSIDE the window
            left = last_seen[ch] + 1                   # jump past the old copy
        last_seen[ch] = right
        best = max(best, right - left + 1)
    return best


s = "abcabcbb"
last_seen, left, best = {}, 0, 0
print(f"{'right':>5} {'ch':>3} {'left':>5}  window      best")
for right, ch in enumerate(s):
    if ch in last_seen and last_seen[ch] >= left:
        left = last_seen[ch] + 1
    last_seen[ch] = right
    best = max(best, right - left + 1)
    print(f"{right:>5} {ch:>3} {left:>5}  {s[left:right + 1]:<10}  {best}")

for text in ["abcabcbb", "bbbbb", "pwwkew", "", "abba", " "]:
    assert length_of_longest_substring(text) == longest_unique_brute(text)
print("matches brute force on the examples; 'abba' →", length_of_longest_substring("abba"))

distinct = "".join(chr(0x4E00 + i) for i in range(3_000))        # 3,000 different characters: worst case for brute force
print(f"n = 3,000 distinct characters: brute force {best_time(longest_unique_brute, distinct, repeat=1) * 1e3:.0f} ms | "
      f"sliding window {best_time(length_of_longest_substring, distinct) * 1e3:.2f} ms")
````

**The classic bug:** forgetting `last_seen[ch] >= left`. On `"abba"`, when the second `a` arrives, the old `a` (index 0) is already *outside* the window `"b"`, so `left` must not move backwards. Without the check the answer comes out as 3 instead of 2.

**Complexity:** O(n) time; O(k) space for k distinct characters.

```python
def min_window(s, t):
    if not t:
        return ""
    need = Counter(t)                    # how many more of each character the window still needs (can go negative = surplus)
    missing = len(t)                     # total characters still needed
    best_start, best_len = 0, float("inf")
    left = 0
    for right, ch in enumerate(s):
        if need[ch] > 0:
            missing -= 1
        need[ch] -= 1
        while missing == 0:              # valid window → try to shrink it
            if right - left + 1 < best_len:
                best_start, best_len = left, right - left + 1
            need[s[left]] += 1
            if need[s[left]] > 0:        # we just gave away a character we actually needed
                missing += 1
            left += 1
    return "" if best_len == float("inf") else s[best_start:best_start + best_len]


check("Minimum Window Substring", min_window, MIN_WINDOW_TESTS, large=MIN_WINDOW_LARGE)
```

**Complexity:** O(|s| + |t|) time — each index enters and leaves the window once — and O(k) space for k distinct characters.

### 💡 Interview angle

**Q9. When does a sliding window NOT work, and what do you use instead?**

<details><summary>Show answer</summary>

- **30-second answer:** It needs a monotonic rule: shrinking an invalid window must eventually make it valid, and growing a valid window must not un-break an invalid one. With **negative numbers**, "subarray sum equals / at most k" breaks that (adding a negative makes the sum smaller), so use **prefix sums + a hash map** (section 4). Non-contiguous selections ("subsequence") aren't windows at all — that's usually DP.
- **Go deeper:** "Exactly K distinct" isn't monotonic, but `exactly(K) = atMost(K) − atMost(K − 1)`, and each "at most" *is* a valid window. Shortest subarray with sum ≥ k **with negatives** needs prefix sums plus a monotonic deque.
- **❌ Common wrong answer:** "Sliding window works for any subarray problem."

</details>

**Q10. Streaming follow-up: how do you show the average and p99 latency over the last 5 minutes of a live service?**

<details><summary>Show answer</summary>

Percentiles aren't decomposable like sums, so either keep the window in a sorted structure (O(log n) per update) or use an approximate **sketch/histogram** (HDR histogram, t-digest, DDSketch, or Prometheus histogram buckets) that merges cheaply across time buckets and servers.

## 4. Prefix Sums 🟡

### 🔍 Recognize it

- **Many** "sum of elements from i to j" questions on an array that doesn't change,
- "number of subarrays whose sum equals k" — especially with **negative numbers** (where a sliding window fails),
- sums over **rectangles** of a grid (an image, a heat map),
- "running total", "balance", "cumulative count".

A **prefix sum** array stores the running total: `prefix[i]` = sum of the first `i` elements. Then any range sum is a subtraction:

```
nums   =      [ 3,  1,  4,  1,  5]
prefix =  [0,   3,  4,  8,  9, 14]          prefix[i] = nums[0] + … + nums[i-1]
sum(nums[1..3]) = prefix[4] - prefix[1] = 9 - 3 = 6  (1 + 4 + 1)
```

### 🧰 The template

```python
prefix = [0]
for x in nums:
    prefix.append(prefix[-1] + x)            # O(n) once
range_sum = prefix[right + 1] - prefix[left] # O(1) per query

# "how many subarrays sum to k?" — count earlier prefixes equal to running - k
seen = Counter({0: 1})                       # the empty prefix
running = count = 0
for x in nums:
    running += x
    count += seen[running - k]
    seen[running] += 1
```

### 🧪 Worked example — Range Sum Query - Immutable (LeetCode 303)

**Problem:** build an object from an array, then answer many `sum_range(left, right)` queries (inclusive indices).

- **Brute force:** sum the slice every time — O(n) per query, O(q · n) for q queries.
- **Prefix sums:** O(n) preprocessing, then O(1) per query.

````python
class NumArrayBrute:
    def __init__(self, nums):
        self.nums = nums

    def sum_range(self, left, right):
        return sum(self.nums[left:right + 1])                  # O(n) per query


class NumArray:
    def __init__(self, nums):
        self.prefix = [0]
        for x in nums:
            self.prefix.append(self.prefix[-1] + x)            # prefix[i] = sum of the first i numbers

    def sum_range(self, left, right):
        return self.prefix[right + 1] - self.prefix[left]      # O(1)


nums = [-2, 0, 3, -5, 2, -1]
table = NumArray(nums)
print("nums  :", nums)
print("prefix:", table.prefix)
for left, right in [(0, 2), (2, 5), (0, 5)]:
    print(f"sum_range({left}, {right}) = prefix[{right + 1}] - prefix[{left}] = {table.prefix[right + 1]} - ({table.prefix[left]}) "
          f"= {table.sum_range(left, right)}  (brute force: {NumArrayBrute(nums).sum_range(left, right)})")

big = [rng.randint(-100, 100) for _ in range(100_000)]
queries = [sorted((rng.randrange(100_000), rng.randrange(100_000))) for _ in range(2_000)]
brute, fast = NumArrayBrute(big), NumArray(big)
assert all(brute.sum_range(l, r) == fast.sum_range(l, r) for l, r in queries[:50])
t_brute = best_time(lambda: [brute.sum_range(l, r) for l, r in queries], repeat=1)
t_fast = best_time(lambda: [fast.sum_range(l, r) for l, r in queries])
print(f"\n2,000 queries on 100,000 numbers: brute force {t_brute * 1e3:.0f} ms | prefix sums {t_fast * 1e3:.2f} ms")
````

**Why a sliding window can't count "subarrays summing to k" when there are negatives:** in `[1, -1, 5, -2, 3]` with k = 3, both `[1, -1, 5, -2]` and `[5, -2]` sum to 3. But moving `left` forward can make the sum *bigger* (dropping a −1) or *smaller* (dropping a 1), so there's no rule telling you which pointer to move. Prefix sums sidestep the question: a subarray `nums[i..j]` sums to k exactly when `prefix[j + 1] − prefix[i] = k`, i.e. when an **earlier prefix equals `running − k`** — a hash-map lookup.

````python
nums, k = [1, -1, 5, -2, 3], 3
running, seen, count = 0, Counter({0: 1}), 0
print(f"{'x':>3} {'running':>8} {'need running-k':>15} {'times seen':>11}  count")
for x in nums:
    running += x
    matches = seen[running - k]
    count += matches
    seen[running] += 1
    print(f"{x:>3} {running:>8} {running - k:>15} {matches:>11}  {count}")
brute_count = sum(1 for i in range(len(nums)) for j in range(i, len(nums)) if sum(nums[i:j + 1]) == k)
print("subarrays summing to 3:", count, "| brute force agrees:", brute_count == count)
````

**Q11. Walk me through why Subarray Sum Equals K needs `{0: 1}` and a hash map.**

<details><summary>Show answer</summary>

- **30-second answer:** A subarray ending at j sums to k exactly when some earlier prefix sum equals `running − k`, so I count how many earlier prefixes had that value — a hash map makes the lookup O(1) and the whole thing O(n). The `{0: 1}` entry is the empty prefix: it lets subarrays that start at index 0 be counted.
- **Go deeper:** Variants use the same idea with a different map value: *longest* subarray with sum k → store the **first** index of each prefix sum; subarray sum **divisible by k** → store `prefix % k` (Python's `%` is always non-negative, unlike C/Java); equal numbers of 0s and 1s → map 0 to −1 and look for sum 0.
- **❌ Common wrong answer:** "Use a sliding window: shrink when the sum is bigger than k" — wrong as soon as negatives appear.

</details>

**Q12. What if the array also receives updates between range-sum queries?**

<details><summary>Show answer</summary>

- **30-second answer:** A prefix array needs O(n) work per update, so with many updates use a **Fenwick tree** (binary indexed tree) — O(log n) per update and per prefix query, O(n) memory — or a **segment tree**, which also supports range min/max and range updates with lazy propagation.
- **Go deeper:** If updates are rare and queries frequent, rebuilding the prefix array lazily (only after a batch of updates) is simpler. Square-root decomposition gives O(√n) for both and is easy to write under interview pressure.
- **❌ Common wrong answer:** "Just recompute the prefix array after every update" without noting it makes each update O(n).

</details>

## 5. Stacks & Monotonic Stacks 🟡

### 🔍 Recognize it

A **stack** is last-in, first-out (LIFO) — like a pile of plates. In Python, a `list` is a perfect stack: `append` pushes, `pop()` pops, `stack[-1]` peeks, all O(1).

- **Plain stack:** matching brackets / tags, nested structures, undo history, evaluating expressions, simulating recursion.
- **Monotonic stack** — a stack kept in increasing (or decreasing) order: "**next greater** / next smaller element", "how many days until a warmer day", "how far can this bar extend", "largest rectangle".

### 🧰 The template

```python
# Next greater element to the right, for every index
answer = [-1] * len(nums)
stack = []                                   # indices still waiting for their answer; their values are decreasing
for i, x in enumerate(nums):
    while stack and nums[stack[-1]] < x:     # x is the next greater element for everything smaller on top
        j = stack.pop()
        answer[j] = i                        # (or x, or i - j — whatever the problem asks)
    stack.append(i)
```

It looks nested, but each index is pushed once and popped at most once, so the total work is **O(n)** (an *amortized* argument).

### 🧪 Worked example — Daily Temperatures (LeetCode 739)

**Problem:** for each day, how many days until a strictly warmer temperature? 0 if none.

````python
def daily_temperatures_brute(temps):
    answer = [0] * len(temps)
    for i in range(len(temps)):
        for j in range(i + 1, len(temps)):          # scan forward until warmer: O(n²) when temps keep falling
            if temps[j] > temps[i]:
                answer[i] = j - i
                break
    return answer


def daily_temperatures(temps):
    answer = [0] * len(temps)
    stack = []                                       # indices of days still waiting for a warmer day
    for i, t in enumerate(temps):
        while stack and temps[stack[-1]] < t:        # today is the answer for every colder day on top
            j = stack.pop()
            answer[j] = i - j
        stack.append(i)
    return answer


temps = [73, 74, 75, 71, 69, 72, 76, 73]
answer, stack = [0] * len(temps), []
print(f"{'day':>3} {'temp':>4}  resolved            stack after (as temperatures)")
for i, t in enumerate(temps):
    resolved = []
    while stack and temps[stack[-1]] < t:
        j = stack.pop()
        answer[j] = i - j
        resolved.append(f"day{j}→{i - j}")
    stack.append(i)
    print(f"{i:>3} {t:>4}  {', '.join(resolved) or '-':<18}  {[temps[j] for j in stack]}")
print("answer:", answer, "| brute force agrees:", answer == daily_temperatures_brute(temps))

falling = list(range(4_000, 0, -1))                  # temperatures that only fall: the brute force's worst case
print(f"n = 4,000 falling temps: brute force {best_time(daily_temperatures_brute, falling, repeat=1) * 1e3:.0f} ms | "
      f"monotonic stack {best_time(daily_temperatures, falling) * 1e3:.2f} ms")
````

Notice the stack's temperatures are always **decreasing** from bottom to top: a warmer day clears every colder day off the top.

> 🤖 **AI/ML angle:** the same code answers "for each epoch, how many epochs until validation loss improved?" (next *smaller* element — flip `<` to `>`), which is what an early-stopping patience setting is really tuning.

````python
val_loss = [0.92, 0.85, 0.87, 0.86, 0.80, 0.81, 0.79, 0.83]       # a small made-up loss curve
wait = [0] * len(val_loss)
stack = []
for epoch, loss in enumerate(val_loss):
    while stack and val_loss[stack[-1]] > loss:        # this epoch beats the losses waiting on the stack
        earlier = stack.pop()
        wait[earlier] = epoch - earlier
    stack.append(epoch)
print("epochs until a lower validation loss (0 = never):", wait)
print("longest wait:", max(wait), "epochs → a patience below that would have stopped too early")
````

```python
def eval_rpn(tokens):
    stack = []
    for token in tokens:
        if token in {"+", "-", "*", "/"}:
            right = stack.pop()                      # popped first = written last = the right operand
            left = stack.pop()
            if token == "+":
                stack.append(left + right)
            elif token == "-":
                stack.append(left - right)
            elif token == "*":
                stack.append(left * right)
            else:
                quotient = abs(left) // abs(right)   # integer-only truncation toward zero (no float rounding)
                stack.append(quotient if (left < 0) == (right < 0) else -quotient)
        else:
            stack.append(int(token))                 # int("-11") handles negative numbers
    return stack[-1]
```

```python
def largest_rectangle_area(heights):
    stack = []                                        # (start index, height), heights strictly increasing
    best = 0
    for i, h in enumerate(heights):
        start = i
        while stack and stack[-1][1] >= h:            # taller (or equal) bars can't extend past i
            index, height = stack.pop()
            best = max(best, height * (i - index))
            start = index                             # the new, shorter bar can extend back to here
        stack.append((start, h))
    for index, height in stack:                       # these bars extend all the way to the right edge
        best = max(best, height * (len(heights) - index))
    return best
```

### 💡 Interview angle

**Q13. How do you recognize a monotonic-stack problem, and why is it O(n) with a loop inside a loop?**

<details><summary>Show answer</summary>

- **30-second answer:** The cue is "for each element, find the **nearest** element to the left/right that is greater/smaller" (next warmer day, stock span, how far a bar extends). The inner `while` looks quadratic, but every index is pushed exactly once and popped at most once, so across the whole run there are at most n pops — O(n) total (amortized analysis).
- **Go deeper:** Choose the order by what you pop for: *next greater* → keep a decreasing stack; *next smaller* → an increasing stack. Decide how to treat equal values (`<` vs `<=`) and say it out loud — it changes answers with duplicates. For "previous greater", read the stack top *after* popping.
- **❌ Common wrong answer:** "It's O(n²) in the worst case because of the nested loop."

</details>

**Q14. Min Stack follow-up: can you reduce the extra memory?**

<details><summary>Show answer</summary>

- **30-second answer:** Keep a second stack that pushes a value only when it's `<=` the current minimum and pops it when that value is popped — in typical data it's far smaller than n. Or store a single `min` variable plus encoded values (push `value − min`; a negative entry means the minimum changed, and the old minimum can be recovered) for O(1) extra space.
- **Go deeper:** The encoding trick can overflow in fixed-width languages (not in Python). A "min queue" (for sliding windows) is built from two min-stacks or a monotonic deque — see Timed Drill B2.
- **❌ Common wrong answer:** "Keep one `min` variable" — it can't tell you the new minimum after the minimum is popped.

</details>

## 6. Binary Search 🟡

### 🔍 Recognize it

- The input is **sorted** (or sorted-then-rotated), and you need **O(log n)**.
- "Find the **first / last** position where…", "insert position", "count of values below x".
- **Binary search on the answer:** "the **minimum** speed / capacity / time such that it's possible…", "**minimize the maximum**…" — whenever *feasible(x)* flips from False to True exactly once as x grows.
- A sorted log keyed by **timestamps** ("value at time t").

Binary search keeps a range that must contain the answer and halves it each step: 1,000,000 items need only about 20 steps.

### 🧰 The templates

```python
# 1. Exact match in a sorted list

while lo <= hi:
    mid = (lo + hi) // 2
    if nums[mid] == target: return mid
    if nums[mid] < target:  lo = mid + 1      # answer is right of mid
    else:                   hi = mid - 1      # answer is left of mid

# 2. First value where is_ok flips to True (False False … True True) — use this for everything else
def first_true(lo, hi, is_ok):                 # answer is in [lo, hi]
    while lo < hi:
        mid = (lo + hi) // 2
        if is_ok(mid): hi = mid                # mid might be the answer — keep it
        else:          lo = mid + 1            # mid is definitely not
    return lo

# Built in: bisect.bisect_left(a, x) = first index with a[i] >= x; bisect_right = first index with a[i] > x
```

````python
            lo = mid + 1
````

```python
    return nums[lo]
```

**Why compare with `nums[hi]` and not `nums[lo]`?** For an unrotated array, `nums[mid] > nums[lo]` is true but the minimum is on the left — comparing with `lo` needs an extra special case.

### 💡 Interview angle

**Q15. What are the classic binary search bugs, and how do you avoid them?**

<details><summary>Show answer</summary>

- **30-second answer:** (1) Infinite loops — with `while lo < hi`, setting `lo = mid` when `mid` rounds down never makes progress; use `lo = mid + 1` / `hi = mid`. (2) Off-by-one — mixing `lo <= hi` with `hi = mid`. (3) Wrong boundary with duplicates — "find target" isn't "find the *first* target". Fix all three by always using one template, stating the invariant ("the answer is in `[lo, hi]`"), and testing arrays of length 0, 1 and 2.
- **Go deeper:** In Java/C++, `(lo + hi) / 2` can overflow; write `lo + (hi - lo) / 2`. Python integers don't overflow, but interviewers like hearing you know it. For floating-point answers, loop a fixed number of times (e.g. 100) instead of comparing floats for equality.
- **❌ Common wrong answer:** "Binary search only works on sorted arrays" — it works on any monotonic True/False predicate.

</details>

**Q16. How do you spot "binary search on the answer", and where does it show up in ML engineering?**

<details><summary>Show answer</summary>

- **30-second answer:** The problem asks for the minimum (or maximum) value of something, and you can write a cheap `feasible(x)` check that is monotonic — once True, it stays True. Then binary search x over its possible range: O(log(range) × cost of the check). Examples: ship packages within D days, split an array to minimize the largest sum, Koko's speed.
- **Go deeper:** In ML work: the largest batch size that fits in GPU memory (if b fits, anything smaller fits), the smallest number of replicas that meets a latency SLO under a load test, the lowest decision threshold that still reaches a target precision on a validation set (precision isn't always monotonic in the threshold, so check that assumption), and `git bisect` to find the commit that caused a metric regression.
- **❌ Common wrong answer:** "Try every speed from 1 upwards" — O(max value × n), which is 10⁹ × n here.

</details>

## 7. Intervals & Sorting 🟡

### 🔍 Recognize it

- Inputs are **[start, end] pairs**: meetings, bookings, GPU jobs, ad slots, time ranges in logs.
- "**merge** overlapping", "**insert** into a schedule", "can one person attend all", "remove the **fewest** to avoid overlaps", "**how many rooms / machines** at once".

Almost every interval problem starts by **sorting** — by start time (merging) or by end time (keeping the most intervals) — so the answer is usually O(n log n).

**Two intervals `[a, b]` and `[c, d]` overlap** exactly when `a <= d and c <= b`. Always clarify whether *touching* intervals (`[1, 4]` and `[4, 5]`) count as overlapping.

### 🧰 The templates

### 🧪 Worked example — Merge Intervals (LeetCode 56)

**Problem:** merge all overlapping intervals (touching ones count as overlapping) and return the result sorted.

**Brute force:** repeatedly find any two overlapping intervals and merge them until no pair overlaps — O(n²) pair checks per merge, O(n³) worst case. **Sorting** puts candidates next to each other: after sorting by start, an interval can only overlap the *last* merged one.

````python
def merge_brute(intervals):
    intervals = [interval[:] for interval in intervals]
    changed = True
    while changed:
        changed = False
        for i in range(len(intervals)):
            for j in range(i + 1, len(intervals)):
                (a, b), (c, d) = intervals[i], intervals[j]
                if a <= d and c <= b:                          # overlap (touching counts)
                    intervals[i] = [min(a, c), max(b, d)]
                    intervals.pop(j)
                    changed = True
                    break
            if changed:
                break
    return sorted(intervals)


def merge(intervals):
    merged = []
    for start, end in sorted(intervals):                        # sort by start (then end)
        if merged and start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)             # extend: the new interval may end earlier (nested)
        else:
            merged.append([start, end])
    return merged


example = [[8, 10], [1, 3], [15, 18], [2, 6], [17, 20], [9, 12]]
merged_example = []
print(f"{'interval':>10}  action               merged so far")
for start, end in sorted(example):
    if merged_example and start <= merged_example[-1][1]:
        merged_example[-1][1] = max(merged_example[-1][1], end)
        action = "overlaps → extend"
    else:
        merged_example.append([start, end])
        action = "gap → new interval"
    print(f"{str([start, end]):>10}  {action:<20} {merged_example}")
assert merge(example) == merge_brute(example)
print("merge:", merge(example), "| nested [[1, 10], [2, 3]] →", merge([[1, 10], [2, 3]]), "| touching [[1, 4], [4, 5]] →", merge([[1, 4], [4, 5]]))

fig, ax = plt.subplots(figsize=(7, 3))
for row, (start, end) in enumerate(sorted(example)):
    ax.plot([start, end], [row, row], linewidth=6, solid_capstyle="butt", color="tab:blue", label="input" if row == 0 else None)
for k, (start, end) in enumerate(merge(example)):
    ax.plot([start, end], [-1.5, -1.5], linewidth=10, solid_capstyle="butt", color="tab:orange", label="merged" if k == 0 else None)
ax.set_yticks([])
ax.set_xlabel("time")
ax.set_title("Merge Intervals: sorted input (blue) and merged result (orange)")
ax.legend(loc="upper left")
plt.show()
````

**Complexity:** O(n log n) time for the sort; O(n) space for the output.

**Edge cases to say out loud:** empty input, one interval, nested intervals (`[1, 10]` then `[2, 3]` — that's why we take `max` of the ends), touching intervals, unsorted input.

### 💡 Interview angle

**Q17. Do touching intervals overlap? And why do some interval problems sort by start and others by end?**

<details><summary>Show answer</summary>

- **30-second answer:** It depends on the problem, so ask. Closed intervals `[1, 4]` and `[4, 5]` share the point 4 (Merge Intervals merges them); half-open ones `[1, 4)` and `[4, 5)` don't (a meeting ending at 4 frees the room). Sort by **start** when you need to merge or sweep through time; sort by **end** when you want to *keep as many intervals as possible* (activity selection), because the earliest finisher leaves the most room.
- **Go deeper:** Tie-breaking matters: in a sweep line with half-open intervals, process end events before start events at the same time. When sorting by start for merging, nested intervals are handled by taking `max` of the ends, not the latest interval's end.
- **❌ Common wrong answer:** "Sort by start and greedily keep the first interval" for Non-overlapping Intervals — on `[[1, 100], [2, 3], [4, 5]]` it keeps `[1, 100]` and removes two intervals instead of one.

</details>

**Q18. Meeting Rooms II follow-ups: which GPU does each job get, and what if jobs arrive live?**

<details><summary>Show answer</summary>

- **30-second answer:** Keep `(end_time, gpu_id)` pairs in the busy heap and a second min-heap of free GPU ids; when a job starts, move every finished GPU from the busy heap to the free heap, then take the smallest free id (or add a new GPU). For a live stream in start order the same heap works online: each arrival is O(log n).
- **Go deeper:** Real schedulers add priorities, preemption, GPU memory as a *resource amount* rather than a count, and fairness across teams (Kubernetes with Kueue/Volcano, Slurm). If times are small integers you can use a difference array over time instead of sorting: O(n + T).
- **❌ Common wrong answer:** "The answer is the number of overlapping pairs."

</details>

````python
def _grid(rows, changes=()):
    """Build a 9 × 9 board of single-character strings, optionally changing some cells: (row, col, value)."""
    board = [list(row) for row in rows]
    for r, c, value in changes:
        board[r][c] = value
    return board


_EXAMPLE = ["53..7....", "6..195...", ".98....6.", "8...6...3", "4..8.3..1", "7...2...6", ".6....28.", "...419..5", "....8..79"]
_SOLVED = ["534678912", "672195348", "198342567", "859761423", "426853791", "713924856", "961537284", "287419635", "345286179"]
_EMPTY = ["........."] * 9
SUDOKU_TESTS = [
    ((_grid(_EXAMPLE),), True),
    ((_grid(_EXAMPLE, [(0, 0, "8")]),), False),                       # 8 twice in column 0 and in the top-left box
    ((_grid(_EMPTY),), True),                                         # empty board
    ((_grid(_EMPTY, [(4, 0, "1"), (4, 8, "1")]),), False),            # row duplicate
    ((_grid(_EMPTY, [(0, 2, "2"), (8, 2, "2")]),), False),            # column duplicate
    ((_grid(_EMPTY, [(0, 0, "3"), (1, 1, "3")]),), False),            # box duplicate (different row and column)
    ((_grid(_SOLVED),), True),                                        # a completely filled, valid board
    ((_grid(_EMPTY, [(0, 0, "9"), (4, 4, "9"), (8, 8, "9")]),), True),# same digit in different rows, columns and boxes
    ((_grid(_EMPTY, [(6, 6, "4"), (8, 8, "4")]),), False),            # duplicate in the bottom-right box
]
````

```python
def is_valid_sudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = defaultdict(set)                          # (r // 3, c // 3) → digits in that box
    for r in range(9):
        for c in range(9):
            digit = board[r][c]
            if digit == ".":

            box = (r // 3, c // 3)
            if digit in rows[r] or digit in cols[c] or digit in boxes[box]:

            rows[r].add(digit)
            cols[c].add(digit)
            boxes[box].add(digit)
    return True
```

<details><summary>💡 Show solution</summary>

```python
def trap(height):
    left, right = 0, len(height) - 1
    left_max = right_max = water = 0
    while left < right:
        if height[left] < height[right]:              # the right side has a wall at least this tall → left_max limits
            left_max = max(left_max, height[left])
            water += left_max - height[left]
            left += 1
        else:                                         # symmetric: left side is at least as tall → right_max limits
            right_max = max(right_max, height[right])
            water += right_max - height[right]
            right -= 1
    return water


check("A2 Trapping Rain Water", trap, TRAP_TESTS, large=TRAP_LARGE)
```

**Complexity:** O(n) time, O(1) space. A good way to get there: first write the O(n)-space version with `left_max[i]` and `right_max[i]` arrays (prefix maxima), then notice you only ever need the smaller of the two running maxima.
</details>

**Q19. Trapping Rain Water follow-ups: other approaches, and the 2-D version?**

<details><summary>Show answer</summary>

- **30-second answer:** Three O(n) approaches: prefix/suffix max arrays (O(n) space), two pointers (O(1) space), or a monotonic stack that fills water layer by layer when a taller bar closes a basin. For a 2-D height map (LeetCode 407), start from all border cells in a **min-heap** and flood inward: always expand the lowest wall, trapping `max(0, wall − height)` at each neighbour — O(mn log(mn)).
- **Go deeper:** The two-pointer proof: if `height[left] < height[right]`, the water at `left` is decided by `left_max` alone, because some bar at least `height[right]` tall exists to its right. The heap approach is Dijkstra-like (Part 2).
- **❌ Common wrong answer:** "Water at i is `min(height[i-1], height[i+1]) − height[i]`" — the limiting walls can be far away, not just the neighbours.

</details>

<details><summary>💡 Show solution</summary>

```python
def check_inclusion(s1, s2):
    k = len(s1)
    if k > len(s2):
        return False
    need, window = [0] * 26, [0] * 26
    for ch in s1:
        need[ord(ch) - ord("a")] += 1
    for right, ch in enumerate(s2):
        window[ord(ch) - ord("a")] += 1
        if right >= k:
            window[ord(s2[right - k]) - ord("a")] -= 1        # this letter slid out of the window
        if right >= k - 1 and window == need:                  # comparing two 26-item lists is O(26)
            return True
    return k == 0                                              # only an empty s1 matches an empty s2
```

<details><summary>💡 Show solution</summary>

```python
def max_sliding_window(nums, k):
    window = deque()                                     # indices; their values decrease from front to back
    result = []
    for right, x in enumerate(nums):
        while window and nums[window[-1]] <= x:          # smaller values can never be a maximum again
            window.pop()
        window.append(right)
        if window[0] <= right - k:                       # the front index slid out of the window
            window.popleft()
        if right >= k - 1:
            result.append(nums[window[0]])
    return result


check("B2 Sliding Window Maximum", max_sliding_window, WINDOW_MAX_TESTS, large=WINDOW_MAX_LARGE)
```

**Complexity:** O(n) time — every index enters and leaves the deque at most once — and O(k) space. A heap of `(-value, index)` with lazy removal of out-of-window indices also works in O(n log n) and is a reasonable first answer.
</details>

**Q20. Why a deque for Sliding Window Maximum, and where does this pattern show up in ML systems?**

<details><summary>Show answer</summary>

- **30-second answer:** We need to add at the back, remove expired indices from the front, and discard dominated values from the back — all O(1) with a deque. Keeping values decreasing means the front is always the maximum. Each index is pushed and popped once, so it's O(n) overall versus O(n · k) for recomputing `max` per window.
- **Go deeper:** It's 1-D max pooling with stride 1, "max latency over the last k requests" on a dashboard, and the building block of "shortest subarray with sum ≥ k" (a monotonic deque over prefix sums). For a *time*-based window, store `(timestamp, value)` and expire by time.
- **❌ Common wrong answer:** "Keep one `max` variable and update it as the window slides" — when the maximum leaves the window you must rescan, which is O(k).

</details>

```python
def search_matrix(matrix, target):
    rows = len(matrix)
    if rows == 0:
        return False
    cols = len(matrix[0])
    lo, hi = 0, rows * cols - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        value = matrix[mid // cols][mid % cols]      # position mid in the flattened, sorted order
        if value == target:
            return True
        if value < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return False
```

````python
MIN_INTERVAL_TESTS = [
    (([[1, 4], [2, 4], [3, 6], [4, 4]], [2, 3, 4, 5]), [3, 3, 1, 4]),
    (([[2, 3], [2, 5], [1, 8], [20, 25]], [2, 19, 5, 22]), [2, -1, 4, 6]),
    (([], [1]), [-1]),                                          # no intervals
    (([[1, 1]], [1, 2]), [1, -1]),                              # size-1 interval
    (([[5, 10]], []), []),                                      # no queries
    (([[1, 10], [2, 3]], [2, 4, 11]), [2, 10, -1]),
    (([[1, 5], [1, 5]], [3]), [5]),                             # duplicate intervals
    (([[-3, -1], [0, 0]], [-2, 0, -4]), [3, 1, -1]),            # negatives
]
_iv_rng = random.Random(10)
_queries = [_iv_rng.randrange(-1_000, 201_000) for _ in range(100_000)]
MIN_INTERVAL_LARGE = dict(args=([[i, i + 9] for i in range(0, 200_000, 10)], _queries),
                          expected=[10 if 0 <= q < 200_000 else -1 for q in _queries],
                          seconds=3.0, label="20,000 intervals, 100,000 queries")
````

````python
def min_interval(intervals, queries):
    """intervals[i] = [left, right] (inclusive), size = right - left + 1. For each query q, return the size of the
    smallest interval with left <= q <= right, or -1 if no interval contains q. Answers are in query order.

    Example: ([[1, 4], [2, 4], [3, 6], [4, 4]], [2, 3, 4, 5]) → [3, 3, 1, 4]
    Constraints: 0 ≤ len(intervals), len(queries) ≤ 10⁵. About O((n + q) log n) time.
    """
    raise NotImplementedError  # TODO: your code here


check("C2 Minimum Interval to Include Each Query", min_interval, MIN_INTERVAL_TESTS, large=MIN_INTERVAL_LARGE,
      hint="Answer queries in sorted order (offline). Sort intervals by left; push (size, right) for every interval with left <= q; "
           "pop while heap top's right < q; the top's size is the answer. Map answers back to the original order.")
````

<details><summary>💡 Show solution</summary>

```python
def min_interval(intervals, queries):
    intervals = sorted(intervals)
    heap = []                                        # (size, right end) of intervals that start at or before q
    answer = {}
    i = 0
    for q in sorted(set(queries)):                   # offline: process queries from smallest to largest
        while i < len(intervals) and intervals[i][0] <= q:
            left, right = intervals[i]
            heapq.heappush(heap, (right - left + 1, right))
            i += 1
        while heap and heap[0][1] < q:               # ends before q → useless for q and every later query
            heapq.heappop(heap)
        answer[q] = heap[0][0] if heap else -1
    return [answer[q] for q in queries]


check("C2 Minimum Interval to Include Each Query", min_interval, MIN_INTERVAL_TESTS, large=MIN_INTERVAL_LARGE)
```

**Complexity:** O(n log n + q log q) time, O(n + q) space. Smaller intervals that ended can be popped lazily — only the heap top needs to be valid.
</details>

**Q21. "Offline" vs "online" queries — why were we allowed to sort the queries, and what if we couldn't?**

<details><summary>Show answer</summary>

- **30-second answer:** All queries were known upfront (**offline**), so we could answer them in whatever order is convenient and restore the original order at the end. If queries arrive one at a time and must be answered immediately (**online**), you need a data structure built in advance — e.g. a segment tree over coordinates storing the minimum interval size covering each point, or an interval tree — with O(log n) per query.
- **Go deeper:** Sorting queries together with events is a general trick (Mo's algorithm, sweep lines, "process events in time order"). In production it's the difference between a batch job and a low-latency API; say which one the interviewer's scenario is.
- **❌ Common wrong answer:** "Sorting the queries changes the answer order, so it's not allowed" — you just keep the mapping.

</details>

## 📋 Pattern Cheat Sheet

### Pattern → cues → template → complexity

| Pattern | Cues in the problem statement | Core move | Typical time / extra space |
|---|---|---|---|
| **Hash map / set** | "seen before", count, group, duplicates, pairs in unsorted data | `seen = {}`; look up `target - x` before storing `x` | O(n) / O(n) |
| **Two pointers (opposite ends)** | sorted input, pairs/triplets with a target, palindrome, max area | move the pointer that can't be part of any answer | O(n) (+ O(n log n) to sort) / O(1) |
| **Two pointers (same direction)** | in place, remove/partition, fast & slow | `slow` marks where the next kept item goes | O(n) / O(1) |
| **Sliding window (variable)** | contiguous, longest/shortest, "at most k" | extend `right`; `while invalid: shrink left` | O(n) / O(alphabet or k) |
| **Sliding window (fixed)** | "every window of size k", rolling metrics | add the new element, subtract the one k back | O(n) / O(1) |
| **Prefix sums** | many range sums, subarray sum = k with negatives, rectangle sums | `prefix[r+1] - prefix[l]`; `count += seen[running - k]` | O(n) build, O(1) query / O(n) |
| **Stack** | brackets, nesting, undo, evaluate expressions | push openers/operands, pop on closers/operators | O(n) / O(n) |
| **Monotonic stack** | next/previous greater or smaller, spans, histogram areas | `while stack and nums[stack[-1]] < x: pop and resolve` | O(n) amortized / O(n) |
| **Monotonic deque** | max/min of every sliding window | deque of indices with decreasing values | O(n) / O(k) |
| **Binary search** | sorted, O(log n), first/last position, rotated arrays | `while lo < hi`: keep the half that must contain the answer | O(log n) / O(1) |
| **Binary search on the answer** | "minimum capacity/speed such that…", "minimize the maximum" | `first_true(lo, hi, feasible)` | O(n log range) / O(1) |
| **Intervals: merge** | overlapping ranges, insert into a schedule | sort by start; extend or append | O(n log n) / O(n) |
| **Intervals: keep most** | remove fewest overlaps, activity selection | sort by end; keep if `start >= last_end` | O(n log n) / O(1) |
| **Intervals: max concurrent** | rooms, GPUs, peak load | sort by start + min-heap of end times (or a sweep line) | O(n log n) / O(n) |

### Python idioms that save time in interviews

| Need | Idiom |
|---|---|
| Count things | `Counter(items)`, `counts[x] += 1` (missing keys read as 0) |
| Group things | `groups = defaultdict(list)`; `groups[key].append(item)` |
| Queue / BFS | `q = deque([start])`; `q.popleft()` — never `list.pop(0)` |
| Stack | `stack.append(x)`, `stack.pop()`, `stack[-1]` |
| Min-heap / max-heap | `heapq.heappush(h, x)`, `heapq.heappop(h)`; push `-x` for a max-heap; `(priority, tie_breaker, item)` tuples |
| Sorted search | `bisect_left(a, x)` (first ≥ x), `bisect_right(a, x)` (first > x) |
| Multi-key sort | `sorted(items, key=lambda t: (-t.score, t.name))` |
| Index and value | `for i, x in enumerate(nums)`; pairs of neighbours: `zip(nums, nums[1:])` |
| Infinity | `float("inf")`, `float("-inf")` |
| Integer ceil and truncation | `(a + b - 1) // b` for positive ceil; `//` rounds **down** (−7 // 2 = −4) |
| 2-D grid | `[[0] * cols for _ in range(rows)]` — never `[[0] * cols] * rows` |
| Build a string | collect pieces in a list, then `"".join(pieces)` |
| Reverse loop | `for i in range(n - 1, -1, -1)` or `for x in reversed(nums)` |
| Letter index | `ord(ch) - ord("a")` |
