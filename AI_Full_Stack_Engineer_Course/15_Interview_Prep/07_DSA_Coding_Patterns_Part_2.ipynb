### Recursion in Python — say this before you write a deep recursion

Python's default recursion limit is about 1,000 calls. Recursive DFS on a balanced tree of a million nodes is only ~20 deep (fine), but on a linked list, a skewed tree, or a 1000 × 1000 grid it can hit the limit. In an interview, write recursion when it's clearer, mention the limit, and offer the iterative version (an explicit stack) — every section below shows both where it matters.

## ⚙️ Setup

Run this cell first. It contains the same `check()` helper as Part 1 (✅ / ⏳ / ❌ with the first failing input, a hint, and an optional timed large input) plus:
- **`ListNode`**, **`build_list(values, cycle_at=None)`**, **`list_values(head)`** — linked lists from and to Python lists (`cycle_at` links the tail back to that index to make a cycle).
- **`TreeNode`**, **`build_tree(level_order)`**, **`tree_values(root)`**, **`show_tree(root)`** — binary trees from and to LeetCode's level-order lists with `None` for missing children, and a sideways printout.

````python
# Nothing to install: the course environment already has everything (standard library + numpy).
import bisect
import copy
import heapq
import itertools
import math
import random
import signal
import sys
import threading
import time
from collections import Counter, OrderedDict, defaultdict, deque
from functools import cache

import numpy as np

print(f"Python {sys.version.split()[0]} | NumPy {np.__version__}")
random.seed(0)


# ─── linked lists ──────────────────────────────────────────────────────────────
class ListNode:
    """One node of a singly linked list: a value and a pointer to the next node (None at the end)."""

    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self):
        return f"ListNode({self.val})"


def build_list(values, cycle_at=None):
    """[1, 2, 3] → 1 -> 2 -> 3. With cycle_at=i, the last node points back to node i (for cycle problems)."""
    dummy = tail = ListNode()
    nodes = []
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
        nodes.append(tail)
    if cycle_at is not None and nodes:
        tail.next = nodes[cycle_at]
    return dummy.next


def list_values(head, limit=10**7):
    """1 -> 2 -> 3 → [1, 2, 3]. Raises ValueError if the list is longer than `limit` (a cycle you created by mistake)."""
    values = []
    while head is not None:
        values.append(head.val)
        head = head.next
        if len(values) > limit:
            raise ValueError("list_values: more nodes than expected — did your code create a cycle?")
    return values


# ─── binary trees ──────────────────────────────────────────────────────────────
class TreeNode:
    """One node of a binary tree: a value and up to two children."""

    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def __repr__(self):
        return f"TreeNode({self.val})"


def build_tree(level_order):
    """LeetCode format: [3, 9, 20, None, None, 15, 7] lists nodes level by level; None marks a missing child."""
    if not level_order or level_order[0] is None:
        return None
    root = TreeNode(level_order[0])
    queue = deque([root])
    i = 1
    while queue and i < len(level_order):
        node = queue.popleft()
        for side in ("left", "right"):
            if i < len(level_order) and level_order[i] is not None:
                child = TreeNode(level_order[i])
                setattr(node, side, child)
                queue.append(child)
            i += 1
    return root


def tree_values(root):
    """The inverse of build_tree: level-order values with None for missing children, trailing Nones removed."""
    values, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node is None:
            values.append(None)
        else:
            values.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
    while values and values[-1] is None:
        values.pop()
    return values


def show_tree(root, indent=""):
    """Print a tree sideways: the root on the left, right subtree above, left subtree below."""
    if root is None:
        return
    show_tree(root.right, indent + "      ")
    print(f"{indent}── {root.val}")
    show_tree(root.left, indent + "      ")


def find_node(root, value):
    """Return the first node with this value (iterative DFS) — tests use it to pick nodes like p and q."""
    stack = [root]
    while stack:
        node = stack.pop()
        if node is None:
            continue
        if node.val == value:
            return node
        stack.extend((node.right, node.left))
    return None


# ─── the checker (same contract as Part 1) ─────────────────────────────────────
class _TooSlow(BaseException):
    """Raised by the timer inside check(). A BaseException, so `except Exception` in your code can't swallow it."""


def _run_with_time_limit(fn, args, seconds):
    use_alarm = hasattr(signal, "SIGALRM") and threading.current_thread() is threading.main_thread()
    running = [True]

    def on_alarm(signum, frame):
        if running[0]:
            raise _TooSlow()

    if use_alarm:
        previous = signal.signal(signal.SIGALRM, on_alarm)
        signal.setitimer(signal.ITIMER_REAL, seconds)
    start = time.perf_counter()
    try:
        result = fn(*args)
        running[0] = False
    finally:
        if use_alarm:
            signal.setitimer(signal.ITIMER_REAL, 0)
            signal.signal(signal.SIGALRM, previous)
    elapsed = time.perf_counter() - start
    if elapsed > seconds:
        raise _TooSlow()
    return result, elapsed


def _fresh(args):
    out = []
    for arg in args:
        if isinstance(arg, list) and all(isinstance(x, (int, float, str, bool, type(None))) for x in arg):
            out.append(arg[:])
        else:
            out.append(copy.deepcopy(arg))
    return tuple(out)


def _short(value, limit=110):
    text = repr(value)
    if len(text) <= limit:
        return text
    size = f" … ({len(value):,} items)" if hasattr(value, "__len__") else " …"
    return text[:limit] + size


def _same(same, got, expected):
    try:
        return bool(same(got, expected))
    except Exception:
        return False


def check(name, fn, cases, hint="", same=None, large=None):
    """Run fn(*args) for every (args, expected) in cases and print ✅ / ⏳ / ❌.

    same  : optional same(got, expected) -> bool; default is got == expected.
    large : optional dict(args=..., expected=..., seconds=2.0, label="n = ...") — must finish within `seconds`.
    """
    same = same or (lambda got, expected: got == expected)
    for number, (args, expected) in enumerate(cases, start=1):
        try:
            got = fn(*_fresh(args))
        except NotImplementedError:
            print(f"⏳ {name}: not attempted yet — replace `raise NotImplementedError` with your code.")
            return
        except RecursionError:
            raise AssertionError(f"❌ {name}: test {number} hit Python's recursion limit — infinite recursion or a missing base case?\n"
                                 f"   input   : {_short(args)}\n   hint    : {hint}") from None
        except Exception as error:
            raise AssertionError(f"❌ {name}: test {number} crashed with {type(error).__name__}: {error}\n"
                                 f"   input   : {_short(args)}\n   hint    : {hint}") from None
        if not _same(same, got, expected):
            raise AssertionError(f"❌ {name}: test {number} of {len(cases)} failed\n"
                                 f"   input   : {_short(args)}\n   expected: {_short(expected)}\n"
                                 f"   got     : {_short(got)}\n   hint    : {hint}")
    summary = f"all {len(cases)} tests pass"
    if large is not None:
        seconds, label = large.get("seconds", 2.0), large.get("label", "large input")
        try:
            got, elapsed = _run_with_time_limit(fn, large["args"], seconds)
        except _TooSlow:
            raise AssertionError(f"❌ {name}: the small tests pass, but the large input ({label}) did not finish within "
                                 f"{seconds} s — look for repeated work (a nested loop, or recursion without memoization).\n"
                                 f"   hint    : {hint}") from None
        except RecursionError:
            raise AssertionError(f"❌ {name}: the large input ({label}) hit Python's recursion limit — use an explicit stack or loop.\n"
                                 f"   hint    : {hint}") from None
        except Exception as error:
            raise AssertionError(f"❌ {name}: the large input ({label}) crashed with {type(error).__name__}: {error}\n"
                                 f"   hint    : {hint}") from None
        if not _same(same, got, large["expected"]):
            raise AssertionError(f"❌ {name}: wrong answer on the large input ({label})\n"
                                 f"   expected: {_short(large['expected'])}\n   got     : {_short(got)}\n   hint    : {hint}")
        summary += f" + large input ({label}) in {elapsed:.2f} s"
    print(f"✅ {name}: {summary}")


def any_order(got, expected):
    """Same items, order ignored."""
    return sorted(got) == sorted(expected)


def any_order_nested(got, expected):
    """Same groups, ignoring both the order of the groups and the order inside each group."""
    return sorted(sorted(group) for group in got) == sorted(sorted(group) for group in expected)


def close(got, expected, tol=1e-6):
    """Numbers (or lists of numbers) equal up to floating-point rounding."""
    if isinstance(expected, (list, tuple)):
        return isinstance(got, (list, tuple)) and len(got) == len(expected) and all(close(g, e, tol) for g, e in zip(got, expected))
    return math.isclose(got, expected, rel_tol=tol, abs_tol=tol)


def replay(cls, ops, args):
    """LeetCode-style design test: ops[0] builds cls(*args[0]); every later op calls that method. Returns all outputs."""
    obj = cls(*args[0])
    outputs = [None]
    for op, op_args in zip(ops[1:], args[1:]):
        outputs.append(getattr(obj, op)(*op_args))
    return outputs


def best_time(fn, *args, repeat=3):
    """Fastest of several runs, in seconds."""
    best = float("inf")
    for _ in range(repeat):
        start = time.perf_counter()
        fn(*args)
        best = min(best, time.perf_counter() - start)
    return best


demo_tree = build_tree([3, 9, 20, None, None, 15, 7])
print("build_list/list_values:", list_values(build_list([1, 2, 3])))
print("tree_values(build_tree(...)):", tree_values(demo_tree))
show_tree(demo_tree)
````

## 1. Linked Lists 🟢

A **linked list** is a chain of nodes. Each node stores a value and a pointer (`next`) to the following node; the last node points to `None`. Unlike a Python `list` (an array), the nodes aren't side by side in memory, so:

| Operation | Array (`list`) | Singly linked list |
|---|---|---|
| Read the i-th item | O(1) | O(i) — walk from the head |
| Insert/delete at the front | O(n) — shift everything | O(1) — rewire one pointer |
| Insert/delete after a node you hold | O(n) | O(1) |

```
head
 │
 ▼
[1 | •]──►[2 | •]──►[3 | •]──► None
```

### 🔍 Recognize it

The input is literally a `ListNode`, or the problem needs O(1) insert/delete in the middle of an ordering (LRU caches, schedulers). Cues: "reverse", "merge sorted lists", "detect a cycle", "k-th from the end", "middle node", "reorder / interleave".

### 🧰 The templates

```python
dummy = ListNode(0, head)          # 1) Dummy node: the head is no longer a special case; return dummy.next

prev, curr = None, head            # 2) Reverse: remember next BEFORE you rewire
while curr:
    nxt = curr.next
    curr.next = prev
    prev, curr = curr, nxt         # prev is the new head at the end

slow = fast = head                 # 3) Fast & slow pointers: middle of the list, cycle detection
while fast and fast.next:
    slow, fast = slow.next, fast.next.next
```

### 🧪 Worked example — Reverse Linked List (LeetCode 206), iterative and recursive

**Brute force:** copy the values into a Python list, reverse it, build a new list — O(n) time but O(n) extra space and new nodes. **In place:** walk once and flip each `next` pointer — O(n) time, O(1) space.

````python
def reverse_copy(head):                              # brute force: O(n) extra space
    values = list_values(head)
    return build_list(values[::-1])


def reverse_iterative(head):
    prev, curr = None, head
    while curr is not None:
        nxt = curr.next                              # 1. remember the rest of the list
        curr.next = prev                             # 2. flip this pointer
        prev, curr = curr, nxt                       # 3. step forward
    return prev


def reverse_recursive(head):
    if head is None or head.next is None:            # empty or one node: already reversed
        return head
    new_head = reverse_recursive(head.next)          # reverse everything after head
    head.next.next = head                            # the old next node now points back to head
    head.next = None
    return new_head


head = build_list([1, 2, 3, 4])
prev, curr, step = None, head, 0
print(f"{'step':>4}  {'prev (reversed part)':<22} curr (still to do)")
while curr is not None:
    print(f"{step:>4}  {str(list_values(prev)):<22} {list_values(curr)}")
    nxt = curr.next
    curr.next = prev
    prev, curr = curr, nxt
    step += 1
print(f"{step:>4}  {str(list_values(prev)):<22} {list_values(curr)}  ← prev is the new head")

for values in ([1, 2, 3, 4, 5], [1], []):
    assert list_values(reverse_iterative(build_list(values))) == values[::-1] == list_values(reverse_recursive(build_list(values)))
print("iterative and recursive agree on [1..5], [1], []")
try:
    reverse_recursive(build_list(range(5_000)))
except RecursionError:
    print("reverse_recursive on 5,000 nodes → RecursionError; reverse_iterative handles it:",
          list_values(reverse_iterative(build_list(range(5_000))))[:3], "...")
````

**Complexity:** iterative O(n) time, O(1) space; recursive O(n) time and O(n) call-stack space (and it hits Python's recursion limit on long lists).

**Edge cases to say out loud:** empty list, one node, two nodes (the smallest case where pointers actually flip).

```python
class LRUCacheOrdered:
    def __init__(self, capacity):
        self.capacity = capacity
        self.data = OrderedDict()                        # iteration order = recency order (oldest first)

    def get(self, key):
        if key not in self.data:
            return -1
        self.data.move_to_end(key)                       # mark as most recently used
        return self.data[key]

    def put(self, key, value):
        self.data[key] = value
        self.data.move_to_end(key)
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)                # drop the oldest


class _DNode:
    __slots__ = ("key", "value", "prev", "next")

    def __init__(self, key=None, value=None):
        self.key, self.value, self.prev, self.next = key, value, None, None


class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.nodes = {}                                  # key → node
        self.head, self.tail = _DNode(), _DNode()        # sentinels: head.next = most recent, tail.prev = least recent
        self.head.next, self.tail.prev = self.tail, self.head

    def _remove(self, node):
        node.prev.next, node.next.prev = node.next, node.prev

    def _add_to_front(self, node):
        node.prev, node.next = self.head, self.head.next
        self.head.next.prev = node
        self.head.next = node

    def get(self, key):
        node = self.nodes.get(key)
        if node is None:
            return -1
        self._remove(node)
        self._add_to_front(node)
        return node.value

    def put(self, key, value):
        node = self.nodes.get(key)
        if node is not None:
            node.value = value
            self._remove(node)
        else:
            node = _DNode(key, value)
            self.nodes[key] = node
            if len(self.nodes) > self.capacity:
                lru = self.tail.prev
                self._remove(lru)
                del self.nodes[lru.key]                  # this is why each node stores its key
        self._add_to_front(node)


large = dict(args=LRU_LARGE_ARGS, expected=LRU_LARGE_EXPECTED, seconds=3.0, label="200,000 operations, capacity 1,000")

check("LRU Cache (dict + doubly linked list)", run_lru(LRUCache), LRU_TESTS, large=large)
```

### 💡 Interview angle

**Q1. Why use a dummy (sentinel) node, and when do you need a doubly linked list instead of a singly linked one?**

<details><summary>Show answer</summary>

- **30-second answer:** A dummy node before the head gives every real node a predecessor, so inserting or deleting at the head isn't a special case — you return `dummy.next` at the end. You need a **doubly** linked list when you must delete a node you're holding in O(1) (LRU cache): a singly linked list needs the *previous* node, which takes O(n) to find.
- **Go deeper:** Sentinel head *and* tail nodes remove all `None` checks in a doubly linked list. The cost is two extra pointers per node. `collections.deque` is implemented as a doubly linked list of blocks, and `OrderedDict` keeps a doubly linked list of keys alongside its hash table — which is exactly why `move_to_end` is O(1).
- **❌ Common wrong answer:** "Linked lists are faster than arrays in Python" — traversal is slower (pointer chasing, poor cache locality); they win only for O(1) insert/delete at a known position.

</details>

**Q2. How would you detect where a cycle starts, or find the middle of a list in one pass?**

<details><summary>Show answer</summary>

- **30-second answer:** Middle: move `slow` one step and `fast` two; when `fast` reaches the end, `slow` is at the middle. Cycle start: run Floyd's algorithm until the pointers meet, then reset one pointer to `head` and advance both one step at a time — they meet at the cycle's first node.
- **Go deeper:** Proof sketch: if the head-to-entrance distance is a, and they meet b steps into a cycle of length c, then the fast pointer travelled 2(a + b) = a + b + kc, so a = kc − b — walking a steps from the head and from the meeting point lands both at the entrance. The same idea finds the duplicate in "Find the Duplicate Number" (LeetCode 287) by treating values as next pointers.
- **❌ Common wrong answer:** "Use a set of visited values" — values can repeat without a cycle; track node **identity** (`id(node)` or the node itself).

</details>

> 🤖 **AI/ML angle:** LRU caches sit in front of embedding lookups, feature stores and LLM response caches; `functools.lru_cache` is the same dict + doubly-linked-list design. In LLM serving, vLLM's PagedAttention manages KV-cache memory as fixed-size blocks and can evict cached prefix blocks with an LRU-style policy when memory runs short.

## 2. Trees: DFS & BFS 🟢

A **binary tree** is a set of nodes where each node has at most two children, `left` and `right`, and there's one **root** at the top. A node without children is a **leaf**. The **depth** of a node is its distance from the root; the **height** of a tree is the number of nodes on its longest root-to-leaf path.

```
        3          ← root (depth 0)
       / \
      9   20       ← depth 1
         /  \
        15   7     ← leaves (depth 2); height of the tree = 3
```

Two ways to visit every node:
- **DFS (depth-first search):** go as deep as possible, then back up. Recursive by nature (or an explicit **stack**). Orders: *preorder* (node, left, right), *inorder* (left, node, right), *postorder* (left, right, node).
- **BFS (breadth-first search):** visit level by level using a **queue** (`deque`).

### 🔍 Recognize it

The input is a `TreeNode`, or the data is hierarchical (file systems, org charts, parse trees, decision trees). Cues:
- "depth / height / diameter / balanced / path sum" → **DFS that returns something from each subtree** (postorder),
- "level by level / right side view / minimum depth / zigzag" → **BFS**,
- "compare two trees / mirror / subtree" → DFS on both trees at once.

### 🧰 The templates

```python
def dfs(node):                            # recursive DFS: "what does this function return for a subtree?"
    if node is None:
        return BASE                       # e.g. 0 for height
    left, right = dfs(node.left), dfs(node.right)
    return COMBINE(node.val, left, right) # e.g. 1 + max(left, right)

queue = deque([root])                     # BFS level by level
while queue:
    for _ in range(len(queue)):           # exactly the nodes of the current level
        node = queue.popleft()
        ...                               # process node
        if node.left:  queue.append(node.left)
        if node.right: queue.append(node.right)
```

### 🧪 Worked example — Maximum Depth of Binary Tree (LeetCode 104), three ways

````python
def max_depth_recursive(root):
    if root is None:
        return 0
    return 1 + max(max_depth_recursive(root.left), max_depth_recursive(root.right))


def max_depth_bfs(root):
    if root is None:
        return 0
    depth, queue = 0, deque([root])
    while queue:
        depth += 1                                     # one full level processed per loop
        for _ in range(len(queue)):
            node = queue.popleft()
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
    return depth


def max_depth_iterative_dfs(root):
    best, stack = 0, [(root, 1)]                       # explicit stack of (node, depth of that node)
    while stack:
        node, depth = stack.pop()
        if node is not None:
            best = max(best, depth)
            stack.append((node.left, depth + 1))
            stack.append((node.right, depth + 1))
    return best


root = build_tree([3, 9, 20, None, None, 15, 7])
show_tree(root)
print("depth:", max_depth_recursive(root), max_depth_bfs(root), max_depth_iterative_dfs(root))

# traversal orders on the same tree
def preorder(node):
    return [] if node is None else [node.val] + preorder(node.left) + preorder(node.right)


def inorder(node):
    return [] if node is None else inorder(node.left) + [node.val] + inorder(node.right)


def postorder(node):
    return [] if node is None else postorder(node.left) + postorder(node.right) + [node.val]


print("preorder:", preorder(root), "| inorder:", inorder(root), "| postorder:", postorder(root))

skewed = build_tree([0] + [x for i in range(1, 3_000) for x in (None, i)])   # every node has only a right child: height 3,000
try:
    max_depth_recursive(skewed)
except RecursionError:
    print("recursive DFS on a 3,000-deep skewed tree → RecursionError;",
          "BFS →", max_depth_bfs(skewed), "| iterative DFS →", max_depth_iterative_dfs(skewed))
````

**Complexity:** O(n) time for all three; space O(h) for DFS (h = height: O(log n) if balanced, O(n) if skewed) and O(w) for BFS (w = widest level, up to about n/2).

> The `preorder`/`inorder`/`postorder` one-liners above build many small lists — fine for learning the orders, but O(n²) in the worst case. In an interview, append to one shared result list instead.

### 💡 Interview angle

**Q3. Recursive vs iterative traversal — which do you write, and what are the trade-offs?**

<details><summary>Show answer</summary>

- **30-second answer:** Recursion mirrors the definition of a tree, so it's shorter and less error-prone — I write it first. Its space is the call stack, O(h), and Python raises `RecursionError` around depth 1,000, so for trees that may be skewed (a sorted insert order into a plain BST, a linked-list-like tree) I switch to an explicit stack (DFS) or queue (BFS), which lives on the heap.
- **Go deeper:** Iterative *inorder* needs the "go left as far as possible, then pop" loop; iterative *postorder* is easiest as "preorder with children swapped, then reversed". Morris traversal does inorder in O(1) extra space by temporarily threading right pointers — worth knowing exists, rarely expected.
- **❌ Common wrong answer:** "Recursion is O(1) space because I don't allocate anything."

</details>

**Q4. How do you design a recursive tree function when the answer isn't what the function naturally returns (diameter, max path sum, balanced check)?**

<details><summary>Show answer</summary>

- **30-second answer:** Separate *what the helper returns to its parent* (usually a height or the best one-sided path) from *what it records globally* (the best path that bends at this node). Define both in one sentence each before coding. Then each node does O(1) work and the whole thing is O(n).
- **Go deeper:** Alternatives to `nonlocal`: return a tuple `(height, best_so_far)`, or use an instance attribute. For "is balanced", return −1 as a sentinel for "unbalanced" so you can stop early. This "tree DP" generalizes to House Robber III (return `(rob_this, skip_this)`).
- **❌ Common wrong answer:** Calling a separate `height()` from every node — correct but O(n²) on skewed trees.

</details>

> 🤖 **AI/ML angle:** trees are everywhere in ML code: decision trees and gradient-boosted trees are traversed root-to-leaf at prediction time (depth = latency); hierarchical clustering produces a tree (dendrogram); LLM agents that "think in trees" (tree-of-thought search, MCTS in game-playing systems) are BFS/DFS with scoring; and JSON/HTML/AST parsing for tool outputs is DFS over a tree.

## 3. Binary Search Trees 🟡

A **binary search tree (BST)** is a binary tree with an ordering rule at **every** node: all values in its left subtree are smaller, and all values in its right subtree are larger. Two consequences:
- **Search, insert and delete walk one path** from the root: O(h) for height h — O(log n) if the tree is balanced, O(n) if it degenerates into a chain.
- **An inorder traversal (left, node, right) visits values in sorted order.**

```
        8               inorder: 1 3 4 6 7 8 10 13 14  (sorted)
      /   \
     3     10
    / \      \
   1   6      14
      / \    /
     4   7  13
```

Python has no built-in balanced BST. In practice you use a sorted list with `bisect`, a heap, or the third-party `sortedcontainers.SortedList`; in interviews you're given a `TreeNode`.

### 🔍 Recognize it

The input is a BST (the statement says so), and the question involves **order**: "k-th smallest", "validate", "closest value", "range sum between L and R", "successor", "lowest common ancestor in a BST". The ordering lets you skip whole subtrees.

### 🧰 The templates

```python
node = root                                   # walk one path: O(h)
while node:
    if target < node.val:   node = node.left
    elif target > node.val: node = node.right
    else:                   return node

def valid(node, low, high):                   # every node must lie inside the bounds its ancestors impose
    if node is None: return True
    if not low < node.val < high: return False
    return valid(node.left, low, node.val) and valid(node.right, node.val, high)
```

### 🧪 Worked example — search and insert, and why balance matters

````python
def bst_insert(root, value):
    """Insert value into the BST and return the root (iterative, so deep trees are fine)."""
    new = TreeNode(value)
    if root is None:
        return new
    node = root
    while True:
        if value < node.val:
            if node.left is None:
                node.left = new
                return root
            node = node.left
        else:
            if node.right is None:
                node.right = new
                return root
            node = node.right


def bst_search(root, value):
    """Return (node or None, number of nodes visited)."""
    steps = 0
    while root is not None:
        steps += 1
        if value == root.val:
            return root, steps
        root = root.left if value < root.val else root.right
    return None, steps


def inorder_iterative(root):
    values, stack, node = [], [], root
    while stack or node:
        while node:                                  # go as far left as possible
            stack.append(node)
            node = node.left
        node = stack.pop()
        values.append(node.val)                      # visit
        node = node.right                            # then the right subtree
    return values


root = None
for value in [8, 3, 10, 1, 6, 14, 4, 7, 13]:
    root = bst_insert(root, value)
show_tree(root)
print("inorder:", inorder_iterative(root), "| search 7 →", bst_search(root, 7)[1], "steps | search 5 →", bst_search(root, 5))

keys = list(range(2_000))
shuffled = keys[:]
random.Random(0).shuffle(shuffled)
random_root = sorted_root = None
for key in shuffled:
    random_root = bst_insert(random_root, key)
for key in keys:                                     # inserting keys in sorted order builds a chain
    sorted_root = bst_insert(sorted_root, key)
print(f"2,000 keys inserted in random order: height {max_depth_bfs(random_root)}, finding key 1,999 visits {bst_search(random_root, 1_999)[1]} nodes")
print(f"2,000 keys inserted in sorted order: height {max_depth_bfs(sorted_root)}, finding key 1,999 visits {bst_search(sorted_root, 1_999)[1]} nodes")
````

Same keys, very different trees: random insertion order gives a height of a couple of dozen, sorted order gives a 2,000-node chain where every operation is O(n). **Self-balancing** BSTs (AVL, red–black trees) rotate nodes after inserts to keep the height O(log n); databases use **B-trees**, which are wide balanced trees designed for disk pages.

```python

    stack = [(root, float("-inf"), float("inf"))]     # (node, exclusive lower bound, exclusive upper bound)

        if node is None:

        if not low < node.val < high:

        stack.append((node.left, low, node.val))      # everything on the left must stay below node.val
```

### 💡 Interview angle

**Q5. When is a BST O(log n), when isn't it, and what do real systems use instead?**

<details><summary>Show answer</summary>

- **30-second answer:** Operations cost O(h). A balanced tree has h ≈ log₂ n; inserting keys in sorted order into a plain BST builds a chain with h = n (we measured height 2,000 above). Self-balancing trees — AVL, red–black (Java's `TreeMap`, C++'s `std::map`) — guarantee O(log n) by rotating after updates. Databases and file systems use B-trees/B+-trees: very wide nodes that match disk pages, so a lookup touches only a handful of pages.
- **Go deeper:** In Python, use `bisect` on a sorted list (O(log n) search but O(n) insert), `heapq` if you only need the minimum, or `sortedcontainers.SortedList` (O(log n)-ish inserts and indexing). Skip lists (Redis sorted sets) are a randomized alternative with the same expected bounds.
- **❌ Common wrong answer:** "BST operations are always O(log n)."

</details>

**Q6. Why does checking each node against only its parent fail to validate a BST?**

<details><summary>Show answer</summary>

- **30-second answer:** The BST rule is about *entire subtrees*: in `[5, 4, 6, None, None, 3, 7]`, node 3 is a valid left child of 6, but it's in 5's right subtree, so it must be greater than 5. Passing down a `(low, high)` range — or checking that the inorder sequence is strictly increasing — captures every ancestor's constraint.
- **Go deeper:** Clarify duplicates (allowed on one side or not); use `float("inf")` rather than integer sentinels; the inorder check needs only the previous value, so O(h) space. For a BST with subtree sizes or sums, the same bounds-passing recursion answers "count/sum of keys in [L, R]" in O(h + output).
- **❌ Common wrong answer:** `node.left.val < node.val < node.right.val` at every node.

</details>

> 🤖 **AI/ML angle:** vector databases combine approximate nearest-neighbour search with **metadata filters** ("documents from the last 30 days, tenant = X"), and those filters are served by ordered indexes — B-trees in Postgres/pgvector, sorted segments elsewhere. Decision trees are *not* BSTs, but prediction is the same root-to-leaf walk, so a tree's depth is its inference latency.

## 4. Tries 🟡

A **trie** (pronounced "try", from re*trie*val), or **prefix tree**, stores a set of strings so that words sharing a prefix share a path. Each node has up to one child per next character, and a flag marks "a word ends here".

```
(root)
 ├─ t
 │  ├─ e
 │  │  ├─ a ●            words: tea, ted, ten, to, in, inn
 │  │  ├─ d ●            ● = end of a word
 │  │  └─ n ●
 │  └─ o ●
 └─ i
    └─ n ●
       └─ n ●
```

Inserting or looking up a word of length L costs **O(L)**, no matter how many words are stored — and "does any word start with this prefix?" is also O(L).

### 🔍 Recognize it

"**prefix**", "starts with", "autocomplete / typeahead", "dictionary of words with wildcard search", "find all words from a list in a grid" (Word Search II), "longest common prefix", "replace words by their shortest root".

### 🧰 The template

```python
class TrieNode:
    def __init__(self):
        self.children = {}          # character → TrieNode
        self.is_word = False

def insert(root, word):
    node = root
    for ch in word:
        node = node.children.setdefault(ch, TrieNode())
    node.is_word = True
```

### 🧪 Worked example — Implement Trie (Prefix Tree) (LeetCode 208)

**Brute force:** keep a list of words and scan it: `search` is O(N · L) with a list (O(L) with a set), but `starts_with` needs `any(w.startswith(prefix) for w in words)` — O(N · L) for N words. **Trie:** O(L) for all three operations.

````python
class TrieNode:
    __slots__ = ("children", "is_word")

    def __init__(self):
        self.children = {}
        self.is_word = False


class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for ch in word:
            node = node.children.setdefault(ch, TrieNode())    # create the child only if it's missing
        node.is_word = True

    def _walk(self, text):
        node = self.root
        for ch in text:
            node = node.children.get(ch)
            if node is None:
                return None
        return node

    def search(self, word):
        node = self._walk(word)
        return node is not None and node.is_word

    def starts_with(self, prefix):
        return self._walk(prefix) is not None


def print_trie(node, prefix="", indent=""):
    for ch, child in sorted(node.children.items()):
        print(f"{indent}{ch}{' ●  (' + prefix + ch + ')' if child.is_word else ''}")
        print_trie(child, prefix + ch, indent + "  ")


demo_trie = Trie()
for word in ["tea", "ted", "ten", "to", "in", "inn"]:
    demo_trie.insert(word)
print_trie(demo_trie.root)
print("search('te'):", demo_trie.search("te"), "| starts_with('te'):", demo_trie.starts_with("te"), "| search('inn'):", demo_trie.search("inn"))

word_rng = random.Random(3)
vocabulary = ["".join(word_rng.choices("abcdefghijklmnopqrstuvwxyz", k=word_rng.randint(5, 10))) for _ in range(20_000)]
prefixes = ["".join(word_rng.choices("abcdefghijklmnopqrstuvwxyz", k=4)) for _ in range(200)]
big_trie = Trie()
for word in vocabulary:
    big_trie.insert(word)
brute_answers = [any(word.startswith(p) for word in vocabulary) for p in prefixes]
assert brute_answers == [big_trie.starts_with(p) for p in prefixes]
t_brute = best_time(lambda: [any(word.startswith(p) for word in vocabulary) for p in prefixes], repeat=1)
t_trie = best_time(lambda: [big_trie.starts_with(p) for p in prefixes])
print(f"200 prefix queries over 20,000 words: scanning the list {t_brute * 1e3:.0f} ms | trie {t_trie * 1e3:.2f} ms "
      f"({sum(brute_answers)} of the prefixes exist)")
````

**Complexity:** O(L) per operation; memory O(total characters stored) — tries can be memory-hungry because every node holds a dictionary. Compressed tries (radix trees) merge single-child chains.

```python
        for word, count in word_counts.items():

            node = self.root
            self._offer(node, entry)                    # the root answers the empty prefix

        for ch in prefix:
            node = node.children.get(ch)
            if node is None:
                return []
        return [word for _, word in node.top[:k]]
```

```python
def find_words(board, words):
    root = {}
    for word in words:                                 # dict-of-dicts trie; "$" holds the finished word
        node = root
        for ch in word:
            node = node.setdefault(ch, {})
        node["$"] = word
    rows, cols = len(board), len(board[0])
    found = []

    def dfs(r, c, parent):
        ch = board[r][c]
        node = parent[ch]
        if "$" in node:
            found.append(node.pop("$"))                # record once; popping also prunes duplicates
        board[r][c] = "#"                              # mark as used on this path
        for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):

        board[r][c] = ch                               # restore
        if not node:                                   # no words left below → remove the branch
            parent.pop(ch)

    for r in range(rows):
        for c in range(cols):
            if board[r][c] in root:
                dfs(r, c, root)
    return found
```

**Q7. Trie or hash set — when is a trie worth its memory?**

<details><summary>Show answer</summary>

- **30-second answer:** A hash set answers "is this exact word present?" in O(L) with less memory and simpler code. ## 5. Heaps / Priority Queues 🟡

A **heap** is a binary tree stored inside a plain list, arranged so that every parent is ≤ its children (a **min-heap**). So `heap[0]` is always the smallest item, and for the item at index i its children are at `2i + 1` and `2i + 2`. A **priority queue** is the abstract idea — "always give me the most urgent item" — and a heap is the usual way to build one.

```
list:  [1, 3, 2, 7, 4, 5]          tree view:        1
index:  0  1  2  3  4  5                           /   \
                                                  3     2
                                                 / \   /
                                                7   4 5
```

| Operation | `heapq` call | Cost |
|---|---|---|
| Peek at the smallest | `heap[0]` | O(1) |
| Add an item | `heapq.heappush(heap, x)` | O(log n) — swap upward |
| Remove the smallest | `heapq.heappop(heap)` | O(log n) — swap downward |
| Pop then push in one step | `heapq.heapreplace(heap, x)` | O(log n) |
| Turn a list into a heap | `heapq.heapify(items)` | O(n) |

Python only has a min-heap. For a max-heap push `-x`. For records, push tuples `(priority, tie_breaker, item)` so ties never compare the items themselves.

### 🔍 Recognize it

"**k** largest / smallest / most frequent / closest", "**merge k** sorted streams", "**running median**", "always process the cheapest / earliest / highest-priority next" (schedulers, Dijkstra, event simulation), "top-k results so far" in a stream.

### 🧰 The template — keep the k best with a heap of size k

```python
heap = []                                  # min-heap holding the k largest seen so far
for x in stream:
    if len(heap) < k:
        heapq.heappush(heap, x)
    elif x > heap[0]:                      # better than the worst of the current top k
        heapq.heapreplace(heap, x)
# heap[0] is the k-th largest; memory stays O(k) however long the stream is
```

### 🧪 Worked example — Kth Largest Element in an Array (LeetCode 215)

**Sort:** O(n log n). **Heap of size k:** O(n log k) time and O(k) memory — works on a stream. **Quickselect:** O(n) on average (partition around a random pivot and recurse into one side only), O(n²) worst case.

````python
def kth_largest_sort(nums, k):
    return sorted(nums)[-k]


def kth_largest_heap(nums, k):
    heap = []
    for x in nums:
        if len(heap) < k:
            heapq.heappush(heap, x)
        elif x > heap[0]:
            heapq.heapreplace(heap, x)
    return heap[0]


def kth_largest_quickselect(nums, k):
    target = len(nums) - k                      # index of the answer in sorted order
    items = list(nums)
    pick = random.Random(0)
    lo, hi = 0, len(items) - 1
    while True:
        pivot = items[pick.randint(lo, hi)]
        segment = items[lo:hi + 1]
        smaller = [x for x in segment if x < pivot]      # 3-way partition handles duplicates
        equal = [x for x in segment if x == pivot]
        larger = [x for x in segment if x > pivot]
        items[lo:hi + 1] = smaller + equal + larger
        if target < lo + len(smaller):
            hi = lo + len(smaller) - 1
        elif target < lo + len(smaller) + len(equal):
            return pivot
        else:
            lo = lo + len(smaller) + len(equal)


heap_demo = [7, 3, 5, 1, 4, 2]
heapq.heapify(heap_demo)
print("heapify([7, 3, 5, 1, 4, 2]) →", heap_demo, "| children of index 1:", heap_demo[3], heap_demo[4])

numbers = [3, 2, 3, 1, 2, 4, 5, 5, 6]
print("k = 4:", kth_largest_sort(numbers, 4), kth_largest_heap(numbers, 4), kth_largest_quickselect(numbers, 4))

kth_rng = random.Random(1)
datasets = {"random order": [kth_rng.random() for _ in range(300_000)],
            "ascending order": [i / 300_000 for i in range(300_000)]}          # every new item beats the current top k
contenders = [("sorted()", kth_largest_sort), ("heapq.nlargest", lambda a, k: heapq.nlargest(k, a)[-1]),
              ("size-k heap loop", kth_largest_heap), ("quickselect", kth_largest_quickselect)]
for label, data in datasets.items():
    timings = {name: best_time(fn, data, 10, repeat=1) for name, fn in contenders}
    all_agree = len({fn(data, 10) for _, fn in contenders}) == 1
    print(f"n = 300,000, k = 10, {label}:\n   " + " | ".join(f"{name} {t * 1e3:.1f} ms" for name, t in timings.items())
          + f"\n   all four agree: {all_agree} | fastest here: {min(timings, key=timings.get)}")
````

Big-O is a guide, not a stopwatch — read the **fastest here** lines above. On **random** data a size-k heap rejects almost every item with one comparison against `heap[0]`, so it does very little work. On **ascending** data every item beats the current top k and pays for a `heapreplace`, while `sorted()` finishes an already-sorted list in about O(n) because Timsort detects the existing run. Which approach wins depends on the data; the heap's guaranteed advantages are **O(k) memory** and working on a **stream** you can't sort — say both in an interview.

How does the heap compare with NumPy's `argpartition` on embedding similarities? The vectors below are **random unit vectors** (synthetic on purpose — only the timing and the agreement matter here).

````python
emb_rng = np.random.default_rng(0)
embeddings = emb_rng.normal(size=(100_000, 64)).astype(np.float32)
embeddings /= np.linalg.norm(embeddings, axis=1, keepdims=True)
query = embeddings[123] + 0.1 * emb_rng.normal(size=64).astype(np.float32)
query /= np.linalg.norm(query)
similarities = embeddings @ query                           # cosine similarity: one matrix-vector product

k = 10
start = time.perf_counter()
heap_top = heapq.nlargest(k, range(len(similarities)), key=similarities.__getitem__)
t_heap = time.perf_counter() - start
start = time.perf_counter()
candidates = np.argpartition(-similarities, k)[:k]            # the k best in arbitrary order, O(n) in C
numpy_top = candidates[np.argsort(-similarities[candidates])].tolist()
t_numpy = time.perf_counter() - start
print(f"heapq.nlargest over a NumPy array: {t_heap * 1e3:6.1f} ms → {heap_top[:5]}")
print(f"np.argpartition + sort of k      : {t_numpy * 1e3:6.1f} ms → {numpy_top[:5]}")
print("same ids:", heap_top == numpy_top, "| nearest neighbour of the noisy query is item", numpy_top[0])
faster, slower = ("np.argpartition", "heapq.nlargest") if t_numpy < t_heap else ("heapq.nlargest", "np.argpartition")
print(f"{faster} was {max(t_heap, t_numpy) / min(t_heap, t_numpy):.1f}× faster than {slower} on this run")
del embeddings, similarities
````

When all the scores already sit in one NumPy array, `np.argpartition` does the selection inside compiled code, while `heapq.nlargest` makes a Python-level call for every score — compare the printed speed-up. The heap pattern is what you use when scores arrive **one at a time** (a stream, merged shard results, results from several retrievers) or when you can't hold them all.

### 💡 Interview angle

**Q9. Heap, sort, or quickselect for "top k" — how do you choose?**

<details><summary>Show answer</summary>

- **30-second answer:** Sort when n is small or you need everything ordered: O(n log n). A size-k heap when k ≪ n or the data is a stream: O(n log k) time, O(k) memory. Quickselect (or `np.argpartition`) when all data is in memory and you only need the k items: O(n) average. In Python, constant factors and the data's order matter — the Kth Largest timings above show the fastest method changing between random and ascending input — so measure on realistic data.
- **Go deeper:** For distributed top-k, take the top k from each shard, then merge those k × shards candidates with a heap — correct because the global top k must appear in some shard's top k. For approximate top-k over huge unbounded streams, see Count-Min Sketch + heap (Part 1, Q5).
- **❌ Common wrong answer:** "A heap is O(1) for top-k" — peeking is O(1); building the top k is O(n log k).

</details>

**Q10. Why two heaps for the running median, and what changes for a sliding window or huge streams?**

<details><summary>Show answer</summary>

- **30-second answer:** The median sits at the boundary between the smaller and larger halves. A max-heap exposes the largest of the small half and a min-heap the smallest of the large half, so inserting is O(log n) and reading the median is O(1). For a **sliding window** you must also delete the value leaving the window: lazy deletion (mark it, discard it when it reaches a heap top) keeps O(log n) amortized. For streams too big to store, use approximate quantile sketches (t-digest, KLL) with bounded memory.
- **Go deeper:** Rebalance after every insert so `len(low) - len(high)` ∈ {0, 1}. The same two-heap trick gives any fixed percentile (keep the split at 90/10 instead of 50/50).
- **❌ Common wrong answer:** "Append to a list and sort it on every `find_median` call."

</details>

## 6. Graphs 🟡

A **graph** is a set of **nodes** (vertices) connected by **edges**. Edges can be **directed** (a → b, like "course a is a prerequisite of b") or **undirected** (a friendship), and **weighted** (a road with a length) or unweighted. Trees and linked lists are special graphs; a grid is a graph where each cell connects to its 4 neighbours.

The usual representation in interviews is an **adjacency list** — a dict mapping each node to its neighbours:

```python
graph = defaultdict(list)
for a, b in edges:
    graph[a].append(b)
    graph[b].append(a)          # omit for directed graphs
```

It uses O(V + E) memory for V nodes and E edges (an adjacency matrix uses O(V²)).

### 🔍 Recognize it

| The problem says… | Algorithm |
|---|---|
| "connected regions", "islands", "flood fill", "how many groups" | DFS or BFS from every unvisited node |
| "shortest path / fewest steps", all edges equal | **BFS** (from several starts at once: multi-source BFS) |
| "prerequisites", "dependencies", "build order", "can all tasks finish" | **Topological sort** (Kahn's algorithm) |
| "are a and b connected?", "merge groups", edges arrive one by one, "redundant edge" | **Union–find** |
| "cheapest / fastest path" with non-negative weights | **Dijkstra** with a heap |
| "copy / clone the graph" | DFS or BFS with an `old → new` dictionary |

### 🧰 The templates

```python
visited = {start}                              # BFS: mark when you ENQUEUE, not when you dequeue
queue = deque([start])
while queue:
    node = queue.popleft()
    for neighbour in graph[node]:
        if neighbour not in visited:
            visited.add(neighbour)
            queue.append(neighbour)

indegree = {node: 0 for node in nodes}         # Kahn's topological sort
for a, b in edges: indegree[b] += 1            # edge a → b
queue = deque(n for n in nodes if indegree[n] == 0)
order = []
while queue:
    node = queue.popleft(); order.append(node)
    for nxt in graph[node]:
        indegree[nxt] -= 1
        if indegree[nxt] == 0: queue.append(nxt)
# len(order) < len(nodes)  ⇔  there is a cycle

dist = {start: 0}; heap = [(0, start)]         # Dijkstra
while heap:
    d, node = heapq.heappop(heap)
    if d > dist[node]: continue                # stale entry
    for nxt, w in graph[node]:
        if d + w < dist.get(nxt, float("inf")):
            dist[nxt] = d + w
            heapq.heappush(heap, (d + w, nxt))
```

### 🧪 Worked example — Number of Islands (LeetCode 200)

**Problem:** in a grid of `"1"` (land) and `"0"` (water), count the islands — groups of land cells connected horizontally or vertically.

Scan every cell; each time you find unvisited land, that's a new island — flood-fill it (DFS or BFS) so its cells aren't counted again. Every cell is visited once: O(rows · cols). The recursive DFS is the shortest code, but a large island makes the recursion as deep as the island is big.

````python
def num_islands_recursive(grid):
    grid = [row[:] for row in grid]
    rows, cols = len(grid), len(grid[0])

    def sink(r, c):
        if r < 0 or r >= rows or c < 0 or c >= cols or grid[r][c] != "1":
            return
        grid[r][c] = "0"                               # mark visited by sinking the land
        sink(r + 1, c); sink(r - 1, c); sink(r, c + 1); sink(r, c - 1)

    count = 0
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == "1":
                count += 1
                sink(r, c)
    return count


def num_islands_bfs(grid, labels=None):
    rows, cols = len(grid), len(grid[0])
    seen = [[False] * cols for _ in range(rows)]
    count = 0
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == "1" and not seen[r][c]:
                count += 1
                seen[r][c] = True
                queue = deque([(r, c)])
                while queue:
                    cr, cc = queue.popleft()
                    if labels is not None:
                        labels[cr][cc] = str(count)
                    for nr, nc in ((cr + 1, cc), (cr - 1, cc), (cr, cc + 1), (cr, cc - 1)):
                        if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == "1" and not seen[nr][nc]:
                            seen[nr][nc] = True               # mark when enqueuing → no cell is queued twice
                            queue.append((nr, nc))
    return count


grid = [list("11000"), list("11000"), list("00100"), list("00011"), list("10011")]
labels = [["." for _ in row] for row in grid]
print("islands:", num_islands_bfs(grid, labels), "| recursive DFS agrees:", num_islands_recursive(grid) == num_islands_bfs(grid))
for original, labelled in zip(grid, labels):
    print("".join(original), "  →  ", "".join(labelled))

all_land = [["1"] * 400 for _ in range(400)]
try:
    num_islands_recursive(all_land)
except RecursionError:
    print("400 × 400 all land: recursive DFS → RecursionError (the flood fill is up to 160,000 calls deep)")
print(f"400 × 400 all land: BFS → {num_islands_bfs(all_land)} island in {best_time(num_islands_bfs, all_land, repeat=1) * 1e3:.0f} ms")
````

**Complexity:** O(R · C) time; O(R · C) space for `seen` plus the queue (O(min(R, C)) for the BFS frontier on a full grid). Say out loud whether you may modify the input grid (sinking land saves the `seen` array).

### ✍️ Your Turn 6.1 — Clone Graph 🟡 (20 min)

**Pattern cue:** "deep copy a graph that may contain cycles" → traverse it while keeping a dictionary `original node → its copy`; the dictionary is both your visited set and the way to wire up neighbours.

````python
class GraphNode:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []


def build_graph(adjacency):
    """LeetCode format: adjacency[i] lists the neighbours of node i + 1. Returns node 1 (or None for an empty graph)."""
    if not adjacency:
        return None
    nodes = [GraphNode(i + 1) for i in range(len(adjacency))]
    for node, neighbours in zip(nodes, adjacency):
        node.neighbors = [nodes[j - 1] for j in neighbours]
    return nodes[0]


def describe_graph(start):
    """(adjacency list, set of node object ids) for the graph reachable from start."""
    if start is None:
        return [], set()
    by_value, queue = {start.val: start}, deque([start])
    while queue:
        node = queue.popleft()
        for neighbour in node.neighbors:
            if neighbour.val not in by_value:
                by_value[neighbour.val] = neighbour
                queue.append(neighbour)
    adjacency = [[n.val for n in by_value[v].neighbors] for v in range(1, len(by_value) + 1)]
    return adjacency, {id(node) for node in by_value.values()}


def clone_and_describe(adjacency):
    original = build_graph(adjacency)
    copied = clone_graph(original)
    _, original_ids = describe_graph(original)
    copied_adjacency, copied_ids = describe_graph(copied)
    return copied_adjacency, bool(original_ids & copied_ids)      # expected: same structure, and NO shared node objects


CLONE_TESTS = [
    (([],), ([], False)),                                           # empty graph
    (([[]],), ([[]], False)),                                       # one node, no edges
    (([[2], [1]],), ([[2], [1]], False)),
    (([[2, 4], [1, 3], [2, 4], [1, 3]],), ([[2, 4], [1, 3], [2, 4], [1, 3]], False)),     # a 4-cycle
    (([[2, 3, 4], [1], [1], [1]],), ([[2, 3, 4], [1], [1], [1]], False)),                 # a star
    (([[2], [1, 3], [2, 4], [3]],), ([[2], [1, 3], [2, 4], [3]], False)),                 # a path
    (([[2, 3, 4], [1, 3, 4], [1, 2, 4], [1, 2, 3]],), ([[2, 3, 4], [1, 3, 4], [1, 2, 4], [1, 2, 3]], False)),   # complete
    (([[2, 3], [1, 3], [1, 2]],), ([[2, 3], [1, 3], [1, 2]], False)),                     # a triangle
]
_ring = [[(i - 1) % 50_000 + 1, (i + 1) % 50_000 + 1] for i in range(50_000)]
CLONE_LARGE = dict(args=(_ring,), expected=(_ring, False), seconds=4.0, label="a ring of 50,000 nodes (too deep for recursion)")
````

### ✍️ Your Turn 6.2 — 🤖 Course Schedule II: topological order of a pipeline 🟡 (25 min)

**Pattern cue:** "prerequisites / dependencies → an order, or detect that none exists" → **Kahn's algorithm**: repeatedly take a node with no remaining prerequisites (indegree 0). If some nodes never reach indegree 0, they sit on a **cycle**.

It's the same problem as ordering the steps of an ML pipeline (ingest → clean → featurize → train → evaluate) — Airflow, Kubeflow and `make` all run a topological sort of a DAG (directed acyclic graph).

````python
def order_verdict(num_courses, prerequisites):
    order = find_order(num_courses, prerequisites)
    if order == []:
        return "impossible"
    position = {course: i for i, course in enumerate(order)}
    is_valid = sorted(order) == list(range(num_courses)) and all(position[pre] < position[course] for course, pre in prerequisites)
    return "valid order" if is_valid else f"INVALID order: {order[:12]}"


COURSE_TESTS = [
    ((2, [[1, 0]]), "valid order"),                                    # take 0, then 1
    ((2, [[1, 0], [0, 1]]), "impossible"),                             # a 2-cycle
    ((1, []), "valid order"),                                          # one course, no prerequisites
    ((3, []), "valid order"),                                          # any order works
    ((4, [[1, 0], [2, 0], [3, 1], [3, 2]]), "valid order"),            # a diamond
    ((3, [[0, 1], [1, 2], [2, 0]]), "impossible"),                     # a 3-cycle
    ((2, [[0, 0]]), "impossible"),                                     # a course that requires itself
    ((5, [[1, 0], [2, 1], [3, 2], [4, 3]]), "valid order"),            # a chain
    ((6, [[1, 0], [2, 1], [0, 2], [4, 3], [5, 4]]), "impossible"),     # one component has a cycle
    ((3, [[1, 0], [1, 0]]), "valid order"),                            # a duplicated edge
]
_chain_edges = [[i + 1, i] for i in range(99_999)]
random.Random(7).shuffle(_chain_edges)
COURSE_LARGE = dict(args=(100_000, _chain_edges), expected="valid order", seconds=3.0,
                    label="100,000 courses in one prerequisite chain")
````

```python
def find_order(num_courses, prerequisites):
    graph = defaultdict(list)
    indegree = [0] * num_courses
    for course, prerequisite in prerequisites:
        graph[prerequisite].append(course)             # edge: prerequisite → course
        indegree[course] += 1
    queue = deque(c for c in range(num_courses) if indegree[c] == 0)
    order = []
    while queue:
        course = queue.popleft()
        order.append(course)
        for nxt in graph[course]:
            indegree[nxt] -= 1                          # one prerequisite of nxt is done
            if indegree[nxt] == 0:
                queue.append(nxt)
    return order if len(order) == num_courses else []   # leftovers are stuck on a cycle

```

**DFS alternative:** three colours per node (unvisited / on the current path / done); reaching an "on the path" node means a cycle; the reverse of the finishing order is a topological order. It's elegant but recursive — the 100,000-course chain is too deep for Python's default limit.
</details>

Python's standard library ships this algorithm as `graphlib.TopologicalSorter`, including grouping steps that can run **in parallel** — exactly what a pipeline orchestrator needs:

````python
from graphlib import CycleError, TopologicalSorter

pipeline = {                                   # step → the steps it depends on
    "clean": {"ingest"},
    "split": {"clean"},
    "featurize": {"clean"},
    "train": {"featurize", "split"},
    "evaluate": {"train", "split"},
    "deploy": {"evaluate"},
}
print("one valid order:", list(TopologicalSorter(pipeline).static_order()))

sorter = TopologicalSorter(pipeline)
sorter.prepare()
stage = 0
while sorter.is_active():
    ready = sorted(sorter.get_ready())         # every step whose dependencies are all done
    print(f"stage {stage}: run in parallel → {ready}")
    sorter.done(*ready)
    stage += 1

try:
    list(TopologicalSorter({"train": {"evaluate"}, "evaluate": {"train"}}).static_order())
except CycleError as error:
    print("cycle detected:", error.args[1])
````

```python
def oranges_rotting(grid):
    rows, cols = len(grid), len(grid[0])
    queue = deque()
    fresh = 0
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 2:
                queue.append((r, c))              # every rotten orange is a BFS source
            elif grid[r][c] == 1:
                fresh += 1
    minutes = 0
    while queue and fresh:
        minutes += 1
        for _ in range(len(queue)):               # one BFS level = one minute
            r, c = queue.popleft()
            for nr, nc in ((r + 1, c), (r - 1, c), (r, c + 1), (r, c - 1)):
                if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1:
                    grid[nr][nc] = 2
                    fresh -= 1
                    queue.append((nr, nc))
    return minutes if fresh == 0 else -1


check("Rotting Oranges", oranges_rotting, ORANGES_TESTS, large=ORANGES_LARGE)
```

**Complexity:** O(R · C) time and space. Simulating minute by minute with a full grid scan each minute is O((R · C) · minutes) — up to 250,000 × 998 cell checks on the large test.

Stopping the loop when `fresh == 0` matters: otherwise the last level (rotten oranges with no fresh neighbours) adds an extra minute.

### ✍️ Your Turn 6.4 — Union–Find: connected components & redundant connection 🟡 (25 min)

**Pattern cue:** "edges arrive and you must keep asking **are these two connected?**" or "which edge closes a cycle?" → a **disjoint set union (union–find)**: every node points toward a representative ("root") of its group.
- `find(x)`: follow parent pointers to the root, and **compress the path** (point nodes closer to the root).
- `union(a, b)`: attach the root of the smaller/shorter tree under the other (**union by rank/size**); return False if they already share a root.

With both tricks each operation costs almost O(1) — formally O(α(n)), where α (the inverse Ackermann function) is below 5 for any realistic n.

```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))           # every node starts as its own group
        self.rank = [0] * n                    # an upper bound on each tree's height

    def find(self, x):
        root = x
        while self.parent[root] != root:       # 1) walk up to the root
            root = self.parent[root]
        while self.parent[x] != root:          # 2) path compression: point every node on the path at the root
            self.parent[x], x = root, self.parent[x]
        return root

    def union(self, a, b):
        root_a, root_b = self.find(a), self.find(b)
        if root_a == root_b:
            return False                       # already connected → this edge would close a cycle
        if self.rank[root_a] < self.rank[root_b]:
            root_a, root_b = root_b, root_a    # attach the shorter tree under the taller one
        self.parent[root_b] = root_a
        if self.rank[root_a] == self.rank[root_b]:
            self.rank[root_a] += 1
        return True


def count_components(n, edges):
    uf = UnionFind(n)
    components = n
    for a, b in edges:
        if uf.union(a, b):
            components -= 1                    # every successful merge removes one group
    return components


def find_redundant_connection(edges):
    uf = UnionFind(len(edges) + 1)             # nodes are 1..n and there are n edges
    for a, b in edges:
        if not uf.union(a, b):
            return [a, b]
    return []


check("Number of Connected Components", count_components, COMPONENTS_TESTS, large=COMPONENTS_LARGE)
check("Redundant Connection", find_redundant_connection, REDUNDANT_TESTS, large=REDUNDANT_LARGE)
```

**Complexity:** O(E · α(V)) ≈ O(E) time for E edges, O(V) space. Without rank *and* without compression, a chain of unions can make `find` O(V) — O(V · E) overall. The iterative `find` avoids recursion limits on long chains.

**BFS/DFS also counts components** in O(V + E) — prefer union–find when edges arrive one at a time (online) or you need "connected?" queries between edge insertions.

### ✍️ Your Turn 6.5 — 🤖 Group near-duplicate documents 🟡 (20 min)

Deduplicating a training corpus or a RAG knowledge base: two texts are **near-duplicates** if the Jaccard similarity of their character 3-gram sets ("shingles") is at least a threshold. Groups are **transitive** — if A ≈ B and B ≈ C, all three belong together even when A and C aren't similar enough on their own. That's connected components, so union–find does the grouping.

The similarity helpers are given; your job is the grouping.

````python
def shingles(text, size=3):
    """The set of overlapping character n-grams of text (a text shorter than size is a single shingle)."""
    return {text[i:i + size] for i in range(len(text) - size + 1)} if len(text) >= size else {text}


def jaccard(a, b):
    """|A ∩ B| / |A ∪ B| for the shingle sets of two texts: 1.0 for identical sets, 0.0 for nothing in common."""
    sa, sb = shingles(a), shingles(b)
    return len(sa & sb) / len(sa | sb)


_DOCS = ["the cat sat on the mat", "the cat sat on a mat", "the cat sat on the mat.", "a dog barked in the fog",
         "a dog barked in fog", "quarterly revenue grew 12%", "unrelated sentence here"]
print("similarities of doc 0 to the others:", [round(jaccard(_DOCS[0], d), 3) for d in _DOCS])
print("doc 1 vs doc 2:", round(jaccard(_DOCS[1], _DOCS[2]), 3), "← below 0.68, yet both are ≥ 0.68 similar to doc 0")
_CHAIN = ["abcdefgh", "abcdefxy", "abcdwxyz", "zzzz"]
NEAR_DUP_TESTS = [
    ((_DOCS, 0.68), [[0, 1, 2], [3, 4], [5], [6]]),              # docs 1 and 2 are grouped through doc 0 (transitivity)
    ((_DOCS, 0.8), [[0, 2], [1], [3], [4], [5], [6]]),           # a stricter threshold
    (([], 0.5), []),                                             # no documents
    ((["hello"], 0.5), [[0]]),                                   # one document
    ((["same text", "same text"], 1.0), [[0, 1]]),               # identical texts at threshold 1.0
    ((["ab", "ab", "abc"], 0.5), [[0, 1], [2]]),                 # texts shorter than a shingle
    ((["x", "y"], 0.0), [[0, 1]]),                               # threshold 0 → everything is "similar"
    ((_CHAIN, 0.3), [[0, 1], [2], [3]]),
    ((_CHAIN, 0.2), [[0, 1, 2], [3]]),
]
````

```python
def group_near_duplicates(texts, threshold):
    parent = list(range(len(texts)))

    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]         # path halving: a simple form of path compression
            x = parent[x]
        return x

    for i in range(len(texts)):
        for j in range(i + 1, len(texts)):
            if jaccard(texts[i], texts[j]) >= threshold:
                parent[find(i)] = find(j)         # union the two groups
    groups = defaultdict(list)
    for i in range(len(texts)):
        groups[find(i)].append(i)
    return list(groups.values())


check("Group near-duplicates", group_near_duplicates, NEAR_DUP_TESTS, same=any_order_nested)
```

Transitivity cuts both ways: long chains can merge documents that aren't similar at all. Production pipelines sometimes cap group sizes or re-check against a group representative.

### ✍️ Your Turn 6.6 — Network Delay Time (Dijkstra) 🟡 (25 min)

```python
def network_delay_time(times, n, k):
    graph = defaultdict(list)
    for u, v, w in times:
        graph[u].append((v, w))
    dist = {}                                         # node → final shortest distance
    heap = [(0, k)]
    while heap:
        d, node = heapq.heappop(heap)
        if node in dist:
            continue                                  # an older, longer entry for a node that's already final
        dist[node] = d                                # the smallest distance in the heap is final (weights ≥ 0)
        for nxt, w in graph[node]:
            if nxt not in dist:
                heapq.heappush(heap, (d + w, nxt))
    return max(dist.values()) if len(dist) == n else -1
```

### 💡 Interview angle

**Q11. BFS or DFS — and why do graph searches loop forever?**

<details><summary>Show answer</summary>

- **30-second answer:** Both visit every reachable node in O(V + E). **BFS** explores in rings of increasing distance, so it finds **shortest paths in unweighted graphs** and is naturally iterative; **DFS** goes deep first, which suits cycle detection, topological order, connected components and backtracking-style exploration. Graphs (unlike trees) have cycles, so without a `visited` set a search revisits nodes forever — mark nodes when you **enqueue** them in BFS so nothing is queued twice.
- **Go deeper:** Bidirectional BFS (search from both ends) cuts the explored area from about b^d to 2·b^(d/2). For weights that are only 0 or 1, a deque-based "0-1 BFS" gives shortest paths in O(V + E). Recursive DFS needs the recursion-limit caveat in Python; an explicit stack doesn't.
- **❌ Common wrong answer:** "DFS finds the shortest path if you stop at the first time you reach the target."

</details>

**Q12. Dijkstra, Bellman–Ford, BFS, topological-order DP — how do you pick a shortest-path algorithm?**

<details><summary>Show answer</summary>

- **30-second answer:** Unweighted → BFS, O(V + E). Non-negative weights → Dijkstra with a heap, O(E log V). Negative weights (no negative cycles) → Bellman–Ford, O(V · E), which also detects negative cycles. A DAG (even with negative weights) → relax edges in topological order, O(V + E). All pairs on a small dense graph → Floyd–Warshall, O(V³).
- **Go deeper:** Dijkstra fails with negative edges because it finalizes a node the first time it's popped, and a later negative edge could have made it cheaper. A* is Dijkstra plus an admissible heuristic (e.g. straight-line distance) that steers the search toward the goal — route planners and game AI use it.
- **❌ Common wrong answer:** "Dijkstra works for any weights if you let nodes be revisited" — with negative cycles distances aren't even defined.

</details>

**Q13. When is union–find the right tool, and what can't it do?**

<details><summary>Show answer</summary>

- **30-second answer:** Use it for **dynamic connectivity with only additions**: edges arrive one at a time and you ask "are x and y connected?", count groups, detect the edge that closes a cycle in an undirected graph, or build a minimum spanning tree with Kruskal's algorithm. With path compression and union by rank each operation is effectively O(1).
- **Go deeper:** It can't *delete* edges (splitting a group) cheaply — offline tricks process deletions in reverse as additions. It also can't give you the path between two nodes, only whether one exists. Weighted union–find stores relative values on edges (e.g. "Evaluate Division", ratios between variables).
- **❌ Common wrong answer:** "Union–find finds shortest paths."

</details>

## 7. Backtracking 🟡

**Backtracking** builds a solution one choice at a time and **undoes** a choice as soon as it can't lead to a valid answer — like exploring a maze: walk down a corridor, and at a dead end step back to the last junction and try the next corridor.

Every backtracking search walks a **decision tree**. For "all subsets of [1, 2, 3]" each level decides whether to include one number:

```
                         []
               /                   \
          take 1                  skip 1
          [1]                       []
        /      \                 /      \
    [1,2]      [1]            [2]        []
    /   \     /   \          /   \      /  \
[1,2,3][1,2][1,3] [1]    [2,3]  [2]   [3]  []     ← 2³ = 8 leaves = 8 subsets
```

### 🔍 Recognize it

- "Return **all** subsets / permutations / combinations / partitions / valid arrangements",
- "place items under constraints" (N-Queens, Sudoku), "find a path of letters on a grid" (Word Search),
- small inputs (n ≤ ~20): the output itself is exponential, so no polynomial algorithm exists.

If the question asks only for the **number** of ways or the **best** way and subproblems repeat, it's probably DP (section 8) instead.

### 🧰 The template — choose, explore, un-choose

```python
results = []

def backtrack(start, path):              # path = the partial solution so far
    if is_complete(path):
        results.append(path[:])          # COPY — path keeps changing afterwards
        return
    for i in range(start, len(choices)):
        if not is_valid(choices[i], path):
            continue                     # prune: skip branches that can't work
        path.append(choices[i])          # choose
        backtrack(i + 1, path)           # explore (i + 1: each item used at most once)
        path.pop()                       # un-choose

backtrack(0, [])
```

### 🧪 Worked example — Subsets (LeetCode 78)

**Problem:** return all subsets of a list of distinct integers. There are 2ⁿ of them, so every correct algorithm is at least O(n · 2ⁿ) (each subset costs up to n to copy). The interesting part is producing them cleanly.

````python
def subsets_bitmask(nums):                    # "brute force": every number 0..2ⁿ−1 is a yes/no pattern
    n = len(nums)
    return [[nums[i] for i in range(n) if mask >> i & 1] for mask in range(1 << n)]


calls = 0


def subsets_backtrack(nums):
    global calls
    results, path = [], []

    def backtrack(start):
        global calls
        calls += 1
        results.append(path[:])               # every node of the tree is a subset
        for i in range(start, len(nums)):
            path.append(nums[i])              # choose
            backtrack(i + 1)                  # explore
            path.pop()                        # un-choose

    backtrack(0)
    return results


print("bitmask  :", subsets_bitmask([1, 2, 3]))
print("backtrack:", subsets_backtrack([1, 2, 3]), f"({calls} calls)")

path = []


def trace(nums, start=0, depth=0):
    print("  " * depth + f"path={path}")
    for i in range(start, len(nums)):
        path.append(nums[i])
        trace(nums, i + 1, depth + 1)
        path.pop()


trace([1, 2, 3])
for n in (10, 15):
    print(f"n = {n}: {len(subsets_backtrack(list(range(n)))):,} subsets")
````

**Complexity:** O(n · 2ⁿ) time and O(n · 2ⁿ) output space; the recursion itself uses O(n) extra space.

**The bug everyone makes once:** `results.append(path)` instead of `results.append(path[:])` — every stored "subset" is the same list object, which ends up empty after the last `pop()`.

```python
def exist(board, word):
    rows, cols = len(board), len(board[0])
    board_counts = Counter(ch for row in board for ch in row)
    if any(board_counts[ch] < need for ch, need in Counter(word).items()):
        return False                                    # prune: a letter the board can't supply
    if board_counts[word[0]] > board_counts[word[-1]]:
        word = word[::-1]                               # start from the rarer end: fewer starting points

    def dfs(r, c, i):
        if r < 0 or r >= rows or c < 0 or c >= cols or board[r][c] != word[i]:
            return False
        if i == len(word) - 1:
            return True
        saved, board[r][c] = board[r][c], "#"          # choose: mark as used on this path
        found = dfs(r + 1, c, i + 1) or dfs(r - 1, c, i + 1) or dfs(r, c + 1, i + 1) or dfs(r, c - 1, i + 1)
        board[r][c] = saved                             # un-choose: restore for other paths
        return found

    return any(dfs(r, c, 0) for r in range(rows) for c in range(cols))


check("Word Search", exist, WORD_SEARCH_TESTS, large=WORD_SEARCH_LARGE)
```

### ✍️ Your Turn 7.4 — N-Queens 🔴 (optional, 35 min)

**Pattern cue:** "place n pieces so none attack each other, return all boards" → one queen per row; for each row try each column, and keep **sets** of used columns and diagonals so each validity check is O(1). Cells on the same "↘" diagonal share `row − col`; on the same "↙" diagonal they share `row + col`.

````python
def _queens_reference(n):                               # a bitmask search (a different implementation) builds expected boards
    boards = []

    def place(row, cols, diag1, diag2, placed):
        if row == n:
            boards.append(["." * c + "Q" + "." * (n - c - 1) for c in placed])
            return
        free = ~(cols | diag1 | diag2) & ((1 << n) - 1)
        while free:
            bit = free & -free
            free ^= bit
            c = bit.bit_length() - 1
            place(row + 1, cols | bit, (diag1 | bit) << 1, (diag2 | bit) >> 1, placed + [c])

    place(0, 0, 0, 0, [])
    return boards


N_QUEENS_TESTS = [((n,), _queens_reference(n)) for n in (1, 2, 3, 4, 5, 6, 7, 8)]
N_QUEENS_TESTS[3] = ((4,), [[".Q..", "...Q", "Q...", "..Q."], ["..Q.", "Q...", "...Q", ".Q.."]])   # n = 4, written out
N_QUEENS_LARGE = dict(args=(10,), expected=_queens_reference(10), seconds=3.0, label="n = 10 → 724 boards")
print("solutions per n:", [len(expected) for _, expected in N_QUEENS_TESTS], "| n = 10:", len(N_QUEENS_LARGE["expected"]))
````

```python
    cols, diagonals, anti_diagonals = set(), set(), set()

    def backtrack(row):

            results.append(["." * c + "Q" + "." * (n - c - 1) for c in queen_col])

            backtrack(row + 1)
            queen_col.pop()

    return results
```

### 💡 Interview angle

**Q14. How do you estimate the complexity of a backtracking solution, and how do you speed it up?**

<details><summary>Show answer</summary>

- **30-second answer:** Count the leaves of the decision tree times the work per leaf: subsets 2ⁿ × O(n), permutations n! × O(n), a grid word search R · C · 3^L. Speed-ups don't change the worst case but cut real work: **prune early** (sorted candidates + `break`, constraint sets, letter counts), **order choices** (most constrained first, rarer end of a word first), avoid copying until a solution is complete, and **memoize** if the same sub-state repeats — at which point it has become DP.
- **Go deeper:** Bitmasks replace sets for small n (N-Queens with three integers). For "does any solution exist?" return as soon as one is found. Constraint solvers (SAT/CP-SAT) generalize these ideas with propagation and learning.
- **❌ Common wrong answer:** "Backtracking is O(n²) because there are two nested loops."

</details>

**Q15. Why does Combination Sum pass `i` but Subsets passes `i + 1`, and how do you avoid duplicate results when the input has duplicates?**

<details><summary>Show answer</summary>

- **30-second answer:** The start index controls reuse: `i` lets the same element be chosen again (unlimited reuse), `i + 1` moves past it (each element at most once). Never restarting from 0 is what keeps results in non-decreasing index order, so `[2, 3]` and `[3, 2]` can't both appear. With duplicate values (Subsets II, Combination Sum II), sort and skip a value equal to the previous one at the same tree level: `if i > start and nums[i] == nums[i - 1]: continue`.
- **Go deeper:** For permutations with duplicates the skip rule uses a `used` array: skip `nums[i]` if it equals `nums[i-1]` and `nums[i-1]` isn't used. A `set` of tuples also removes duplicates but wastes the work of generating them.
- **❌ Common wrong answer:** "Generate everything and deduplicate with a set at the end."

</details>

## 8. Dynamic Programming 🔴

**Dynamic programming (DP)** solves a problem by combining answers to smaller versions of the same problem, and **stores** each sub-answer so it's computed only once. It applies when:
1. **Optimal substructure:** the best answer for size n can be built from best answers for smaller sizes, and
2. **Overlapping subproblems:** a plain recursion would solve the same smaller problem many times.

Analogy: working out 1 + 1 + 1 + 1 + 1 = 5, then being asked "and one more?" — you don't recount, you remember 5.

### 🔍 Recognize it

- "**How many ways**…", "**minimum / maximum** cost / length / profit…", "**is it possible** to…",
- choices at each step that affect later options (rob this house or skip it, take this item or not),
- two strings compared character by character (LCS, edit distance), or a prefix of a string (word break),
- a brute-force recursion you can write, but whose call tree clearly repeats arguments.

### 🧰 The recipe (say these steps out loud)

1. **State:** define `dp[i]` (or `dp[i][j]`) in one English sentence — "the fewest coins that make amount i".
2. **Recurrence:** how `dp[i]` follows from smaller states — "`dp[i] = 1 + min(dp[i − c] for each coin c)`".
3. **Base cases:** the smallest states you know directly — "`dp[0] = 0`".
4. **Order:** compute states so that everything a state needs is ready (or memoize a recursion and let it figure out the order).
5. **Answer:** where it lives — `dp[amount]`, `max(dp)`, `dp[n][m]`.
6. **Space:** if `dp[i]` only needs the previous row or two values, keep only those.

```python
@cache                                   # top-down: write the recursion, add a cache
def solve(i):
    if i <= BASE: return BASE_VALUE
    return COMBINE(solve(i - 1), solve(i - 2))

dp = [BASE_VALUE] * (n + 1)              # bottom-up: fill a table from small to large
for i in range(2, n + 1):
    dp[i] = COMBINE(dp[i - 1], dp[i - 2])
```

### 🧪 Worked example — Climbing Stairs (LeetCode 70): recursion → memo → table → O(1) space

**Problem:** you can climb 1 or 2 steps at a time. How many distinct ways reach step n? The last move was either a 1-step (from n − 1) or a 2-step (from n − 2), so `ways(n) = ways(n − 1) + ways(n − 2)` with `ways(0) = ways(1) = 1`.

````python
call_count = 0


def climb_naive(n):                                  # 1) plain recursion: O(2ⁿ) calls
    global call_count
    call_count += 1
    if n <= 1:
        return 1
    return climb_naive(n - 1) + climb_naive(n - 2)


@cache
def climb_memo(n):                                   # 2) top-down memoization: each n computed once → O(n)
    if n <= 1:
        return 1
    return climb_memo(n - 1) + climb_memo(n - 2)


def climb_table(n):                                  # 3) bottom-up table: O(n) time, O(n) space
    dp = [1] * (n + 1)
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]


def climb_two_vars(n):                               # 4) keep only the last two values: O(n) time, O(1) space
    prev, curr = 1, 1
    for _ in range(n - 1):
        prev, curr = curr, prev + curr
    return curr


for n in (20, 25, 30):
    call_count = 0
    start = time.perf_counter()
    value = climb_naive(n)
    print(f"n = {n}: naive recursion {value:>9,} ways, {call_count:>9,} calls, {(time.perf_counter() - start) * 1e3:7.1f} ms")
print("memo / table / two variables for n = 30:", climb_memo(30), climb_table(30), climb_two_vars(30))
print("n = 90 in O(1) space:", climb_two_vars(90))

climb_memo.cache_clear()
try:
    climb_memo(5_000)
except RecursionError:
    print("climb_memo(5000) → RecursionError: top-down recursion is still ~5,000 calls deep; bottom-up →",
          str(climb_table(5_000))[:20] + "…", f"({len(str(climb_table(5_000)))} digits)")
````

The naive call count grows by about 1.6× per extra step (it *is* the Fibonacci sequence), while the memoized version makes n + 1 distinct calls. `functools.cache` (Python 3.9+) is `lru_cache(maxsize=None)`: it stores every result keyed by the arguments.

**Top-down vs bottom-up:** top-down is often faster to *write* (it's the recursion plus one decorator) and only computes the states it needs; bottom-up avoids recursion limits and function-call overhead and makes space optimization obvious. In Python interviews, write the recursion + `@cache` first if it's clearer, then convert.

```python
    dp = [0] * (capacity + 1)                         # dp[c] = best value using the items so far with capacity c
```

### 💡 Interview angle

**Q16. How do you turn a brute-force recursion into DP, step by step?**

<details><summary>Show answer</summary>

- **30-second answer:** (1) Write the recursion with the smallest set of arguments that determines the answer — that's the **state**. (2) Check that calls repeat. (3) Add `@cache` — top-down DP, done. (4) For bottom-up, allocate a table indexed by the state, fill base cases, and loop in an order where every state's dependencies are already computed. (5) If each row only reads the previous row, keep two rows (or two variables). Complexity = number of states × work per state.
- **Go deeper:** Arguments that are lists or sets aren't hashable — use indices into the original input instead (`i`, `j`), or a bitmask for small sets. Memoized recursion in Python hits the recursion limit near depth 1,000 and has per-call overhead; bottom-up avoids both. Watch memory: `@cache` on a 2-D state of 10⁴ × 10⁴ is 10⁸ entries.
- **❌ Common wrong answer:** "DP means filling a 2-D table" — the table is an implementation detail; the state and recurrence are the idea.

</details>

**Q17. Top-down memoization vs bottom-up tabulation — which would you pick?**

<details><summary>Show answer</summary>

- **30-second answer:** Top-down is faster to write correctly from a recursion and only visits reachable states (great when most states are never needed, e.g. word break with few matching prefixes). Bottom-up has no recursion limit, lower constant factors in Python, and makes space optimization possible. In an interview I write whichever I can explain fastest, then offer the other with its trade-off.
- **Go deeper:** Bottom-up needs a valid evaluation order, which can be awkward for DP on trees or DAGs (use postorder / topological order). `functools.cache` never evicts — clear it between independent test cases (`f.cache_clear()`) or it leaks memory and stale answers across inputs that share argument values.
- **❌ Common wrong answer:** "Top-down is always slower" or "they have different time complexity" — both visit each state once; they differ in constants, space and recursion depth.

</details>

**Q18. Where does dynamic programming show up in AI/ML systems?**

<details><summary>Show answer</summary>

- **30-second answer:** Edit distance powers WER/CER metrics and fuzzy matching; the Viterbi algorithm decodes the most likely state sequence in HMMs and CRFs (named-entity tagging); CTC loss in speech recognition sums over alignments with a forward DP; dynamic time warping aligns time series; and reinforcement learning's Bellman equation is DP over states (value iteration).
- **Go deeper:** Many of these run on GPUs as batched tensor recurrences (CTC in PyTorch's `F.ctc_loss`). Sequence alignment in bioinformatics (Needleman–Wunsch, Smith–Waterman) is edit distance with scores. Beam search is *not* DP — it's pruned search, and can miss the optimum that Viterbi would find for models where Viterbi applies.
- **❌ Common wrong answer:** "DP is only for coding interviews."

</details>

## 9. Greedy & Bit Manipulation 🟡

### Greedy

A **greedy** algorithm makes the choice that looks best *right now* and never reconsiders it. It's usually simpler and faster than DP — but it's only correct when you can argue that a locally best choice never blocks the global best (an **exchange argument**: "any optimal solution can be changed to include my greedy choice without getting worse"). When you can't argue that, test a small counterexample:

````python
def greedy_coins(coins, amount):
    count = 0
    for c in sorted(coins, reverse=True):           # always take the biggest coin that fits
        count += amount // c
        amount %= c
    return count if amount == 0 else -1


def dp_coins(coins, amount):
    dp = [0] + [float("inf")] * amount
    for a in range(1, amount + 1):
        dp[a] = min((dp[a - c] + 1 for c in coins if c <= a), default=float("inf"))
    return dp[amount]


for coins, amount in [([1, 5, 10, 25], 63), ([1, 3, 4], 6)]:
    print(f"coins {coins}, amount {amount}: greedy uses {greedy_coins(coins, amount)}, DP (optimal) uses {dp_coins(coins, amount)}")
````

US coins happen to be a "canonical" system where greedy is optimal; `[1, 3, 4]` isn't (greedy takes 4 + 1 + 1, the best is 3 + 3).

- **Bit manipulation:** "every element appears twice except one", "count set bits", "subsets as bitmasks", "O(1) extra space" puzzles, powers of two.

### 🧪 Worked example — Jump Game (LeetCode 55)

**Problem:** `nums[i]` is the maximum jump length from index i. Starting at index 0, can you reach the last index?

**DP (brute force):** `reachable[i]` = any earlier reachable j with `j + nums[j] >= i` — O(n²). **Greedy:** track the farthest index reachable so far; if you ever stand beyond it, you're stuck — O(n).

````python
def can_jump_dp(nums):
    reachable = [False] * len(nums)
    reachable[0] = True
    for i in range(1, len(nums)):
        reachable[i] = any(reachable[j] and j + nums[j] >= i for j in range(i))
    return reachable[-1]


def can_jump(nums):
    farthest = 0
    for i, jump in enumerate(nums):
        if i > farthest:
            return False
        farthest = max(farthest, i + jump)
    return True


nums = [2, 3, 1, 1, 4]
farthest = 0
print(f"{'i':>2} {'nums[i]':>7} {'farthest before':>15} {'farthest after':>14}")
for i, jump in enumerate(nums):
    before = farthest
    farthest = max(farthest, i + jump)
    print(f"{i:>2} {jump:>7} {before:>15} {farthest:>14}")
for case in ([2, 3, 1, 1, 4], [3, 2, 1, 0, 4], [0], [1, 0, 1, 0]):
    assert can_jump(case) == can_jump_dp(case)
print("[3, 2, 1, 0, 4] →", can_jump([3, 2, 1, 0, 4]), "(everything leads to the 0 at index 3)")

long_case = [1] * 5_000
print(f"n = 5,000: DP {best_time(can_jump_dp, long_case, repeat=1) * 1e3:.0f} ms | greedy {best_time(can_jump, long_case) * 1e3:.2f} ms")
````

**Why greedy is safe here:** the set of reachable indices is always a prefix `[0, farthest]` — if you can reach index k, you can reach every index before it. So one number describes the whole state, and updating it can never hurt.

```python
def serialize(root):
    tokens, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node is None:
            tokens.append("#")                        # a missing child keeps the shape unambiguous
        else:
            tokens.append(str(node.val))
            queue.append(node.left)
            queue.append(node.right)
    return ",".join(tokens)


def deserialize(data):
    tokens = data.split(",")
    if tokens[0] == "#":
        return None
    root = TreeNode(int(tokens[0]))
    parents, i = deque([root]), 1
    while parents:
        parent = parents.popleft()
        for side in ("left", "right"):                # the next two tokens are this parent's children
            if tokens[i] != "#":
                child = TreeNode(int(tokens[i]))
                setattr(parent, side, child)
                parents.append(child)
            i += 1
    return root
```

```python
def pacific_atlantic(heights):
    rows, cols = len(heights), len(heights[0])

    def reachable(starts):
        seen = set(starts)
        queue = deque(starts)
        while queue:
            r, c = queue.popleft()
            for nr, nc in ((r + 1, c), (r - 1, c), (r, c + 1), (r, c - 1)):
                if (0 <= nr < rows and 0 <= nc < cols and (nr, nc) not in seen
                        and heights[nr][nc] >= heights[r][c]):     # uphill in reverse = downhill for the water
                    seen.add((nr, nc))
                    queue.append((nr, nc))
        return seen

    pacific = reachable([(0, c) for c in range(cols)] + [(r, 0) for r in range(1, rows)])
    atlantic = reachable([(rows - 1, c) for c in range(cols)] + [(r, cols - 1) for r in range(rows - 1)])
    return [[r, c] for r, c in pacific & atlantic]


check("B1 Pacific Atlantic Water Flow", pacific_atlantic, PACIFIC_TESTS, same=any_order, large=PACIFIC_LARGE)
```

**Complexity:** O(R · C) time and space — each BFS visits every cell at most once. The reversal trick ("search from the destinations instead of from every source") is worth naming; it also solves "Surrounded Regions" and "walls and gates".

<details><summary>💡 Show solution</summary>

```python
def ladder_length(begin_word, end_word, word_list):
    words = set(word_list)
    if end_word not in words:
        return 0
    buckets = defaultdict(list)                        # "h*t" → every word matching that pattern
    for word in words:
        for i in range(len(word)):
            buckets[word[:i] + "*" + word[i + 1:]].append(word)
    visited = {begin_word}
    queue = deque([(begin_word, 1)])
    while queue:
        word, length = queue.popleft()
        if word == end_word:
            return length
        for i in range(len(word)):
            pattern = word[:i] + "*" + word[i + 1:]
            for neighbour in buckets.pop(pattern, []):  # pop: each bucket only needs to be expanded once
                if neighbour not in visited:
                    visited.add(neighbour)
                    queue.append((neighbour, length + 1))
    return 0


check("B2 Word Ladder", ladder_length, LADDER_TESTS, large=LADDER_LARGE)
```

**Complexity:** O(n · L²) time and space for n words of length L (n · L patterns, each built in O(L)). Popping a bucket after using it prevents re-scanning big buckets from many words. Every edge has weight 1, so plain BFS gives the shortest sequence.
</details>

**Q21. How would you speed up Word Ladder further, or return the actual path?**

<details><summary>Show answer</summary>

- **30-second answer:** **Bidirectional BFS**: expand from both ends and always grow the smaller frontier; stop when they meet. With branching factor b and distance d, that explores about 2·b^(d/2) nodes instead of b^d. To return a path, store each word's parent when you first reach it and walk back from the end; for *all* shortest paths (Word Ladder II), BFS first to build the layered parent graph, then backtrack over it.
- **Go deeper:** Precomputing the pattern buckets is the adjacency list of an implicit graph — the same idea as bucketing embeddings by LSH hash to avoid all-pairs comparisons. If some transformations cost more (weighted edges), switch to Dijkstra.
- **❌ Common wrong answer:** "DFS and keep the minimum length" — exponential, because DFS explores every path.

</details>

<details><summary>💡 Show solution</summary>

```python
def least_interval(tasks, n):
    counts = Counter(tasks)
    top = max(counts.values())                                     # the most frequent task sets the skeleton
    tied = sum(1 for c in counts.values() if c == top)             # tasks sharing that highest count fill the last row
    frame = (top - 1) * (n + 1) + tied                             # A _ _ A _ _ A (with the tied tasks after the last A)
    return max(len(tasks), frame)                                  # extra variety fills gaps; it never makes it shorter


check("C1 Task Scheduler", least_interval, TASK_SCHEDULER_TESTS, large=TASK_SCHEDULER_LARGE)
```

**Complexity:** O(len(tasks)) time, O(1) space (at most 26 labels). The heap simulation — each cycle of n + 1 slots, pop the most frequent available tasks, push back the decremented counts — is O(len(tasks) · log 26) if it jumps over idle stretches, and it generalizes to "return the actual schedule". A tick-by-tick simulation is O(answer), 10 million steps on the large test.
</details>

<details><summary>💡 Show solution</summary>

```python
def longest_increasing_path(matrix):
    if not matrix or not matrix[0]:
        return 0
    rows, cols = len(matrix), len(matrix[0])
    memo = {}

    def longest(r, c):                                      # longest increasing path that STARTS at (r, c)
        if (r, c) not in memo:
            best = 1
            for nr, nc in ((r + 1, c), (r - 1, c), (r, c + 1), (r, c - 1)):
                if 0 <= nr < rows and 0 <= nc < cols and matrix[nr][nc] > matrix[r][c]:
                    best = max(best, 1 + longest(nr, nc))
            memo[(r, c)] = best
        return memo[(r, c)]

    return max(longest(r, c) for r in range(rows) for c in range(cols))


check("C2 Longest Increasing Path in a Matrix", longest_increasing_path, LIP_TESTS, large=LIP_LARGE)
```

**Complexity:** O(R · C) time and space — each cell's answer is computed once from at most 4 neighbours. Recursion depth equals the longest path (299 here); for a pathological 200 × 200 snake (40,000 deep) use the iterative alternative: a topological sort of cells by value (Kahn's algorithm from local maxima), where the answer is the number of BFS layers.
</details>

**Q22. Why doesn't Longest Increasing Path need a `visited` set, and how is it related to topological sorting?**

<details><summary>Show answer</summary>

- **30-second answer:** Edges only go from a smaller value to a strictly larger one, so the implied graph is a **DAG** — a path can't come back to a cell it left, and there are no cycles to guard against. That's also why memoization is valid: the longest path from a cell doesn't depend on how you arrived there. Longest path in a DAG = DP in topological order.
- **Go deeper:** In a general graph with cycles, "longest simple path" is NP-hard; the DAG structure is what makes it O(V + E). The same reasoning powers scheduling with dependencies (critical path method) and DP over a pipeline DAG to find its slowest chain of steps.
- **❌ Common wrong answer:** "Add a visited set and do plain DFS from every cell" — correct, but exponential without memoization.

</details>

### Turning a recursion into dynamic programming

1. Write the brute-force recursion with the **fewest arguments** that determine the answer (indices, remaining capacity) — those arguments are the **state**.
2. Check that the same arguments repeat (draw part of the call tree).
3. Add `@functools.cache` → **top-down DP**. Complexity = states × work per state.
4. For **bottom-up**, make a table indexed by the state, fill the **base cases**, and loop so that each state's dependencies are already filled.
5. Point to where the **answer** lives (`dp[n]`, `max(dp)`, `dp[n][m]`).
6. If a row only reads the previous row, keep **two rows or two variables** (space optimization) — and loop capacities *downwards* for 0/1 choices.

### Graph algorithm selection

| Algorithm | Use it when | Time | Space | Watch out for |
|---|---|---|---|---|
| BFS | Unweighted shortest paths, levels, multi-source spreading | O(V + E) | O(V) | Mark visited when enqueuing |
| DFS (iterative or recursive) | Components, flood fill, cycle detection, path existence | O(V + E) | O(V) | Python's recursion limit |
| 0-1 BFS (deque) | Edge weights are only 0 or 1 | O(V + E) | O(V) | 0-weight edges go to the front |
| Dijkstra (heap) | Single-source shortest paths, weights ≥ 0 | O(E log V) | O(V + E) | Wrong with negative edges; skip stale heap entries |
| Bellman–Ford | Negative weights allowed; detect negative cycles | O(V · E) | O(V) | Slow for large graphs |
| Topological sort (Kahn / DFS) | Ordering a DAG; shortest/longest paths in a DAG | O(V + E) | O(V + E) | Leftover nodes mean a cycle |
| Union–find | Connectivity as edges are added; Kruskal's MST; undirected cycle check | ≈ O(α(V)) per operation | O(V) | No edge deletions, no paths |
| Floyd–Warshall | All-pairs shortest paths on small graphs | O(V³) | O(V²) | Only for a few hundred nodes |
| A* search | One target with a heuristic that never overestimates (maps, games) | ≤ Dijkstra in practice | O(V) | A bad heuristic makes it Dijkstra (or wrong, if it overestimates) |
| Prim / Kruskal | Minimum spanning tree of a weighted undirected graph | O(E log V) | O(V + E) | Undirected graphs only |
