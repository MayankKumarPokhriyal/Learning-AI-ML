## 🎯 How This Round Works

A **behavioral question** asks about something you actually did in the past: *"Tell me about a time you…"*. The idea behind it is simple: **how you handled real situations before is the best available predictor of how you will handle them again.** A **project deep dive** is the technical cousin: you describe one project and the interviewer keeps digging until they find the edge of what you know.

Think of it like a **cooking show judge**. The judge does not want the recipe you read online; they want to taste *your* dish and ask why you added each ingredient. Anyone can recite "I value teamwork" — the interviewer is listening for a specific moment, what *you* did, and what happened.

### Common formats

| Format | Who runs it | What it looks like | Typical length |
|---|---|---|---|
| **Recruiter / behavioral screen** | Recruiter or HR partner | "Tell me about yourself", "Why this role?", 1–2 light behavioral questions, logistics | 20–30 min |
| **Hiring-manager round** | Your future manager | Career story, motivation, 3–5 behavioral questions, how you work with people and handle ambiguity | 45–60 min |
| **Project deep dive** | A senior engineer or scientist | One project for the whole session: problem → data → metric → model → failures → deployment → "what would you change?" | 45–60 min |
| **Dedicated behavioral round** | A trained interviewer (e.g. Amazon's Leadership Principles questions) | Several "tell me about a time" questions with aggressive follow-ups | 45–60 min |
| **Behavioral slice inside a technical round** | Any interviewer | 5–10 minutes at the start: "tell me about a project you're proud of" | 5–10 min |

### What interviewers score

Most companies use a written rubric. The labels differ ("Leadership Principles", "values", "competencies") but they boil down to these:

| Signal | What "strong" sounds like | Red flags |
|---|---|---|
| **Ownership** | "I noticed X was broken, nobody owned it, so I…" | Everything was "we"; or "that wasn't my job" |
| **Impact** | A concrete result, ideally a number and who benefited | "It went well"; no result at all |
| **Collaboration** | Credits others precisely, explains how they worked together | Takes credit for the team's work — or can't say what *they* did |
| **Handling conflict** | Disagreed with data and respect, reached a decision, relationship intact | Blames or belittles a colleague; "I've never had a conflict" |
| **Handling ambiguity** | Clarified the goal, made assumptions explicit, shipped something small first | Waited for perfect requirements; or charged ahead without checking |
| **Learning from failure** | Owns the mistake, explains the fix *and* what changed afterwards | A disguised success ("I worked too hard"); blames circumstances |
| **Communication with non-technical people** | Translates metrics into business terms, uses examples, checks understanding | Jargon dumps; "they just didn't get it" |
| **Judgement & ethics** | Raises risks (bias, privacy, misuse) early and proposes options | Ships anything the metric rewards |

### Red flags that sink otherwise good candidates

- **Invented or borrowed stories.** Follow-ups ("what exactly did the dashboard show?") expose them within two questions — and it is dishonest. Every example answer in this notebook is **clearly labelled as an example**; use it to learn the *shape*, never as your content.
- **Inflated metrics.** "99% accuracy" on an imbalanced dataset, or "improved revenue by 40%" when you cannot explain how it was measured.
- **No "I".** The interviewer can't hire your team. Say "we" for context and "I" for your actions.
- **Rambling.** A 6-minute answer with the result buried at the end. Aim for **~2 minutes**, then let them ask follow-ups.
- **Negativity** about past employers, teammates, or interviewers.
- **No reflection.** Every story should end with what you learned or would do differently.

## 1. Answer Structures: STAR and CAR/CARL 🟢

Without a structure, stories wander: background for three minutes, then "…and it worked out". A structure is a **coat hanger** — the same shape every time, so the listener always knows where they are.

| Format | Parts | Best for |
|---|---|---|
| **STAR** | **S**ituation → **T**ask → **A**ction → **R**esult | The default. Most interviewers and many company prep guides use these words |
| **CAR** | **C**ontext (situation + task in one) → **A**ction → **R**esult | Shorter answers, phone screens, résumé bullets |
| **CARL** | **C**ontext → **A**ction → **R**esult → **L**earning | Failure and conflict questions, where reflection is half the score |

**Time budget for a ~2-minute answer:**

```
Situation + Task  ███░░░░░░░░░░░░░░░░░   ~15–20%   "just enough context to follow"
Action            ████████████░░░░░░░░   ~50–60%   "what YOU did, step by step, and why"
Result            ███░░░░░░░░░░░░░░░░░   ~15%      "what changed — a number if you have one"
Learning          ██░░░░░░░░░░░░░░░░░░   ~10%      "what you'd repeat or do differently"
```

The most common mistake is flipping that budget: long situation, one-line action, no result.

### Weak vs strong: the same story, rewritten

*Both versions below are an **illustrative example** written for this notebook — not a real project.*

**❌ Weak** (question: *"Tell me about a time you improved something"*)

> "So at my internship we had this forecasting pipeline and it was really slow, like really slow, and everyone complained about it. We looked into it and there were a lot of problems with the code. We ended up refactoring a lot of things and using better pandas functions, and it was a team effort. In the end it was much faster and people were happy. I learned a lot about pandas."

**✅ Strong (STAR + a learning)**

> **Situation:** "During my internship, the team's weekly sales-forecast pipeline took about six hours, so analysts received numbers a day late."
> **Task:** "My lead asked me to get it under an hour without changing the forecasts."
> **Action:** "I profiled it first instead of guessing, and found most of the time was in a row-by-row `apply`. I rewrote that step as a vectorized group-by. Because the output fed finance reports, I also wrote a regression test that compared old and new outputs on three months of history before switching over, and I walked the analyst who owned the report through the diff."
> **Result:** "The run dropped to about 40 minutes, outputs matched exactly, and analysts got forecasts the same morning."
> **Learning:** "Profiling before optimizing saved me from rewriting the parts that weren't slow — I now profile first by default."

Why the second one wins: a clear goal and constraint, "I" for actions, a *reason* behind each action, a verification step, a concrete result, and a reflection.

A quick, transparent signal you can compute on your own written drafts: **how often you say "I" vs "we"**, and whether the story contains any number at all. It is a *nudge*, not a grade — a great teamwork story legitimately has plenty of "we".

````python
WEAK_STORY = (
    "So at my internship we had this forecasting pipeline and it was really slow, like really slow, and everyone "
    "complained about it. We looked into it and there were a lot of problems with the code. We ended up refactoring "
    "a lot of things and using better pandas functions, and it was a team effort. In the end it was much faster and "
    "people were happy. I learned a lot about pandas."
)
STRONG_STORY = (
    "During my internship, the team's weekly sales-forecast pipeline took about six hours, so analysts received "
    "numbers a day late. My lead asked me to get it under an hour without changing the forecasts. I profiled it first "
    "instead of guessing, and found most of the time was in a row-by-row apply. I rewrote that step as a vectorized "
    "group-by. Because the output fed finance reports, I also wrote a regression test that compared old and new "
    "outputs on three months of history before switching over, and I walked the analyst who owned the report through "
    "the diff. The run dropped to about 40 minutes, outputs matched exactly, and analysts got forecasts the same "
    "morning. Profiling before optimizing saved me from rewriting the parts that weren't slow."
)

NUMBER_WORDS = {"one", "two", "three", "four", "five", "six", "seven", "eight", "nine", "ten", "twice", "half", "hundred", "thousand"}


def story_signals(text):
    """Count ownership pronouns, filler words, and concrete quantities in a written story draft."""
    words = re.findall(r"[a-z']+|\d+", text.lower())
    counts = Counter(words)
    return {
        "words": len(words),
        "I / my": counts["i"] + counts["my"] + counts["me"],
        "we / our": counts["we"] + counts["our"] + counts["us"],
        "fillers (so, like, really, a lot)": counts["so"] + counts["like"] + counts["really"] + text.lower().count("a lot"),
        "quantities": sum(1 for w in words if w.isdigit() or w in NUMBER_WORDS),
    }


pd.DataFrame({"weak": story_signals(WEAK_STORY), "strong": story_signals(STRONG_STORY)})
````

The table is computed from the two texts above: the strong version is longer but carries its weight in *actions* ("I profiled…", "I rewrote…", "I walked…") and concrete quantities, while the weak version leans on "we" and fillers.

## 2. Building a Story Bank 🟢

You cannot invent a good story in 10 seconds under pressure. So you prepare a **story bank**: 8–10 real experiences written down *once*, each tagged with the competencies it can demonstrate. In the interview you just pick the best-fitting story — like a musician choosing from a practised set list instead of composing on stage.

**Where stories come from** (early-career and career-switchers have more than they think):
- internships, jobs (including non-tech jobs — conflict and deadlines are universal), research, teaching assistant work
- course capstones and personal projects — especially the parts that *went wrong*
- open-source contributions, hackathons, Kaggle competitions, study groups, volunteering

### The template

Write each story as one row. Keep the numbers you remember **exactly as they were** — if you don't know a number, write what you *do* know ("the job went from failing weekly to not failing in the two months I was there").

| Field | What to write | Example (illustrative) |
|---|---|---|
| **Title** | A 3–6 word handle you'll remember | "Silent zero-filled column" |
| **Context** | 1–2 sentences: where, when, what was at stake | "Internship; churn scores fed a weekly retention campaign" |
| **Your actions** | 3–5 bullets of what *you* did and why | "Compared feature distributions; found renamed column; added schema check" |
| **Result** | What changed, with a number if one exists | "Campaign ran on time; check caught 2 later upstream changes" |
| **Learning** | What you'd repeat or change | "Validate inputs at pipeline boundaries, not just model outputs" |
| **Competencies** | Which question types it can answer | ownership, incomplete data, learning from failure |

### Reusing one story for several questions

One strong story can answer 3–4 different questions **by changing the emphasis**, not the facts:

| Question | Same story, different spotlight |
|---|---|
| "Tell me about a time you took ownership" | Nobody owned data validation; you noticed and fixed it |
| "Tell me about a mistake" | You shipped the pipeline without input checks in the first place |
| "A decision with incomplete data" | You had to decide by Friday whether to trust the scores |
| "Explaining something technical" | How you explained "silent zero-filling" to the retention manager |

Aim for **every competency covered by at least two stories**, so you never have to use the same story twice with the same interviewer panel.

### The story-bank coverage checker

The code below turns a story bank into a **coverage matrix** and reports gaps. The competency list matches the question groups in [section 5](#5.-Common-Behavioral-Questions-🟡). The example bank uses **placeholder stories** — replace them with yours.

````python
COMPETENCIES = [
    "ownership", "impact", "collaboration", "conflict", "ambiguity", "failure",
    "non_technical_communication", "learning_quickly", "ethics_fairness",
    "incomplete_data_decision", "pushing_back", "deadline_pressure",
]

# Placeholder bank — titles describe the *kind* of story, not real events. Replace every row with your own.
example_bank = [
    {"title": "Speeding up a slow data pipeline", "competencies": ["ownership", "impact"]},
    {"title": "Silent zero-filled column in production", "competencies": ["ownership", "incomplete_data_decision", "deadline_pressure"]},
    {"title": "Disagreement over the evaluation metric", "competencies": ["conflict", "pushing_back", "collaboration"]},
    {"title": "Vague 'add AI search' request", "competencies": ["ambiguity", "non_technical_communication"]},
    {"title": "Learning PyTorch for a hackathon", "competencies": ["learning_quickly", "deadline_pressure"]},
    {"title": "Explaining precision vs recall to ops", "competencies": ["non_technical_communication", "impact"]},
    {"title": "Group capstone with an absent teammate", "competencies": ["collaboration", "conflict"]},
    {"title": "Choosing a model with only 3 weeks of data", "competencies": ["incomplete_data_decision", "ambiguity", "learning_quickly"]},
]


def coverage_report(bank, competencies=COMPETENCIES, min_stories=2):
    """Return a coverage matrix plus gaps (0 stories), thin spots (< min_stories) and unknown tags (likely typos)."""
    known = set(competencies)
    unknown = sorted({tag for story in bank for tag in story["competencies"] if tag not in known})
    matrix = pd.DataFrame(
        [["✓" if comp in story["competencies"] else "" for comp in competencies] for story in bank],
        index=[story["title"] for story in bank], columns=competencies,
    )
    counts = {comp: sum(comp in story["competencies"] for story in bank) for comp in competencies}
    gaps = [comp for comp, n in counts.items() if n == 0]
    thin = [comp for comp, n in counts.items() if 0 < n < min_stories]
    return {"matrix": matrix, "counts": counts, "gaps": gaps, "thin": thin, "unknown_tags": unknown}


report = coverage_report(example_bank)
print("❌ gaps (no story at all):", report["gaps"])
print("⚠️ thin (only one story) :", report["thin"])
print("❓ unknown tags (typos?)  :", report["unknown_tags"])
report["matrix"]
````

Read the output: the example bank has **no story** for `failure` or `ethics_fairness` — two of the most frequently asked question types — and only one for `pushing_back`. That is precisely the kind of gap you want to find *before* the interview.

**Why a coverage check instead of just "more stories"?** Candidates naturally collect success stories. The checker forces you to find a real failure story and a real judgement-call story, which are the ones people freeze on.

## 3. The Project Deep Dive 🟡

In a deep dive the interviewer picks **one project** (usually the top one on your résumé) and spends 30–60 minutes on it. They want to know three things: *Did you really do this? Do you understand why each decision was made? Can you reason about what you'd do differently?*

### The 2-minute pitch

Open with a crisp overview so the interviewer can choose where to dig. Six beats, in order:

| # | Beat | Question it answers | Share of 2 min |
|---|---|---|---|
| 1 | **Problem** | What was the goal and who cared? | ~15 s |
| 2 | **Constraints** | Data size, labels, latency, budget, deadline, privacy | ~15 s |
| 3 | **Approach** | Baseline first, then the model/system you built and why | ~30 s |
| 4 | **Trade-offs** | What you chose *not* to do, and the cost of your choice | ~20 s |
| 5 | **Results with metrics** | Offline metric vs a baseline, and any real-world effect | ~20 s |
| 6 | **What you'd change** | Honest limitations and next steps | ~20 s |

People speak at very roughly **130–160 words per minute** in conversation, so a 2-minute pitch is about **250–300 words**. The helper below estimates speaking time from a written draft (at 140 wpm by default — time yourself to find your own pace).

````python
def speaking_seconds(text, words_per_minute=140):
    """Estimate how long a written answer takes to say out loud."""
    n_words = len(re.findall(r"\S+", text))
    return n_words, round(n_words / words_per_minute * 60)


EXAMPLE_PITCH = """\
Problem: a subscription business wanted to know which customers were likely to cancel next month, so the retention team could call the riskiest ones first.
Constraints: about seven thousand customers with a churn rate near one in four, tabular data only, and the team could only call a few hundred people a week, so ranking quality at the top mattered more than overall accuracy.
Approach: I started with a majority-class baseline and a logistic regression, then tried gradient boosting inside a scikit-learn pipeline so preprocessing was fitted on training folds only. I tuned on a validation split and touched the test set once.
Trade-offs: boosting was a little better than logistic regression but harder to explain, so I kept both and reported feature effects with permutation importance. I chose ROC-AUC plus precision in the top decile instead of accuracy, because accuracy looks great on imbalanced data even for a useless model.
Results: on the held-out test set the boosted model beat the logistic regression baseline on both metrics, and I served it behind a small FastAPI endpoint with input validation and a test suite.
What I'd change: the data is a single snapshot, so I can't measure drift; next I'd do a time-based split and estimate the value of a call against its cost to pick the threshold.
"""
n_words, seconds = speaking_seconds(EXAMPLE_PITCH)
verdict = "fits in 2 minutes ✅" if seconds <= 120 else "too long — cut it ✂️"
print(f"{n_words} words ≈ {seconds} s at 140 wpm → {verdict}")
````

*The pitch above is an illustrative example of the **shape**. It deliberately says "beat the baseline" instead of quoting numbers — your pitch must quote **your own** measured numbers (see section 4).*

### Handling drill-down questions

After the pitch, the interviewer digs. Almost every follow-up falls into one of these buckets — prepare one honest sentence for each **before** the interview:

| Bucket | Typical question | What they're checking | Prepare |
|---|---|---|---|
| **Data** | "Where did the labels come from? How clean were they?" | You looked at the data, not just `.fit()` | Size, source, label quality, one surprising thing you found |
| **Leakage & splits** | "How did you split? Could information leak?" | You understand evaluation validity | Split strategy and *why* (random / time / group) |
| **Metrics** | "Why that metric?" | You connected the metric to the goal | Metric ↔ business cost; why not accuracy |
| **Baselines** | "Compared to what?" | Your number means something | The simplest baseline and its score |
| **Failures** | "What didn't work?" | Honesty and learning | Two things you tried that failed, and why |
| **Scale** | "What if data were 100× bigger / latency 10× tighter?" | You can reason past the prototype | The first bottleneck and one concrete fix |
| **Ownership** | "Which parts did *you* do?" | Team project vs your contribution | A precise split of who did what |

**The golden rule:** if you don't know, say so and reason out loud — *"I didn't measure that. My guess is X because Y, and I'd check it by Z."* Interviewers respect that far more than a confident guess they can disprove.

- **30-second answer:** Tie the metric to the decision the model supports. *Example:* "The team could only act on the top few hundred predictions each week, so I optimized ranking at the top — precision in the top decile — and tracked ROC-AUC as a threshold-free summary. - **Go deeper:** Mention the cost asymmetry (false negatives vs false positives), how you picked a threshold on the validation set, and what online metric would confirm the offline one.
- **❌ Common wrong answer:** "Accuracy, because it's the standard metric" — on imbalanced data a model that never predicts the minority class can look excellent.

</details>

**Q3. "What was your baseline, and how much better was your model?"**

<details><summary>Show answer</summary>

- **30-second answer:** Name the simplest reasonable baseline (majority class, mean prediction, a rule, logistic regression, BM25 for search) and state both numbers on the **same** held-out data. If the gain was small, say so and explain why the extra complexity was or wasn't worth it.
- **Go deeper:** Mention variance: with a small test set, a 1-point difference may be noise — cross-validation spread or a bootstrap interval shows whether the gain is real.
- **❌ Common wrong answer:** "I didn't have a baseline, but 87% is good." A number without a comparison can't be judged.

</details>

**Q4. "What didn't work?"**

<details><summary>Show answer</summary>

- **30-second answer:** Give one or two genuine dead ends, what you learned from each, and how that changed your approach. *Example:* "A deeper network overfit badly on 2,000 rows; the validation loss told me within a few epochs, so I went back to gradient boosting and spent the time on features instead."
- **Go deeper:** The best failure answers show a **fast feedback loop** — you noticed quickly because you had a validation set, a baseline, and logging.
- **❌ Common wrong answer:** "Everything worked as planned." It sounds either untrue or like the project was too easy.

</details>

**Q5. "How would this work at 100× the data or traffic?"**

<details><summary>Show answer</summary>

- **30-second answer:** Identify the first bottleneck rather than listing technologies: *"Training on one machine still fits at 100× for this tabular model, but feature computation in pandas wouldn't — I'd move it to SQL/Spark. For serving, batch scoring nightly is cheaper than an API if predictions don't need to be real-time."*
- **Go deeper:** Discuss what changes besides compute: drift monitoring, retraining schedule, data validation, latency budgets, cost per prediction, and for LLM apps caching and token cost.
- **❌ Common wrong answer:** "Put it on Kubernetes" with no bottleneck analysis.

</details>

### ⚠️ The one rule: your numbers, or no numbers

Everything in `{curly braces}` below is a **blank you fill from your own run**. Never copy a number from a tutorial, from someone else's repo, or from this notebook. If you didn't measure something, say "I didn't measure that" — then say how you would.

To make that rule mechanical, `render_pitch()` fills a template from a dictionary of facts and **refuses to produce a pitch while any blank is missing** — it will never invent a value for you.

````python
def template_fields(template):
    """Names of all {placeholders} in a template, in order of first appearance."""
    seen = []
    for _, field, _, _ in string.Formatter().parse(template):
        if field and field not in seen:
            seen.append(field)
    return seen


def render_pitch(template, facts):
    """Fill the template only if every blank has a real value; otherwise list what is missing."""
    missing = [f for f in template_fields(template) if facts.get(f) in (None, "")]
    if missing:
        print(f"⏳ Not ready — measure/fill these first: {missing}")
        return None
    return template.format(**facts)


ML_PITCH = (
    "Problem: {who} needed to predict {target} so they could {decision}.\n"
    "Constraints: {n_rows} rows, {positive_rate} positive class, {constraint}.\n"
    "Approach: baseline {baseline_model}, then {final_model} in a leak-free pipeline, tuned on a validation split.\n"
    "Trade-offs: I chose {tradeoff_choice} over {tradeoff_alternative} because {tradeoff_reason}.\n"
    "Results: on the untouched test set, {metric} was {final_score} vs {baseline_score} for the baseline; served via {serving}.\n"
    "What I'd change: {next_step}."
)

print("Blanks in the ML pitch:", template_fields(ML_PITCH))
my_facts = {"who": "a retention team", "target": "which customers will cancel", "decision": "call the riskiest first",
            "baseline_model": "logistic regression", "final_model": "gradient boosting", "serving": "a FastAPI endpoint"}
pitch = render_pitch(ML_PITCH, my_facts)       # the qualitative parts are filled, the measured ones are not
````

That refusal is the point: the qualitative facts are known before you start, but `n_rows`, `positive_rate`, `final_score` and `baseline_score` only exist after **you** run the project. Paste this template into your own notes, fill it from your results, and time it with `speaking_seconds()` from section 3.

### 🧮 Capstone 01 — End-to-End ML Project

**Pitch skeleton:** *"{Who} needed to predict {target} to {decision}. The data had {n_rows} rows with {class balance / target distribution}. I built a {baseline} baseline, then a {final model} inside a scikit-learn pipeline so preprocessing was fitted only on training data. I picked {metric} because {why it matches the decision}. On the held-out test set it reached {your score} vs {baseline score}. I explained predictions with {method}, packaged the model behind {serving}, and wrote tests for {what}. **Likely follow-ups:**

2. Why this metric? What threshold would you use, and who decides it?

4. How much better than the baseline was it, and is that difference bigger than the noise?
5. What happens when the input distribution shifts next quarter?
6. How does the served model get the *same* preprocessing as training?

<details><summary>What good answers cover</summary>

- **Split & leakage:** stratified (or time-based if rows have dates) split made *before* any fitting; imputers/scalers/encoders inside the `Pipeline`; features that exist only after the outcome removed.
- **Metric & threshold:** metric tied to the cost of errors; threshold picked on validation data from a precision/recall trade-off, owned by the business team.
- **Feature importance:** permutation importance on held-out data rather than only built-in tree importances; a sanity check for suspiciously strong features.
- **Noise:** cross-validation spread or a bootstrap interval on the test metric.
- **Drift:** monitor input distributions and prediction rates; retrain on a schedule or trigger; see the drift notebook in module 09.
- **Train/serve skew:** save the whole fitted pipeline (not just the model) and test the API against training-time predictions.

</details>

### 🖼️ Capstone 02 — End-to-End DL Project

**Pitch skeleton:** *"The goal was to classify {inputs} into {classes} for {use}. I had {n examples} labelled examples, split into train/validation/test. I started from {baseline — e.g. a small model or a pretrained backbone} and trained with {augmentation, optimizer, schedule}, using the validation set for early stopping and checkpoint selection. Test {metric} was {your score} vs {baseline}. The main errors were {confusion pattern}, which I found by {error analysis}. I exported the model with {export format} and served it via {serving}, measuring {latency} per request on {hardware}. Next I'd {improvement}."*

**Likely follow-ups:**
1. Why a pretrained model (or why not)? What did you freeze and why?
2. How did you know when to stop training? Show me an overfitting sign in your curves.
3. What does the confusion matrix say — which classes get mixed up and why?
4. Which augmentations did you use, and could any of them change the label?
5. What would you do if the model had to run on a phone or a CPU-only server?
6. How did you make the run reproducible?

<details><summary>What good answers cover</summary>

- **Transfer learning:** little data → pretrained backbone, freeze early layers first, then fine-tune with a lower learning rate.
- **Stopping:** validation loss rising while training loss falls = overfitting; early stopping + keeping the best checkpoint; the test set only once at the end.
- **Error analysis:** looking at actual misclassified examples, not only aggregate accuracy; per-class metrics for imbalance.
- **Augmentation validity:** e.g. horizontal flips are wrong for digits or text-like images where orientation carries meaning.
- **Deployment constraints:** smaller architecture, quantization, ONNX/TorchScript-style export, batch size and latency measured on the target hardware.
- **Reproducibility:** seeds, pinned versions, logged configs and metrics (MLflow or W&B from module 09).

</details>

### 📚 Capstone 03 — LLM RAG Assistant Project

**Pitch skeleton:** *"{Users} needed answers from {documents}, and {pain today}. I built a retrieval-augmented assistant: documents are chunked {chunking strategy}, embedded with {embedding model}, retrieved with {retrieval — e.g. hybrid keyword + vector}, and the model answers with citations. I wrote an evaluation set of {n} questions and measured retrieval ({recall@k / MRR}) separately from answer quality ({faithfulness / correctness method}). Compared with {baseline}, {your metric change}. The biggest failure mode was {failure}, which I addressed with {fix}. It runs on {model / hosting} at {latency or cost per query}. Next I'd {improvement}."*

**Likely follow-ups:**
1. How did you choose chunk size and overlap? What happened when you changed it?
2. How do you know an answer is grounded in the retrieved text rather than invented?
3. How did you build the evaluation set, and how big is it? Is it biased toward easy questions?
4. Why hybrid retrieval (or why not)? When does keyword search beat embeddings?
5. What does a query cost, and how would you cut latency and cost by 5×?
6. How do you handle prompt injection hidden inside a document?

<details><summary>What good answers cover</summary>

- **Chunking:** a measured comparison on your eval set, not a default copied from a tutorial; the trade-off between precise citations (small chunks) and complete context (large chunks).
- **Grounding:** citations required, an "I don't know" path when retrieval scores are low, and a faithfulness check (human spot checks or an LLM judge validated against human labels).
- **Eval set:** written *before* tuning, covering hard cases (multi-hop, no-answer, near-duplicate documents); retrieval and generation evaluated separately so you know which half failed.
- **Hybrid search:** keywords win on exact IDs, error codes and rare names; embeddings win on paraphrases; rank fusion combines them.
- **Cost & latency:** caching, smaller models for easy queries, fewer/shorter retrieved chunks, streaming for perceived latency.
- **Injection:** treat retrieved text as data, not instructions; restrict what the model can do; test with adversarial documents.

</details>

### 🤖 Capstone 04 — AI Agent Project

**Pitch skeleton:** *"The task was {multi-step task} for {users}, which previously took {manual effort}. I built an agent that plans and calls {n tools — list them} through {framework / plain tool-calling loop}. Guardrails: {input validation, allowed actions, step limits, human approval for risky actions}. I evaluated it on {n} scripted tasks, measuring {task success rate, steps, cost, failure categories}; versus {baseline — e.g. a single prompt or fixed workflow}, {your result}. The most common failure was {failure}, fixed by {fix}. Every run is traced with {tracing}, so I can replay failures. Next I'd {improvement}."*

**Likely follow-ups:**
1. Why an agent at all — would a fixed workflow have been simpler and more reliable?
2. What stops the agent from looping forever or taking a harmful action?
3. How do you evaluate something non-deterministic? How many runs per task?
4. What happens when a tool fails, times out, or returns malformed output?
5. How do you debug a bad run?
6. What does one task cost in tokens and time, and how would you reduce it?

<details><summary>What good answers cover</summary>

- **Agent vs workflow:** use a fixed pipeline when the steps are known; an agent only when the path genuinely depends on intermediate results. Saying "I started with a workflow and only added autonomy where it helped" is a strong answer.
- **Safety:** max steps/time/budget, an allow-list of tools, schema-validated arguments, read-only by default, human confirmation for irreversible actions.
- **Evaluation:** a fixed task suite with programmatic success checks, several runs per task to report a success *rate*, failure categories counted, not anecdotes.
- **Tool errors:** retries with backoff for transient errors, structured error messages returned to the model, graceful give-up.
- **Debugging:** traces of every model call and tool call with inputs/outputs; replay a failing trace after a fix.
- **Cost:** fewer steps via better tool descriptions, smaller models for simple sub-steps, caching tool results.

</details>

## 6. Writing Résumé Bullets for ML Projects 🟢

A résumé bullet is a **one-line trailer** for a story you are ready to tell in full. Recruiters skim; engineers who read it will pick a bullet and ask you about it. A widely used shape is:

> **Action verb** + **what you built, with the method** + **measurable result** (compared with something)

| Part | ❌ Weak | ✅ Better |
|---|---|---|
| Action | "Responsible for models" | "Built", "Cut", "Designed", "Automated", "Diagnosed" |
| Method | "using AI" | "gradient boosting in a leak-free scikit-learn pipeline" |
| Result | "improved performance" | "ROC-AUC from {baseline} to {yours} on a held-out test set" |

### Common mistakes

1. **Tools without outcomes** — "Python, PyTorch, Docker, AWS" belongs in a skills line, not a project bullet.
2. **Weak openers** — "Responsible for", "Worked on", "Helped with" hide what *you* did.
3. **Unverifiable claims** — "significantly improved", "state-of-the-art", "various models".
4. **Inflated or context-free metrics** — "99% accuracy" on a dataset that is 99% one class; "increased revenue 30%" when you cannot explain how it was measured or what share was yours.
5. **Walls of text** — more than ~2 lines per bullet gets skimmed past.

**Honesty test for every bullet:** *Could I defend this number for 10 minutes, including how it was measured, what it was compared against, and what I personally did?* If not, soften it to what you can defend ("on a 60-question test set I wrote") or remove it.

### The résumé-bullet checker

A small, transparent rule-based checker. Each rule is a few lines you can read — no model, no magic. It checks **shape, not truth**: it cannot tell whether a number is real. That part is on you.

| Rule | Flags a bullet when… |
|---|---|
| `no_metric` | there is no quantity — years (`2024`), names with digits (`GPT-4`, `YOLOv8`, `F1`, `p95`) and versions after a tool (`Python 3.12`) don't count |
| `weak_opener` | it starts with a passive phrase like "Responsible for" or "Worked on" |
| `tool_list` | it names ≥3 tools and either has no metric or is at least half tool names |
| `too_long` | it has more than 30 words |
| `check_claim` | it claims ≥99% accuracy/precision/recall/F1 — not necessarily wrong, but be ready to explain class balance and leakage checks |

````python
WEAK_OPENERS = ("responsible for", "worked on", "helped with", "helped", "assisted with", "assisted",
                "involved in", "participated in", "tasked with", "duties included", "in charge of", "exposure to")
TOOLS = {"python", "pytorch", "tensorflow", "keras", "scikit-learn", "sklearn", "pandas", "numpy", "polars", "sql",
         "postgres", "docker", "kubernetes", "aws", "gcp", "azure", "fastapi", "flask", "mlflow", "airflow", "dvc",
         "spark", "pyspark", "langchain", "langgraph", "llamaindex", "faiss", "xgboost", "lightgbm", "catboost", "git",
         "transformers", "huggingface", "openai", "opencv", "jupyter", "streamlit", "redis", "linux", "tableau", "excel"}
STOPWORDS = {"a", "an", "the", "and", "or", "of", "to", "in", "on", "for", "with", "using", "used", "via", "by",
             "from", "into", "at", "as", "&", "+"}
EDGE_PUNCT = ",.;:()[]\"'!?"


def clean_tokens(text):
    return [tok.strip(EDGE_PUNCT) for tok in text.split() if tok.strip(EDGE_PUNCT)]


def metric_tokens(bullet):
    """Tokens that look like real quantities (not years, versions, or model names)."""
    tokens = clean_tokens(bullet)
    found = []
    for i, tok in enumerate(tokens):
        if not any(ch.isdigit() for ch in tok):
            continue
        if re.fullmatch(r"(19|20)\d\d", tok):                      # a year
            continue
        if tok[0].isalpha() and not tok.endswith("%"):             # GPT-4, YOLOv8, F1, p95, recall@5 (but keep top-10%)
            continue
        if i > 0 and tokens[i - 1].lower() in TOOLS:               # Python 3.12, pandas 3.0
            continue
        found.append(tok)
    return found


def tool_mentions(bullet):
    return [tok for tok in clean_tokens(bullet) if tok.lower() in TOOLS]


def suspicious_claim(bullet):
    low = bullet.lower()
    for m in re.finditer(r"(\d+(?:\.\d+)?)\s?%", low):
        window = low[max(0, m.start() - 25): m.end() + 25]
        if float(m.group(1)) >= 99 and re.search(r"accura|precision|recall|f1", window):
            return True
    return False


def review_bullet(bullet, max_words=30):
    """Return a list of rule names the bullet breaks ([] means it passes every rule)."""
    text = bullet.strip().lstrip("-•*▪ ").strip()
    issues = []
    metrics = metric_tokens(text)
    if not metrics:
        issues.append("no_metric")
    if text.lower().startswith(WEAK_OPENERS):
        issues.append("weak_opener")
    tools = tool_mentions(text)
    content = [t for t in clean_tokens(text) if t.lower() not in STOPWORDS]
    if len(tools) >= 3 and (not metrics or len(tools) / len(content) >= 0.5):
        issues.append("tool_list")
    if len(text.split()) > max_words:
        issues.append("too_long")
    if suspicious_claim(text):
        issues.append("check_claim")
    return issues


# Illustrative bullets — the "strong" ones use made-up numbers purely to show the format
sample_bullets = [
    "Responsible for building machine learning models for the data science team.",
    "Python, PyTorch, Docker, FastAPI, MLflow, AWS",
    "Worked on a churn prediction project using Python 3.12, pandas and scikit-learn in 2024.",
    "Achieved 99.8% accuracy on credit-card fraud detection with XGBoost.",
    "Built a churn model in scikit-learn that raised ROC-AUC from 0.71 (logistic baseline) to 0.84 on a held-out test set.",
    "Cut p95 API latency from 800 ms to 120 ms by batching requests and caching embeddings in Redis.",
    "Designed and implemented an end-to-end machine learning pipeline including data ingestion, validation, feature "
    "engineering, hyperparameter tuning with Optuna, experiment tracking, containerisation, monitoring and deployment for the "
    "analytics team, reducing manual work by 10 hours per week.",
]
pd.DataFrame({"bullet": sample_bullets,
              "issues": [", ".join(review_bullet(b)) or "✅ passes" for b in sample_bullets],
              "quantities found": [metric_tokens(b) for b in sample_bullets]})
````

Reading the table: bullet 3 mentions `3.12` and `2024`, yet it is still flagged `no_metric` — neither is a result. Bullet 4 has a number but trips `check_claim`: fraud datasets are usually heavily imbalanced, so an interviewer's first question will be "what fraction of transactions were fraud?". The last bullet has a real result buried after 30+ words.

**Limits (say these if asked about the tool):** a rule-based checker misses synonyms it hasn't been told about, can't judge whether a verb is strong, and can't verify truth. It is a linter, like `flake8` for prose — useful for catching the obvious, useless as a judge of quality.

## 7. Questions to Ask and Follow-Up Etiquette 🟢

"Do you have any questions for us?" is part of the interview. Good questions show you're thinking about doing the job well — and the answers help *you* decide whether you want it.

| Topic | Example questions | What you learn |
|---|---|---|
| **The work** | "What does the path from a notebook experiment to production look like here?" · "How are models evaluated before launch, and who decides they're good enough?" | ML maturity, whether you'll ship |
| **The role** | "What would success look like for this role after six months?" · "What's a problem the team is stuck on right now?" | Expectations, real priorities |
| **The team** | "How are disagreements about technical direction resolved?" · "How do ML engineers and product managers work together?" | Culture, collaboration |
| **Growth** | "How do people learn new things here — reading groups, conference time, mentoring?" · "What does onboarding look like?" | Support for early-career people |
| **Data & ethics** | "How does the team handle model risks like bias or privacy reviews?" | Whether responsibility is real or a slogan |

**Avoid** questions answered on the first page of their website, and leave salary/benefits for the recruiter conversation unless the interviewer brings it up. Tailor to the person: ask an engineer about the stack and on-call, a manager about priorities and growth.

### Follow-up etiquette

- **Thank-you note:** a short email (3–4 sentences) within about a day, ideally mentioning one specific thing you discussed. It rarely changes a decision, but it's courteous and cheap.
- **Waiting:** ask about the timeline at the end of the interview. If it passes, one polite check-in with the recruiter is fine; repeated chasing is not.
- **Rejection:** reply graciously, ask if any feedback is available (many companies have policies against sharing it — don't push), and keep the door open. People who handled rejection well are often contacted again.
- **Confidentiality:** don't post the exact questions you were asked if you agreed not to share them, and never record an interview without explicit consent.
- **Honesty after the fact:** if you realise you said something wrong in the interview (e.g. a wrong formula), a brief correction in the thank-you note is fine and can come across well.

**Q37. The interviewer asks "Do you have any questions for me?" and you've run out of prepared ones. What do you say?**

- **30-second answer:** Ask about something *they* said during the interview: *"You mentioned the team retrains the ranking model weekly — what usually breaks when that happens?"* It shows you were listening and is always available.
- **Go deeper:** Keep one or two universal questions in reserve ("What do you enjoy most about working here?", "What would make someone really successful in their first six months?"). Ending with a clear, warm close ("Thanks — I enjoyed talking about X") matters too.
- **❌ Common wrong answer:** "No, I think you covered everything." It signals low interest.

## 🏋️ Timed Drills

Three mock sets. Do one set per sitting, **out loud**, ideally recorded on your phone or with a friend playing interviewer.

**How the 2-minute timer works:**
1. Read the question, then take **up to 10 seconds** to pick a story from your bank (you'd pause like this in a real interview too).
2. Start a **2-minute** timer and answer. Aim to finish the Result and Learning by roughly 1:45.
3. When the timer rings, stop — even mid-sentence. If you were cut off before the Result, that's the most important thing to fix.
4. Have your partner (or yourself, playing back the recording) ask **one follow-up** from the list and answer it in under 1 minute.
5. Score the answer with the rubric below *before* moving on.

### Set A — Hiring-manager round (warm-up)

| # | Question | Follow-up to expect |
|---|---|---|
| A1 | Tell me about yourself. | "Why did you leave / want to leave your current role?" |
| A2 | Tell me about a time you failed. | "What would you do differently today?" |
| A3 | Tell me about a time the requirements were unclear. | "How did you know the first version was good enough?" |
| A4 | Explain one of your projects to me as if I were a product manager. | "How would I know if it's working?" |
| A5 | Why this role? | "What would you want to learn in your first six months?" |

### Set B — Dedicated behavioral round (pressure)

| # | Question | Follow-up to expect |
|---|---|---|
| B1 | Tell me about a conflict with a teammate. | "What did they say exactly? What if they still disagreed?" |
| B2 | Tell me about a time you missed a deadline. | "When did you first know it was at risk?" |
| B3 | Tell me about a time you pushed back on a stakeholder. | "What would you have done if they had said no?" |
| B4 | Tell me about a time you raised an ethics or fairness concern. | "How did you measure the problem?" |
| B5 | Tell me about a decision you made with incomplete data. | "Was it the right decision? How do you know?" |

### Set C — Project deep dive (technical storytelling)

| # | Question | Follow-up to expect |
|---|---|---|
| C1 | Give me the 2-minute pitch of your strongest project. | "Which part did you personally build?" |
| C2 | Why that metric, and what was your baseline? | "Is the difference bigger than noise?" |
| C3 | What didn't work? | "How quickly did you find out, and how?" |
| C4 | How would this change at 100× the data or traffic? | "What breaks first?" |
| C5 | What are you proudest of, and what would you change? | "Why didn't you do that the first time?" |

### Self-scoring rubric (0–2 per row, 10 points per answer)

| Dimension | 0 | 1 | 2 |
|---|---|---|---|
| **Structure** | Rambling; hard to follow | Mostly ordered, some backtracking | Clear STAR/CARL or six-beat pitch |
| **Specificity & ownership** | Generic; "we" throughout | One concrete moment, some "I" | Specific moment, clear personal actions and reasons |
| **Result** | No result | Vague result ("it went well") | Concrete result, with a number or a clear comparison when one exists |
| **Reflection** | None | Generic lesson | Specific lesson or change in behaviour since |
| **Time & delivery** | Over 2:30 or under 0:45 | 2:00–2:30, or rushed | 1:15–2:00, calm pace, stops to invite follow-ups |

**Reading your score:** 9–10 ready · 6–8 re-tell the story once more, fixing the lowest row · ≤ 5 rewrite the story in your bank first. Track your scores per story; after a week of drills your weakest *row* (often Result or Reflection) tells you what to practise.

## 📋 Answer Framework & Cheat Sheet

### STAR / CARL one-pager

| Part | Say | Share of a 2-min answer | Watch out for |
|---|---|---|---|
| **S**ituation / **C**ontext | Where, when, what was at stake (1–2 sentences) | ~15% | Over-explaining background |
| **T**ask | Your specific responsibility or goal | ~5% | Describing the team's goal only |
| **A**ction | 3–5 steps *you* took and **why** | ~50–60% | "We", skipping the reasoning |
| **R**esult | What changed; a number or comparison if real | ~15% | No result; inflated or unverifiable numbers |
| **L**earning | What you'd repeat or change | ~10% | Generic lessons ("communication is important") |

### 2-minute project pitch template

```
Problem      : {who} needed {outcome} because {pain}.
Constraints  : {data size / labels / latency / budget / privacy / deadline}.
Approach     : Baseline {simple thing} → {what I built} because {reason}.
Trade-offs   : I chose {X} over {Y}; the cost was {Z}.
Results      : {metric} = {my measured number} vs {baseline number} on {evaluation set}.
What I'd change: {honest limitation} → {next step}.
Close        : "Which part would you like me to go deeper on?"
```

### Résumé bullet formula

`Action verb` + `what you built, with the method` + `measured result vs a comparison` — ≤ 30 words, no tool lists, no numbers you can't defend.

### Drill-down question buckets

**Data** (source, size, labels) · **Splits & leakage** · **Metric** (why this one) · **Baseline** (compared to what) · **Failures** (what didn't work) · **Scale** (100× what breaks) · **Ownership** (what *you* did)

### ✅ Pre-interview checklist

- [ ] Story bank of 8–10 **true** stories, every competency covered by ≥ 2 (run `coverage_report`)
- [ ] Each story rehearsed out loud at least twice with a 2-minute timer
- [ ] "Tell me about yourself" (60–90 s) and "Why this company?" (specific reasons) prepared
- [ ] One project you can deep-dive for 45 minutes: baseline, metric, split, biggest failure, what you'd change
- [ ] Every number on your résumé: you know how it was measured and what it was compared against
- [ ] Every résumé bullet maps to a story (run `review_bullet` on each)
- [ ] Read the job description; mapped 3 requirements to 3 stories
- [ ] Researched the company's product and one recent engineering blog post or talk
- [ ] 3–5 questions to ask, tailored to each interviewer's role
- [ ] Logistics: time zone, video link, quiet room, water, charged laptop, notes with story titles (not scripts)
- [ ] Plan to send a short thank-you note within about a day
