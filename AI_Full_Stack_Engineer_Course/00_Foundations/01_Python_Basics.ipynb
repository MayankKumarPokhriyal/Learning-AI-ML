## 2. Variables and Data Types <a id='2'></a>

### What is a Variable?

Think of a variable like a **labeled box**. You put something in the box, give it a name, and later you can use that name to get back what's inside.

```
┌─────────────┐
│  age = 15   │  ← The box is named 'age' and holds the value 15
└─────────────┘
```

Python has several built-in data types:

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `str` | `"hello"` | Text (string) |
| `bool` | `True` / `False` | Yes or No |
| `None` | `None` | Nothing / empty |

````python
# --- Integers (whole numbers) ---
age = 15
year = 2024
negative_temp = -10

# --- Floats (decimal numbers) ---
height = 5.9
pi = 3.14159
price = 19.99

# --- Strings (text) ---
name = "Alice"         # double quotes work
city = 'New York'      # single quotes also work
greeting = "Hi there!" 

# --- Booleans (True or False only) ---
is_student = True
has_pet = False

# --- None (represents 'nothing') ---
result = None          # like an empty box

# --- Check the type of any variable ---
print(type(age))       # <class 'int'>
print(type(height))    # <class 'float'>
print(type(name))      # <class 'str'>
print(type(is_student))# <class 'bool'>
print(type(result))    # <class 'NoneType'>
````

````python
# --- Type Conversion (changing types) ---
# Sometimes you need to convert between types

age_str = "25"        # This is text, not a number!
age_num = int(age_str) # Convert text to integer
print(age_num + 5)     # Now we can do math: 30

pi_str = str(3.14159)  # Convert number to text
print("Pi is: " + pi_str)  # Now we can join text

num = float("9.5")     # Text to decimal number
print(num * 2)         # 19.0

# Check if something is a specific type
print(isinstance(42, int))     # True
print(isinstance("hi", str))   # True
print(isinstance(3.14, float)) # True
````

## 3. Operators <a id='3'></a>

Operators let you do calculations and comparisons.

### Arithmetic Operators
| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `10 - 4` | `6` |
| `*` | Multiplication | `4 * 3` | `12` |
| `/` | Division | `15 / 4` | `3.75` |
| `//` | Floor Division | `15 // 4` | `3` |
| `%` | Modulus (remainder) | `15 % 4` | `3` |
| `**` | Power | `2 ** 8` | `256` |

````python
# Arithmetic
print(10 + 3)   # 13
print(10 - 3)   # 7
print(10 * 3)   # 30
print(10 / 3)   # 3.333... (always gives float)
print(10 // 3)  # 3 (throws away the decimal — floor division)
print(10 % 3)   # 1 (remainder: 10 = 3*3 + 1)
print(2 ** 10)  # 1024 (2 to the power of 10)

# Useful trick: check if a number is even or odd
number = 42
if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")

# Comparison operators — always return True or False
print(5 == 5)   # True  (equal to)
print(5 != 3)   # True  (not equal to)
print(10 > 5)   # True  (greater than)
print(10 < 5)   # False (less than)
print(5 >= 5)   # True  (greater than or equal)
print(4 <= 3)   # False (less than or equal)

# Logical operators
print(True and False)   # False (both must be True)
print(True or False)    # True  (at least one must be True)
print(not True)         # False (flips True to False)
````

## 4. Strings <a id='4'></a>

A string is any text wrapped in quotes. Strings are everywhere — names, messages, URLs, file paths.

**Analogy:** A string is like a necklace of beads, where each bead is a character. You can count beads, cut the necklace, and rearrange it.

````python
# --- Creating strings ---
name = "Alice"
message = 'Hello, World!'
multiline = """This is
a multiline
string"""

# --- String length ---
print(len(name))      # 5 (number of characters)

# --- Indexing (accessing individual characters) ---
# Positions:  A  l  i  c  e
# Index:      0  1  2  3  4
# Negative:  -5 -4 -3 -2 -1
print(name[0])    # 'A' (first character)
print(name[-1])   # 'e' (last character)
print(name[1:3])  # 'li' (slice: from index 1 up to but not including 3)
print(name[:3])   # 'Ali' (from start to index 3)
print(name[2:])   # 'ice' (from index 2 to end)
print(name[::-1]) # 'ecilA' (reverse the string!)

# --- String concatenation (joining) ---
first = "Hello"
second = "World"
print(first + " " + second)  # 'Hello World'
print(first * 3)              # 'HelloHelloHello'
````

````python
# --- Essential string methods ---
sentence = "  Hello, Python World!  "

print(sentence.upper())          # '  HELLO, PYTHON WORLD!  '
print(sentence.lower())          # '  hello, python world!  '
print(sentence.strip())          # 'Hello, Python World!' (removes spaces at edges)
print(sentence.strip().title())  # 'Hello, Python World!'
print(sentence.replace("Python", "Amazing"))  # replaces a word
print(sentence.find("Python"))   # 8 (index where 'Python' starts)
print(sentence.count("l"))       # 3 (how many times 'l' appears)
print(sentence.startswith("  Hello"))  # True
print(sentence.endswith("!  "))        # True

# --- Splitting and joining ---
words = "apple,banana,cherry"
word_list = words.split(",")  # ['apple', 'banana', 'cherry']
print(word_list)

joined = " | ".join(word_list)  # 'apple | banana | cherry'
print(joined)

# --- f-strings (the modern, easy way to format strings) ---
name = "Alice"
age = 25
score = 98.567

# Old way (ugly):
print("My name is " + name + " and I am " + str(age) + " years old.")

# New way with f-strings (much nicer!):
print(f"My name is {name} and I am {age} years old.")
print(f"Score: {score:.2f}")  # 2 decimal places: 98.57
print(f"2 + 2 = {2 + 2}")     # You can put expressions inside {}
````

## 5. Lists <a id='5'></a>

A list is an **ordered collection** of items. Think of it like a shopping list — items have positions (first item, second item, etc.), and you can add, remove, or change items.

```
fruits = ['apple', 'banana', 'cherry']
           index 0   index 1   index 2
```

````python
# --- Creating lists ---
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [42, "hello", True, 3.14]  # lists can hold different types
empty = []                          # empty list

# --- Accessing items ---
print(fruits[0])   # 'apple' (first item)
print(fruits[-1])  # 'cherry' (last item)
print(fruits[1:3]) # ['banana', 'cherry'] (slice)
print(len(fruits)) # 3

# --- Modifying lists ---
fruits[0] = "mango"          # Change first item
print(fruits)                 # ['mango', 'banana', 'cherry']

fruits.append("grape")        # Add to the END
print(fruits)                 # ['mango', 'banana', 'cherry', 'grape']

fruits.insert(1, "pineapple") # Insert at position 1
print(fruits)                 # ['mango', 'pineapple', 'banana', 'cherry', 'grape']

fruits.remove("banana")       # Remove by value
print(fruits)

popped = fruits.pop()         # Remove and return the last item
print(f"Removed: {popped}")   
print(fruits)

popped_at = fruits.pop(1)     # Remove at index 1
print(f"Removed at index 1: {popped_at}")
````

````python
# --- More list operations ---
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

print(sorted(numbers))          # [1, 1, 2, 3, 3, 4, 5, 5, 6, 9] (doesn't change original)
numbers.sort()                  # sorts IN PLACE (changes original)
print(numbers)

numbers.sort(reverse=True)      # sort descending
print(numbers)

numbers.reverse()               # flip the order
print(numbers)

print(numbers.count(5))         # how many times 5 appears
print(numbers.index(9))         # what position is 9 at

# --- Checking membership ---
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)        # True
print("mango" in fruits)        # False
print("mango" not in fruits)    # True

# --- Combining lists ---
list1 = [1, 2, 3]
list2 = [4, 5, 6]
combined = list1 + list2         # [1, 2, 3, 4, 5, 6]
list1.extend(list2)              # adds list2 items to list1
print(list1)
````

## 6. Tuples <a id='6'></a>

A tuple is like a list, but **immutable** — once created, you can't change it. Think of GPS coordinates: (latitude, longitude) — these are fixed values that shouldn't be accidentally changed.

**Use tuples when:** the data shouldn't change (coordinates, RGB colors, database records).

````python
# --- Creating tuples ---
coordinates = (40.7128, -74.0060)    # NYC coordinates
rgb_red = (255, 0, 0)                # RGB color
person = ("Alice", 25, "Engineer")   # record
singleton = (42,)                     # single-item tuple NEEDS a comma!

# --- Accessing (same as lists) ---
print(coordinates[0])   # 40.7128
print(coordinates[1])   # -74.006
print(person[-1])       # 'Engineer'

# --- Tuples are immutable (cannot be changed) ---
# coordinates[0] = 0  # This would cause a TypeError!

# --- Tuple unpacking (very useful!) ---
latitude, longitude = coordinates
print(f"Latitude: {latitude}, Longitude: {longitude}")

name, age, job = person
print(f"{name} is {age} years old and works as an {job}")

# --- Swapping variables using tuples ---
a, b = 10, 20
print(f"Before: a={a}, b={b}")
a, b = b, a             # swap in one line!
print(f"After: a={a}, b={b}")

# --- Converting between list and tuple ---
my_list = [1, 2, 3]
my_tuple = tuple(my_list)   # list → tuple
back_to_list = list(my_tuple)  # tuple → list
````

## 7. Dictionaries <a id='7'></a>

A dictionary stores data as **key-value pairs** — like a real dictionary where you look up a word (key) to get its definition (value).

```
{'name': 'Alice', 'age': 25, 'city': 'NYC'}
   key    value    key   value   key   value
```

**Analogy:** A phone book — you look up a name (key) to find a phone number (value).

````python
# --- Creating dictionaries ---
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York",
    "hobbies": ["coding", "reading"]  # value can be a list!
}

# --- Accessing values ---
print(person["name"])           # 'Alice'
print(person["hobbies"])        # ['coding', 'reading']

# Safe access — won't crash if key doesn't exist
print(person.get("age"))        # 25
print(person.get("salary"))     # None (key doesn't exist)
print(person.get("salary", 0))  # 0 (default value if missing)

# --- Modifying dictionaries ---
person["age"] = 26              # update existing key
person["email"] = "alice@email.com"  # add new key
print(person)

del person["city"]              # delete a key
print(person)

# --- Useful dictionary methods ---
print(person.keys())            # dict_keys(['name', 'age', 'email', 'hobbies'])
print(person.values())          # dict_values([...])
print(person.items())           # dict_items([('name', 'Alice'), ...])

# --- Check if a key exists ---
print("name" in person)         # True
print("salary" in person)       # False

# --- Iterating over a dictionary ---
for key, value in person.items():
    print(f"  {key}: {value}")
````

````python
# --- Nested dictionaries ---
students = {
    "alice": {"grade": "A", "score": 95},
    "bob":   {"grade": "B", "score": 82},
    "charlie": {"grade": "A", "score": 91}
}

print(students["alice"]["score"])   # 95

# --- Dictionary comprehension ---
scores = {"alice": 95, "bob": 82, "charlie": 91}
passed = {name: score for name, score in scores.items() if score >= 90}
print(passed)  # {'alice': 95, 'charlie': 91}

# --- Merging dictionaries ---
defaults = {"theme": "light", "language": "en", "notifications": True}
user_prefs = {"theme": "dark", "font_size": 14}
merged = {**defaults, **user_prefs}  # user_prefs overrides defaults
print(merged)
````

## 8. Sets <a id='8'></a>

A set is an **unordered collection of unique items** — no duplicates allowed, no fixed order.

**Analogy:** A bag of marbles where each color only appears once.

````python
# --- Creating sets ---
colors = {"red", "green", "blue"}
print(colors)   # order may vary!

# Duplicates are automatically removed
numbers = {1, 2, 2, 3, 3, 3, 4}
print(numbers)  # {1, 2, 3, 4}

# Convert list with duplicates to set to get unique values
tagged = ["python", "java", "python", "c++", "java"]
unique_tags = set(tagged)
print(unique_tags)  # {'python', 'java', 'c++'}

# --- Modifying sets ---
colors.add("yellow")       # add one item
colors.discard("red")      # remove (no error if missing)
print(colors)

# --- Set operations (like Venn diagrams!) ---
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(set_a | set_b)   # Union: {1,2,3,4,5,6,7,8} (all items)
print(set_a & set_b)   # Intersection: {4, 5} (items in BOTH)
print(set_a - set_b)   # Difference: {1, 2, 3} (in A but not B)
print(set_a ^ set_b)   # Symmetric diff: {1,2,3,6,7,8} (in one but not both)

# --- Membership check (sets are FAST for this) ---
print(3 in set_a)       # True
print(9 in set_a)       # False
````

## 9. Conditional Statements <a id='9'></a>

Conditionals let your program make decisions. Like a fork in the road — depending on a condition, it goes left or right.

````python
# --- Basic if / elif / else ---
temperature = 28

if temperature > 35:
    print("It's very hot! Stay hydrated.")
elif temperature > 25:
    print("It's warm and pleasant.")
elif temperature > 15:
    print("It's a bit cool, bring a jacket.")
else:
    print("It's cold! Stay warm.")

# --- Multiple conditions with 'and' / 'or' ---
age = 20
has_id = True

if age >= 18 and has_id:
    print("Welcome!")
else:
    print("Sorry, you cannot enter.")

# --- One-line if (ternary operator) ---
score = 85
result = "Pass" if score >= 60 else "Fail"
print(f"Result: {result}")

# --- Checking for None ---
user_input = None
if user_input is None:
    print("No input provided")
if user_input is not None:
    print("Got input:", user_input)
````

## 10. Loops <a id='10'></a>

Loops repeat code. Instead of writing the same thing 100 times, you write it once and tell Python to repeat it.

**For loop:** "Do this for each item in a collection"  
**While loop:** "Keep doing this as long as a condition is true"

````python
# --- for loop over a list ---
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# --- for loop with range() ---
for i in range(5):          # 0, 1, 2, 3, 4
    print(i, end=" ")       # end=" " prints on same line
print()                     # newline

for i in range(1, 6):       # 1, 2, 3, 4, 5
    print(i, end=" ")
print()

for i in range(0, 10, 2):   # 0, 2, 4, 6, 8 (step of 2)
    print(i, end=" ")
print()

# --- enumerate: get index AND value ---
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

# --- iterating over a dictionary ---
student = {"name": "Alice", "age": 20, "grade": "A"}
for key, value in student.items():
    print(f"  {key} = {value}")
````

````python
# --- while loop ---
counter = 0
while counter < 5:
    print(f"Counter: {counter}")
    counter += 1    # counter = counter + 1

# --- break: exit the loop early ---
for num in range(10):
    if num == 5:
        print("Found 5! Stopping.")
        break       # exits the loop
    print(num, end=" ")
print()

# --- continue: skip to next iteration ---
for num in range(10):
    if num % 2 == 0:    # skip even numbers
        continue
    print(num, end=" ") # only prints odd numbers
print()

# --- Nested loops ---
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i}x{j}={i*j}", end="  ")
    print()
````

## 11. List Comprehensions <a id='11'></a>

List comprehensions are a **short, elegant way** to create lists. Instead of writing a loop to build a list, you can do it in one line.

```python
# The pattern:
[expression for item in iterable if condition]
```

````python
# --- Regular loop vs list comprehension ---

# Old way (loop):
squares = []
for x in range(10):
    squares.append(x ** 2)
print("Loop:", squares)

# New way (comprehension) — same result, one line:
squares = [x ** 2 for x in range(10)]
print("Comprehension:", squares)

# --- With a condition ---
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print("Even squares:", even_squares)  # [0, 4, 16, 36, 64]

# --- String processing ---
words = ["hello", "world", "python", "is", "great"]
upper_long = [w.upper() for w in words if len(w) > 4]
print(upper_long)  # ['HELLO', 'WORLD', 'PYTHON', 'GREAT']

# --- Nested comprehension (creates a matrix) ---
matrix = [[i * j for j in range(1, 4)] for i in range(1, 4)]
for row in matrix:
    print(row)

# --- Dictionary comprehension ---
word_lengths = {word: len(word) for word in words}
print(word_lengths)
````

## 12. Functions <a id='12'></a>

A function is a **reusable block of code**. You define it once, then call it as many times as you want.

**Analogy:** A function is like a blender — you put ingredients in (parameters), it blends (processes), and gives you a smoothie (return value).

````python
# --- Basic function ---
def greet(name):
    """This is a docstring — it explains what the function does."""
    message = f"Hello, {name}!"
    return message

result = greet("Alice")
print(result)  # Hello, Alice!

# --- Multiple parameters ---
def add(a, b):
    return a + b

print(add(3, 5))   # 8
print(add(10, 20)) # 30

# --- Default parameter values ---
def greet_with_title(name, title="Mr."):
    return f"Hello, {title} {name}!"

print(greet_with_title("Smith"))          # Hello, Mr. Smith!
print(greet_with_title("Smith", "Dr."))   # Hello, Dr. Smith!

# --- Returning multiple values ---
def min_max(numbers):
    return min(numbers), max(numbers)   # returns a tuple

low, high = min_max([3, 1, 4, 1, 5, 9])
print(f"Min: {low}, Max: {high}")  # Min: 1, Max: 9
````

````python
# --- *args: accept any number of positional arguments ---
def sum_all(*numbers):      # *numbers collects all args into a tuple
    total = 0
    for n in numbers:
        total += n
    return total

print(sum_all(1, 2))           # 3
print(sum_all(1, 2, 3, 4, 5)) # 15

# --- **kwargs: accept any number of keyword arguments ---
def display_info(**details):   # **details collects all kwargs into a dict
    for key, value in details.items():
        print(f"  {key}: {value}")

display_info(name="Alice", age=25, city="NYC")

# --- Combining all: positional, *args, keyword, **kwargs ---
def mixed(required, *args, keyword_only=None, **kwargs):
    print(f"Required: {required}")
    print(f"Extra args: {args}")
    print(f"Keyword only: {keyword_only}")
    print(f"Extra kwargs: {kwargs}")

mixed("hello", 1, 2, 3, keyword_only="yes", x=10, y=20)
````

## 13. Lambda Functions <a id='13'></a>

A lambda is a **tiny, anonymous function** written in one line. Use it for simple operations where creating a full function would be overkill.

```python
lambda arguments: expression
```

````python
# --- Lambda basics ---
double = lambda x: x * 2
print(double(5))   # 10

add = lambda a, b: a + b
print(add(3, 4))   # 7

greet = lambda name: f"Hello, {name}!"
print(greet("Bob")) # Hello, Bob!

# --- Lambdas shine with sorted(), map(), filter() ---

# Sort a list of dicts by a specific key
students = [
    {"name": "Alice", "score": 95},
    {"name": "Bob",   "score": 82},
    {"name": "Charlie", "score": 91}
]
sorted_students = sorted(students, key=lambda s: s["score"], reverse=True)
for s in sorted_students:
    print(f"  {s['name']}: {s['score']}")

# map: apply a function to every item
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared)   # [1, 4, 9, 16, 25]

# filter: keep only items that pass a test
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)     # [2, 4]
````

## 14. Error Handling <a id='14'></a>

Errors happen. Error handling lets your program deal with problems **gracefully** instead of crashing.

**Analogy:** Like a safety net for a trapeze artist. If something goes wrong, the net catches you instead of letting you hit the ground.

````python
# --- Without error handling (CRASHES) ---
# print(int("abc"))   # ValueError: invalid literal for int() with base 10: 'abc'

# --- With try/except ---
try:
    result = int("abc")   # This will fail
    print(result)
except ValueError:
    print("That's not a valid number!")  # This runs instead of crashing

# --- Catching specific errors ---
try:
    numbers = [1, 2, 3]
    print(numbers[10])      # IndexError: list index out of range
except IndexError:
    print("Index out of range!")
except TypeError:
    print("Wrong type!")

# --- Multiple errors in one except ---
try:
    x = int(input("Enter a number: ") if False else "abc")
except (ValueError, TypeError) as e:
    print(f"Error: {e}")

# --- else: runs if NO error occurred ---
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero!")
else:
    print(f"Success! Result: {result}")  # This runs

# --- finally: runs ALWAYS (for cleanup) ---
try:
    file_data = open("nonexistent.txt")
except FileNotFoundError:
    print("File not found!")
finally:
    print("This always runs — good for cleanup!")  # Always executes
````

````python
# --- Raising your own errors ---
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero!")  # raise forces an error
    return a / b

try:
    print(divide(10, 2))   # 5.0
    print(divide(10, 0))   # Raises ValueError
except ValueError as e:
    print(f"Caught: {e}")

# --- Real-world example: safe user input ---
def get_age():
    user_input = "twenty-five"   # simulating bad input
    try:
        age = int(user_input)
        if age < 0 or age > 150:
            raise ValueError("Age must be between 0 and 150")
        return age
    except ValueError as e:
        print(f"Invalid age: {e}")
        return None

result = get_age()
print(f"Age: {result}")
````

## 15. File I/O <a id='15'></a>

File I/O (Input/Output) means reading from and writing to files. This is how programs save data permanently.

````python
import os

# --- Writing to a file ---
with open("sample.txt", "w") as f:   # 'w' = write (creates or overwrites)
    f.write("Hello, File!\n")
    f.write("This is line 2\n")
    f.write("Python file I/O is easy!\n")

print("File written successfully.")

# --- Reading the entire file ---
with open("sample.txt", "r") as f:   # 'r' = read
    content = f.read()
    print("Full content:")
    print(content)

# --- Reading line by line ---
with open("sample.txt", "r") as f:
    for line_num, line in enumerate(f, 1):
        print(f"Line {line_num}: {line.strip()}")

# --- Reading all lines into a list ---
with open("sample.txt", "r") as f:
    lines = f.readlines()
    print(f"Total lines: {len(lines)}")

# --- Appending to a file (doesn't overwrite) ---
with open("sample.txt", "a") as f:   # 'a' = append
    f.write("This line was added later!\n")

# --- Cleanup ---
os.remove("sample.txt")
print("File deleted.")
````

## 16. 🚀 Mini Project: Student Grade Analyzer <a id='16'></a>

Let's build a complete program that:
1. Stores student scores
2. Calculates statistics
3. Assigns letter grades
4. Saves results to a file

````python
import os

# --- Student data ---
students = {
    "Alice":   [88, 92, 79, 95, 84],
    "Bob":     [72, 68, 75, 71, 80],
    "Charlie": [95, 98, 92, 96, 99],
    "Diana":   [60, 55, 63, 58, 70],
    "Eve":     [85, 88, 90, 82, 87],
}

# --- Function to assign letter grade ---
def get_letter_grade(avg):
    if avg >= 90: return "A"
    elif avg >= 80: return "B"
    elif avg >= 70: return "C"
    elif avg >= 60: return "D"
    else: return "F"

# --- Analyze each student ---
results = []
for name, scores in students.items():
    avg = sum(scores) / len(scores)
    grade = get_letter_grade(avg)
    results.append({
        "name": name,
        "scores": scores,
        "average": round(avg, 2),
        "grade": grade,
        "highest": max(scores),
        "lowest": min(scores)
    })

# --- Sort by average (highest first) ---
results.sort(key=lambda r: r["average"], reverse=True)

# --- Print report ---
print("=" * 55)
print(f"{'STUDENT GRADE REPORT':^55}")
print("=" * 55)
print(f"{'Name':<12} {'Average':>8} {'Grade':>6} {'High':>6} {'Low':>6}")
print("-" * 55)
for r in results:
    print(f"{r['name']:<12} {r['average']:>8.2f} {r['grade']:>6} {r['highest']:>6} {r['lowest']:>6}")

# Class statistics
all_avgs = [r["average"] for r in results]
print("-" * 55)
print(f"Class average:  {sum(all_avgs)/len(all_avgs):.2f}")
print(f"Top performer:  {results[0]['name']} ({results[0]['average']})")
print(f"Grade distribution: { {g: sum(1 for r in results if r['grade']==g) for g in 'ABCDF'} }")
print("=" * 55)

# --- Save results to file ---
with open("grade_report.txt", "w") as f:
    f.write("Student Grade Report\n")
    f.write("=" * 40 + "\n")
    for r in results:
        f.write(f"{r['name']}: {r['average']:.2f} ({r['grade']})\n")

print("\nReport saved to grade_report.txt")
os.remove("grade_report.txt")
````

## 17. 🎤 Interview Questions & Answers <a id='17'></a>

**Q1: What's the difference between a list and a tuple?**  
**A:** Lists are mutable (can be changed), tuples are immutable (cannot be changed). Use tuples for fixed data like coordinates. Lists are slower to iterate but can be modified; tuples are faster and hashable (can be used as dict keys).

**Q2: What is a list comprehension and when should you use it?**  
**A:** A one-line way to create lists: `[x**2 for x in range(10)]`. Use it for simple transformations. Avoid it when the logic is complex — a regular loop is clearer.

**Q3: What's the difference between `=` and `==`?**  
**A:** `=` assigns a value to a variable (`x = 5`). `==` compares two values for equality (`x == 5` returns `True` or `False`).

**Q4: What is `None` in Python?**  
**A:** `None` is Python's way of saying "nothing" or "no value". Functions that don't have a `return` statement return `None`. Always check `if x is None` (not `if x == None`).

**Q5: What's the difference between `is` and `==`?**  
**A:** `==` checks if values are equal. `is` checks if they are the same object in memory. Use `is` only for `None`, `True`, `False`. Example: `[1,2] == [1,2]` is True, but `[1,2] is [1,2]` is False.

**Q6: What are *args and **kwargs?**  
**A:** `*args` lets a function accept any number of positional arguments (stored as a tuple). `**kwargs` lets a function accept any number of keyword arguments (stored as a dict).

**Q7: Why use f-strings over `.format()` or `%` formatting?**  
**A:** f-strings (Python 3.6+) are the most readable and fastest. They let you embed expressions directly: `f"{name.upper()} is {age} years old"`.

**Q8: What is the purpose of the `with` statement?**  
**A:** `with` is a context manager that automatically handles setup and cleanup. `with open('file.txt') as f:` automatically closes the file when done, even if an error occurs.
