## 2. Variables and Core Types 🟢

A **variable** is a *name* that refers to a value. Think of it as a **name tag stuck on an object**, not a box that holds it — section 13 shows why this difference matters.

| Type | Example | Meaning |
|---|---|---|
| `int` | `42`, `-7`, `2**100` | whole numbers of **any** size |
| `float` | `3.14`, `1e-3` | decimal numbers (stored approximately) |
| `str` | `"hello"` | text ("string" of characters) |
| `bool` | `True`, `False` | yes / no |
| `NoneType` | `None` | "no value" |

Python is **dynamically typed** (you don't declare types; the *value* has a type) and **strongly typed** (it won't silently mix incompatible types like `"1" + 1`).

````python
age = 29
height_m = 1.75
name = "Ada"
is_student = False
manager = None

print(type(age), type(height_m), type(name), type(is_student), type(manager))
print("ints never overflow:", 2 ** 100)
print("floats are approximate:", 0.1 + 0.2)

try:                                        # (try/except is explained in section 14)
    "Age: " + age
except TypeError as err:
    print("TypeError:", err)
````

**Converting between types** is something you'll do constantly — data from files, web forms, and APIs arrives as text.

````python
print(int("42") + 1)             # text → int
print(float("3.5") * 2)          # text → float
print(str(42) + "!")             # int → text
print(int(3.99), int(-3.99))     # float → int TRUNCATES toward zero (it does not round)
print(round(2.5), round(3.5))    # round() sends exact ties to the EVEN number ("banker's rounding")

# Truthiness: zero and empty things count as False, everything else as True
print(bool(0), bool(0.0), bool(""), bool([]), bool(None), "|", bool(7), bool("False"), bool([0]))

try:
    int("3.5")                   # int() only parses whole-number text
except ValueError as err:
    print("ValueError:", err)
````

## 3. Operators 🟢

| Kind | Operators | Example → result |
|---|---|---|
| Arithmetic | `+ - * /` | `7 / 2` → `3.5` (always a float) |
| Floor division, remainder, power | `// % **` | `7 // 2` → `3`, `7 % 2` → `1`, `2 ** 10` → `1024` |
| Comparison | `== != < <= > >=` | `3 == 3.0` → `True` |
| Logical | `and or not` | `True and not False` → `True` |
| Augmented assignment | `+= -= *= /=` | `x += 1` means `x = x + 1` |

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(-7 // 2, -7 % 2)          # // rounds DOWN (toward −∞), so −3.5 becomes −4, not −3
print(divmod(17, 5))            # (quotient, remainder) in one call

x = 5
print(0 < x < 10)               # chained comparison: 0 < x and x < 10
print(x > 3 and x < 4, not x > 3)

# `and` / `or` return one of their operands, and stop early ("short-circuit")
print(0 or "default", "|", "" or "empty string is falsy", "|", 3 and 4)

score = 10
score += 5
score *= 2
print("score:", score)
````

> 💡 **Interview angle:** "What is `-7 // 2`?" — `-4`, because floor division rounds toward negative infinity (and `-7 % 2` is `1`, so `(-7 // 2) * 2 + (-7 % 2) == -7` still holds).

## 4. Strings 🟢

A **string** is a sequence of characters. Each character has a position (**index**) starting at **0**; negative indexes count from the end. A **slice** `s[start:stop:step]` takes a piece — `stop` is **excluded**.

```
  text   =   P   y   t   h   o   n
  index  =   0   1   2   3   4   5
  negative  -6  -5  -4  -3  -2  -1        text[1:4] → "yth"     text[::-1] → "nohtyP"
```

Strings are **immutable**: you can't change one in place — methods always return a *new* string.

````python
text = "Python"
print(text[0], text[-1], text[1:4], text[:2], text[2:], text[::2], text[::-1])
print("length:", len(text), "| 'th' in text:", "th" in text)

try:
    text[0] = "J"
except TypeError as err:
    print("TypeError:", err)
print("build a new one instead:", "J" + text[1:])
````

````python
sentence = "  Hello, Python World!  "
print(repr(sentence.strip()))               # remove whitespace at both ends
print(sentence.lower(), "|", sentence.upper())
print(sentence.replace("World", "Community"))
print("find('Python') :", sentence.find("Python"))   # 9 — the two leading spaces count as positions 0 and 1
print("find('Java')   :", sentence.find("Java"))     # -1 means "not found"
print("count('o')     :", sentence.count("o"))
print("startswith     :", sentence.strip().startswith("Hello"))

try:
    sentence.index("Java")                   # like find, but raises an error when not found
except ValueError as err:
    print("index('Java')  → ValueError:", err)

words = "the quick brown fox".split()        # split on whitespace → list of words
print(words, "→", "-".join(words))           # join glues a list of strings together
````

## 5. Lists and Tuples 🟢

- A **list** `[...]` is an **ordered, changeable (mutable)** sequence — like a shopping list you can add to and cross off.
- A **tuple** `(...)` is an **ordered, unchangeable (immutable)** sequence — like GPS coordinates or an RGB colour: a fixed record.

Both use the same indexing and slicing as strings.

````python
scores = [88, 92, 79]
scores.append(95)              # add one item at the end
scores.extend([60, 71])        # add several items at the end
scores.insert(0, 100)          # insert at a position (everything after it shifts right)
print("list  :", scores)

last = scores.pop()            # remove and return the last item
first = scores.pop(0)          # remove and return the item at index 0
scores.remove(79)              # remove the first item EQUAL to 79
print("popped:", last, first, "| now:", scores, "| len:", len(scores))

print("sorted copy :", sorted(scores), "| original unchanged:", scores)
scores.sort(reverse=True)      # sorts IN PLACE and returns None
print("sorted list :", scores)

people = [("Chen", 31), ("Ana", 25), ("Ben", 31)]
print("by age      :", sorted(people, key=lambda person: person[1]))   # key= says WHAT to sort by
print("slice copy  :", scores[1:3], "| every other:", scores[::2])
````

````python
point = (40.7128, -74.0060)    # a tuple
single = (42,)                 # a one-item tuple NEEDS the comma
not_a_tuple = (42)             # just the number 42 in parentheses
print(type(single).__name__, type(not_a_tuple).__name__)

try:
    point[0] = 0
except TypeError as err:
    print("TypeError:", err)

lat, lon = point               # unpacking: one name per item
print(f"lat={lat}, lon={lon}")

a, b = 1, 2
a, b = b, a                    # swap without a temporary variable (it builds and unpacks a tuple)
print("swapped:", a, b)

head, *rest = [10, 20, 30, 40] # star-unpacking collects "the rest" into a list
print("head:", head, "| rest:", rest)
````

## 6. Dictionaries and Sets 🟢

- A **dictionary** (`dict`) maps **keys → values**, like a phone book: look up a name, get a number. Since Python 3.7, dicts remember **insertion order**.
- A **set** holds **unique** items with **no order**, like a bag of distinct marbles. Great for removing duplicates and for "is X in here?" questions.

Keys of a dict and items of a set must be **hashable** — roughly, *immutable*: `str`, `int`, `float`, `tuple` of hashables. Lists and dicts can't be keys.

````python
model_card = {"name": "bert-base", "params_m": 110, "tasks": ["classification", "ner"]}

print(model_card["name"])                          # lookup by key
print(model_card.get("license"))                   # .get → None instead of an error when missing
print(model_card.get("license", "unknown"))        # ... or a default of your choice
model_card["license"] = "apache-2.0"               # add or update
del model_card["params_m"]                         # delete
print("keys:", list(model_card.keys()))

for key, value in model_card.items():              # loop over key-value pairs
    print(f"  {key:>8} → {value}")

defaults = {"lr": 0.001, "epochs": 10}
overrides = {"epochs": 3}
print("merged:", defaults | overrides)             # right side wins for shared keys

try:
    model_card["missing"]
except KeyError as err:
    print("KeyError:", err)

try:
    {["a", "list"]: 1}
except TypeError as err:
    print("TypeError:", err)
````

````python
tags = ["python", "ml", "python", "llm", "ml"]
unique_tags = set(tags)
print("unique:", sorted(unique_tags))              # sets have no order → sort to display

ana_skills = {"python", "sql", "pytorch"}
ben_skills = {"python", "docker", "sql"}
print("both (&)      :", sorted(ana_skills & ben_skills))
print("either (|)    :", sorted(ana_skills | ben_skills))
print("only Ana (-)  :", sorted(ana_skills - ben_skills))
print("not shared (^):", sorted(ana_skills ^ ben_skills))

empty_set = set()                                  # {} is an empty DICT, not a set!
print(type({}).__name__, type(empty_set).__name__)
````

## 7. How Fast Is Each Container? Big-O 🟡

**Big-O notation** describes how the work grows as the data grows (`n` = number of items):

- **O(1) — constant:** same speed for 10 or 10 million items. Like a coat check: your ticket number takes you straight to the hook.
- **O(n) — linear:** 10× more data → ~10× more work. Like searching for a name in an unsorted pile of papers.
- **O(n log n):** a bit worse than linear — good sorting algorithms.

Dicts and sets are **hash tables**: they turn each key into a number (its *hash*) that points almost directly to where it's stored. Lists have to be scanned.

| Operation | `list` | `dict` / `set` |
|---|---|---|
| `x in container` | **O(n)** | **O(1)** average |
| get by position `lst[i]` | O(1) | — |
| get by key `d[k]` | — | O(1) average |
| add at end: `append` / `add` / `d[k] = v` | O(1) *amortized* | O(1) average |
| `insert(0, x)` / `pop(0)` (front) | **O(n)** — every item shifts | — |
| `pop()` from the end | O(1) | — |
| `sort()` / `sorted()` | O(n log n) | — |

*Amortized* O(1) means: usually instant, occasionally slow (when the list grows its storage), but constant on average. *Average* O(1) for hash tables: many keys with colliding hashes can make it O(n) in the worst case.

Let's **measure** it instead of trusting a table.

````python
def microseconds_per_lookup(container, target, repeats=50):
    start = time.perf_counter()
    for _ in range(repeats):
        target in container
    return (time.perf_counter() - start) / repeats * 1e6


sizes = [10_000, 100_000, 1_000_000]
list_times, set_times = [], []
print(f"{'n':>10} | {'list: x in lst':>15} | {'set: x in s':>12}")
for n in sizes:
    numbers_list = list(range(n))
    numbers_set = set(numbers_list)
    missing = -1                                   # worst case for the list: scan everything
    list_times.append(microseconds_per_lookup(numbers_list, missing))
    set_times.append(microseconds_per_lookup(numbers_set, missing))
    print(f"{n:>10,} | {list_times[-1]:>12.1f} µs | {set_times[-1]:>9.3f} µs")

growth = sizes[-1] // sizes[0]
list_growth, set_growth = list_times[-1] / list_times[0], set_times[-1] / set_times[0]
print(f"\nData grew {growth}×: list lookups got {list_growth:.0f}× slower "
      f"({'≈ linear, O(n)' if list_growth > growth / 5 else 'less than linear?'}), "
      f"set lookups {set_growth:.1f}× ({'≈ constant, O(1)' if set_growth < 5 else 'not constant?'}).")
print(f"At n = {sizes[-1]:,} the set is ~{list_times[-1] / set_times[-1]:,.0f}× faster.")
````

````python
n = 30_000
start = time.perf_counter()
front = []
for i in range(n):
    front.insert(0, i)                             # O(n) each time → O(n²) total
front_s = time.perf_counter() - start

start = time.perf_counter()
back = []
for i in range(n):
    back.append(i)                                 # O(1) amortized each time → O(n) total
back_s = time.perf_counter() - start

print(f"insert(0, x) × {n:,}: {front_s * 1000:.0f} ms | append × {n:,}: {back_s * 1000:.1f} ms "
      f"→ {front_s / back_s:.0f}× slower")
print("Need fast adds/removes at BOTH ends? collections.deque — next notebook.")
````

## 8. Conditionals 🟢

`if / elif / else` lets a program choose what to do. The **indentation** (4 spaces) after the colon defines which lines belong to each branch — Python has no `{ }` braces.

```
           ┌─ temperature >= 35 ──► "heat warning"
  start ───┼─ temperature >= 25 ──► "warm"          (checks run top to bottom; the first True wins)

```

## 12. Variable Scope 🟡

**Scope** = where a name is visible. Python looks names up in this order — **LEGB**:

```
  Local  (inside the current function)
    → Enclosing  (an outer function, if nested)
      → Global  (top level of the notebook/module)
        → Built-in  (print, len, ...)
```

The rule that surprises people: **if a function assigns to a name anywhere, that name is local for the whole function.**

````python
threshold = 0.5                                   # global


def classify(probability):
    label = "positive" if probability >= threshold else "negative"   # reading a global is fine
    return label                                                     # `label` is local


print(classify(0.8))
try:
    print(label)
except NameError as err:
    print("NameError:", err, "← locals disappear when the function returns")

call_count = 0


def count_call_buggy():
    call_count += 1                               # assignment → call_count is LOCAL → read before assigned


try:
    count_call_buggy()
except UnboundLocalError as err:
    print("UnboundLocalError:", err)


def count_call(current):                          # ✅ better than `global`: pass the value in, return the new one
    return current + 1


call_count = count_call(call_count)
print("call_count:", call_count)
````

> 💡 **Interview angle:** "Why do I get `UnboundLocalError` when my global variable clearly exists?" — any assignment inside the function makes the name local for the entire function. ## 13. Names vs Objects: Mutability, Aliasing, and Copies 🟡

This is the **most asked** Python fundamentals topic. Three rules explain almost every surprise:

1. **Names are labels, objects are things.** `b = a` never copies — it sticks a second label on the *same* object (**aliasing**).
2. **Mutable** objects (`list`, `dict`, `set`) can change in place; **immutable** ones (`int`, `float`, `str`, `tuple`, `frozenset`) can't — "changing" them makes a new object.
3. **`==` compares values; `is` compares identity** (the same object in memory, see `id()`).

```
  a = [1, 2, 3]        a ──►┌───────────┐
  b = a                b ──►│ [1, 2, 3] │   one list, two names → b.append(4) is visible through a
                            └───────────┘
```

````python
a = [1, 2, 3]
b = a                          # no copy
b.append(4)
print("a:", a, "| a is b:", a is b, "| same id:", id(a) == id(b))

x = 10
y = x
y += 1                         # ints are immutable: this creates a new int 11 and moves the label y
print("x:", x, "| y:", y)
````

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print("equal values:", list1 == list2, "| same object:", list1 is list2)

big1, big2 = int("1000"), int("1000")
print("1000 == 1000:", big1 == big2, "| is:", big1 is big2)
small1, small2 = int("7"), int("7")
print("7 is 7:", small1 is small2, "← CPython caches small ints (-5..256): an implementation detail, never rely on it")

result = None
print("result is None:", result is None, "← `is` is right for singletons like None, True, False")
````

**Passing arguments.** Python passes a *reference to the object* ("call by sharing"): a function can **mutate** a mutable argument, but **reassigning** the parameter only moves the local label.

````python
def add_item(items):
    items.append("added")      # mutates the caller's list


def replace_items(items):
    items = ["replaced"]       # rebinds the local name only; the caller never sees this


data = ["original"]
add_item(data)
replace_items(data)
print(data)

numbers = [1, 2]
alias = numbers
numbers += [3]                 # for lists, += extends IN PLACE (alias sees it)
print("list += :", numbers, alias)

pair = (1, 2)
pair_alias = pair
pair += (3,)                   # for tuples, += builds a NEW tuple (alias unchanged)
print("tuple +=:", pair, pair_alias)

record = ("ada", [90, 85])     # the tuple can't change which objects it holds...
record[1].append(99)           # ...but the list inside it is still mutable
print("tuple holding a list:", record)
````

**Shallow vs deep copy.**
- A **shallow copy** (`lst.copy()`, `lst[:]`, `dict(d)`, `copy.copy(x)`) makes a new *outer* container, but the items inside are **shared**.
- A **deep copy** (`copy.deepcopy(x)`) copies everything recursively — nothing is shared.

```
  original ─►{ "model", "layers" ─►[64, 32] }
  shallow  ─►{ "model", "layers" ──┘            ← same inner list!
  deep     ─►{ "model", "layers" ─►[64, 32] }   ← its own inner list
```

````python
original = {"model": "bert", "layers": [64, 32]}
shallow = copy.copy(original)
deep = copy.deepcopy(original)

shallow["model"] = "gpt"             # top-level change: only affects the shallow copy
shallow["layers"].append(16)         # nested change: SHARED with the original!

print("original:", original)
print("shallow :", shallow)
print("deep    :", deep)
print("inner list shared? shallow:", original["layers"] is shallow["layers"],
      "| deep:", original["layers"] is deep["layers"])
````

## 14. Errors and Exceptions 🟡

An **exception** is Python's way of saying "I can't continue". Unhandled, it stops the program and prints a **traceback** — read it **from the bottom**: the last line has the error type and message, the lines above show where it happened.

```
  try:      code that might fail
  except:   runs only if a matching error happened      (catch SPECIFIC errors)
  else:     runs only if NO error happened
  finally:  runs ALWAYS — error or not (cleanup: close files, release resources)
```

````python
experiments = [
    ("divide by zero", lambda: 1 / 0),           # lambda = a tiny one-line function (notebook 02)
    ("missing dict key", lambda: {"a": 1}["b"]),
    ("list index", lambda: [1, 2, 3][10]),
    ("wrong type", lambda: len(42)),
    ("bad number text", lambda: float("abc")),
    ("missing file", lambda: Path("no_such_file.txt").read_text()),
]
for description, action in experiments:
    try:
        action()
    except Exception as err:                     # broad catch is OK here ONLY because we print every error
        print(f"{description:>16} → {type(err).__name__}: {err}")
````

````python
def parse_score(text):
    try:
        score = float(text)
    except ValueError:
        print(f"  except : {text!r} is not a number")
        return None
    else:
        print(f"  else   : parsed {score}")
        return score
    finally:
        print("  finally: always runs")


for raw in ["87.5", "eighty"]:
    print(f"parse_score({raw!r}) = {parse_score(raw)}")
````

**Raising your own errors** makes bad input fail loudly and early, with a helpful message. `raise NewError(...) from err` keeps the original cause attached.

````python
def set_learning_rate(lr):
    if not isinstance(lr, (int, float)):
        raise TypeError(f"lr must be a number, got {type(lr).__name__}")
    if not 0 < lr <= 1:
        raise ValueError(f"lr must be in (0, 1], got {lr}")
    return lr


for candidate in [0.01, 5, "fast"]:
    try:
        print("accepted:", set_learning_rate(candidate))
    except (TypeError, ValueError) as err:       # catch several specific types
        print(f"rejected: {type(err).__name__}: {err}")


def get_setting(settings, key):
    try:
        return settings[key]
    except KeyError as err:
        raise ValueError(f"missing required setting {key!r}") from err


try:
    get_setting({"lr": 0.1}, "epochs")
except ValueError as err:
    print("ValueError:", err, "| caused by:", repr(err.__cause__))
````

## 🔧 Build It From Scratch

Interviewers sometimes say *"don't use built-ins — implement it"*. Building them yourself shows you understand what the built-ins actually do (and what they cost). Every function below is tested against the real thing.

### 1) `my_len` and `my_max`

````python
def my_len(iterable):
    count = 0
    for _ in iterable:
        count += 1
    return count


def my_max(iterable, key=None):
    """Largest item; `key` works like max(key=...). Raises ValueError on empty input, like max()."""
    iterator = iter(iterable)                 # iter/next: step through any iterable manually
    try:
        best = next(iterator)
    except StopIteration:
        raise ValueError("my_max() iterable argument is empty") from None
    best_key = best if key is None else key(best)
    for item in iterator:
        item_key = item if key is None else key(item)
        if item_key > best_key:               # strictly greater → keeps the FIRST max, like max()
            best, best_key = item, item_key
    return best


for trial in range(200):
    data = [random.randint(-50, 50) for _ in range(random.randint(1, 30))]
    assert my_len(data) == len(data)
    assert my_max(data) == max(data)
    assert my_max(data, key=abs) == max(data, key=abs)
assert my_max(["pear", "fig", "banana"], key=len) == max(["pear", "fig", "banana"], key=len)
assert my_len(x for x in range(7)) == 7       # works on generators, where len() can't

for func in (max, my_max):
    try:
        func([])
    except ValueError as err:
        print(f"{func.__name__}([]) → ValueError: {err}")
print("✅ my_len and my_max match len and max on 200 random lists")
````

**Why is the real `len()` O(1) while `my_len` is O(n)?** Lists, strings, dicts, and sets store their size, so `len` just reads it. `my_len` has to walk every item.

### 2) `my_sorted` with insertion sort

**Insertion sort** works like sorting playing cards in your hand: take the next card, slide it left past every larger card, drop it in place. It is **O(n²)** but simple and **stable** (equal items keep their original order — just like Python's `sorted`).

````python
def my_sorted(iterable, key=None, reverse=False):
    items = list(iterable)                    # never modify the caller's data
    get_key = key if key is not None else (lambda item: item)
    for i in range(1, len(items)):
        current = items[i]
        current_key = get_key(current)
        j = i - 1
        # shift items that should come AFTER current one step right
        while j >= 0 and (get_key(items[j]) < current_key if reverse else get_key(items[j]) > current_key):
            items[j + 1] = items[j]
            j -= 1
        items[j + 1] = current
    return items


for trial in range(200):
    data = [random.randint(0, 20) for _ in range(random.randint(0, 40))]
    assert my_sorted(data) == sorted(data)
    assert my_sorted(data, reverse=True) == sorted(data, reverse=True)
    pairs = [(random.choice("abc"), i) for i in range(random.randint(0, 20))]
    assert my_sorted(pairs, key=lambda p: p[0]) == sorted(pairs, key=lambda p: p[0])        # stability check
    assert my_sorted(pairs, key=lambda p: p[0], reverse=True) == sorted(pairs, key=lambda p: p[0], reverse=True)
print("✅ my_sorted matches sorted (including key=, reverse=, and stability) on 200 random lists")

timings = {}
for n in (1_000, 2_000):
    data = [random.random() for _ in range(n)]
    start = time.perf_counter(); my_sorted(data); mine = time.perf_counter() - start
    start = time.perf_counter(); sorted(data); builtin = time.perf_counter() - start
    timings[n] = mine
    print(f"n={n:,}: my_sorted {mine * 1000:6.0f} ms | sorted {builtin * 1000:.2f} ms")
print(f"Doubling n made insertion sort {timings[2_000] / timings[1_000]:.1f}× slower (O(n²) predicts ~4×).")
````

### 3) A word-frequency counter

Counting things is everywhere in ML: word counts for text features, label counts for class imbalance, token counts for LLM costs. We count words in **the Zen of Python** (a poem hidden inside Python itself) and compare with `collections.Counter`.

````python
import this                                   # importing it prints the Zen of Python
from collections import Counter

zen_text = "".join(this.d.get(ch, ch) for ch in this.s)   # the module stores the poem ROT13-encoded


def tokenize(text):
    cleaned = []
    for raw_word in text.lower().split():
        word_only = raw_word.strip(".,!?*-'\"()")
        if word_only:
            cleaned.append(word_only)
    return cleaned


def word_frequencies(words):
    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    return counts


def top_k(counts, k):
    # sort by count (high → low), then alphabetically so ties have a fixed order
    return sorted(counts.items(), key=lambda pair: (-pair[1], pair[0]))[:k]


tokens = tokenize(zen_text)
mine = word_frequencies(tokens)
reference = Counter(tokens)
assert mine == dict(reference)
assert top_k(mine, 5) == sorted(reference.items(), key=lambda pair: (-pair[1], pair[0]))[:5]
print(f"\n✅ matches Counter: {len(tokens)} words, {len(mine)} unique")
print("top 5:", top_k(mine, 5))
````

## ⚠️ Common Pitfalls

### ❌ Pitfall 1 — Changing a list while looping over it

````python
values = [2, 4, 5, 6, 8]
for v in values:                              # ❌ removing shifts items, so the loop skips some
    if v % 2 == 0:
        values.remove(v)
print("❌ tried to remove evens:", values)

values = [2, 4, 5, 6, 8]
values = [v for v in values if v % 2 != 0]    # ✅ build a new list instead
print("✅ evens removed        :", values)
````

### ❌ Pitfall 2 — `[[0] * 3] * 3` creates three references to ONE row

````python
grid = [[0] * 3] * 3                          # ❌ the outer * copies the reference, not the row
grid[0][0] = 1
print("❌ every row changed:", grid)

grid = [[0] * 3 for _ in range(3)]            # ✅ a new inner list per row
grid[0][0] = 1
print("✅ only row 0       :", grid)
````

### ❌ Pitfall 3 — Using `is` to compare values

````python
expected_status = "model ready"
received_status = " ".join(["model", "ready"])   # e.g. text built at run time or read from a file
print("❌ is :", received_status is expected_status)   # different objects → False
print("✅ == :", received_status == expected_status)   # same text → True
````

### ❌ Pitfall 4 — `if value:` when 0 is a valid value

````python
def describe_discount_buggy(discount=None):
    if discount:                              # ❌ 0 is falsy, so a real 0% discount looks "missing"
        return f"discount {discount}%"
    return "no discount given"


def describe_discount(discount=None):
    if discount is not None:                  # ✅ only None means "missing"
        return f"discount {discount}%"
    return "no discount given"


print("❌", describe_discount_buggy(0))
print("✅", describe_discount(0))
````

### ❌ Pitfall 5 — A broad `except` hides a real bug

````python
bills = [{"tip": 2.0}, {"tip": 3.5}]


def average_tip_buggy(rows):
    try:
        return sum(row["tip"] for row in rows) / len(rowz)     # typo: rowz
    except Exception:                                          # ❌ swallows the NameError
        return 0.0


def average_tip_strict(rows):
    try:
        return sum(row["tip"] for row in rows) / len(rowz)     # same typo
    except ZeroDivisionError:                                  # ✅ only handle the case we expect (empty input)
        return 0.0


print("❌ silently wrong:", average_tip_buggy(bills))
try:
    average_tip_strict(bills)
except NameError as err:
    print("✅ the real bug surfaces immediately → NameError:", err)
````

### ❌ Pitfall 6 — Comparing floats with `==`

````python
total_cost = 0.1 + 0.2
print("❌ total_cost == 0.3          :", total_cost == 0.3, f"(it is really {total_cost!r})")
print("✅ math.isclose(total_cost, 0.3):", math.isclose(total_cost, 0.3))
````

### Step 2 — Parse rows into typed dicts and validate

`csv.DictReader` gives each row as a dict of **strings**. We convert types in one function and reject bad rows loudly.

````python
def parse_row(raw):
    row = {
        "total_bill": float(raw["total_bill"]),
        "tip": float(raw["tip"]),
        "smoker": raw["smoker"] == "Yes",
        "day": raw["day"],
        "time": raw["time"],
        "size": int(raw["size"]),
    }
    if row["total_bill"] <= 0 or row["tip"] < 0 or row["size"] < 1:
        raise ValueError(f"impossible values in row: {raw}")
    row["tip_pct"] = row["tip"] / row["total_bill"] * 100
    return row


with open(tips_path, newline="", encoding="utf-8") as f:              # newline="" is what the csv docs require
    reader = csv.DictReader(f)
    print("columns:", reader.fieldnames)
    tips = [parse_row(raw) for raw in reader]

print(f"parsed {len(tips)} rows")
print("first row:", tips[0])
print("days:", sorted({r["day"] for r in tips}), "| times:", sorted({r["time"] for r in tips}),
      "| party sizes:", sorted({r["size"] for r in tips}))
assert len(tips) == 244, "unexpected row count — did the download change?"
````

### Step 3 — Overall statistics (with our own `mean` and `median`)

````python
def mean(values):
    return sum(values) / len(values)


def median(values):
    ordered = sorted(values)
    mid = len(ordered) // 2
    return ordered[mid] if len(ordered) % 2 == 1 else (ordered[mid - 1] + ordered[mid]) / 2


bills_all = [r["total_bill"] for r in tips]
tip_pcts = [r["tip_pct"] for r in tips]

print(f"bills           : {len(tips)}")
print(f"revenue         : ${sum(bills_all):,.2f}")
print(f"average bill    : ${mean(bills_all):.2f} (median ${median(bills_all):.2f})")
print(f"average tip     : ${mean([r['tip'] for r in tips]):.2f}")
print(f"tip % of bill   : mean {mean(tip_pcts):.1f}% | median {median(tip_pcts):.1f}% | "
      f"min {min(tip_pcts):.1f}% | max {max(tip_pcts):.1f}%")

skew_word = "above" if mean(tip_pcts) > median(tip_pcts) else "below"
print(f"→ the mean tip % is {skew_word} the median, so a few unusual tips pull the average {'up' if skew_word == 'above' else 'down'}.")

import statistics                                                      # verify our versions against the standard library
assert math.isclose(mean(tip_pcts), statistics.mean(tip_pcts)) and math.isclose(median(tip_pcts), statistics.median(tip_pcts))
````

### 🧠 Concepts

**Q1. What is the difference between mutable and immutable objects? Why does it matter?**

<details><summary>Show answer</summary>

- **30-second answer:** Mutable objects (`list`, `dict`, `set`) can change in place; immutable ones (`int`, `float`, `str`, `tuple`, `frozenset`) can't — any "change" creates a new object. It matters for aliasing (two names seeing the same change), for dict keys and set members (they must be hashable, which in practice means immutable), and for default arguments.
- **Go deeper:** Immutability is shallow: a tuple containing a list can't swap the list out, but the list itself can still change — and such a tuple is no longer hashable. Immutable objects are safe to share between functions and threads without copying.
- **❌ Common wrong answer:** "Immutable means the variable can't be reassigned." Names can always be rebound (`s = s + "x"`); it's the *object* that can't change.

</details>

**Q2. When would you use a list, a tuple, a set, or a dict? What are their lookup costs?**

<details><summary>Show answer</summary>

- **30-second answer:** **List** — ordered, growable collection, access by position (O(1) index, O(n) `in`). **Tuple** — fixed record, immutable, hashable (e.g. `(lat, lon)` as a dict key). **Set** — unique items and fast membership (O(1) average `in`). **Dict** — key → value lookup (O(1) average).
- **Go deeper:** `list.append` and `pop()` are amortized O(1), but `insert(0, x)`/`pop(0)` are O(n) — use `collections.deque` for queues. Sets and dicts use more memory than lists because hash tables keep empty slots. Tuples are slightly smaller and faster to create than lists.
- **❌ Common wrong answer:** "Tuples are just faster lists" — the real reason to choose a tuple is immutability/hashability, not speed.

</details>

**Q3. What is the difference between `is` and `==`?**

<details><summary>Show answer</summary>

- **30-second answer:** `==` compares values (it calls `__eq__`); `is` checks whether both names point to the **same object** (same `id`). Use `is` only for singletons such as `None`: `if x is None`.
- **Go deeper:** CPython caches small integers (-5 to 256) and interns some strings, so `a is b` can be `True` for equal small values by accident — that's an implementation detail. Python 3.8+ even emits a `SyntaxWarning` for `x is 5` with a literal.
- **❌ Common wrong answer:** "`is` is a faster `==`." It answers a different question and gives wrong results for equal-but-distinct objects.

</details>

**Q4. Is Python pass-by-value or pass-by-reference?**

<details><summary>Show answer</summary>

- **30-second answer:** Neither exactly — it's *pass-by-object-reference* ("call by sharing"). The function receives a reference to the same object, so it can **mutate** a mutable argument, but **rebinding** the parameter (`items = [...]`) doesn't affect the caller.
- **Go deeper:** That's why `list.append` inside a function is visible outside but `x += 1` on an int is not (ints are immutable, so `+=` rebinds). If a function must not modify its input, copy it first or return a new object.
- **❌ Common wrong answer:** "Pass-by-reference, so reassigning the parameter changes the caller's variable."

</details>

**Q5. What is the difference between a shallow copy and a deep copy?**

<details><summary>Show answer</summary>

- **30-second answer:** A shallow copy (`lst.copy()`, `lst[:]`, `dict(d)`, `copy.copy`) creates a new outer container whose elements are the **same** objects as the original; a deep copy (`copy.deepcopy`) recursively copies nested objects too, so nothing is shared.
- **Go deeper:** Shallow copies are fine when elements are immutable (a list of ints). `deepcopy` keeps a memo dict so shared references and cycles are copied correctly, but it's slow for large structures — often a better design is to build new objects rather than copy and mutate.
- **❌ Common wrong answer:** "`b = a` makes a copy" or "`list(a)` makes a deep copy."

</details>

**Q6. Why are dict and set lookups O(1) on average? When can they be slower?**

<details><summary>Show answer</summary>

- **30-second answer:** They're hash tables: `hash(key)` picks a slot directly, so finding a key doesn't depend on how many keys there are. Collisions (different keys, same slot) are resolved by probing other slots.
- **Go deeper:** With many colliding hashes the worst case degrades to O(n). The table resizes when it fills up (amortized O(1) inserts). Keys must be hashable and their hash must not change — which is why mutable lists can't be keys. Since Python 3.7, dicts preserve insertion order.
- **❌ Common wrong answer:** "Lookups are always O(1), guaranteed" or "dicts are sorted by key."

</details>

**Q7. What do `*args` and `**kwargs` mean?**

<details><summary>Show answer</summary>

- **30-second answer:** In a function definition, `*args` collects extra positional arguments into a **tuple** and `**kwargs` collects extra keyword arguments into a **dict**. In a call, `*` and `**` do the reverse: unpack a sequence or dict into arguments.
- **Go deeper:** Parameters after `*` (or after `*args`) are keyword-only, e.g. `def f(x, *, verbose=False)`. Wrappers and decorators use `def wrapper(*args, **kwargs): return func(*args, **kwargs)` to forward any arguments. The names are convention — only the stars matter.
- **❌ Common wrong answer:** "`**kwargs` is a list of keyword arguments."

</details>

**Q8. List comprehension vs generator expression — what's the difference?**

<details><summary>Show answer</summary>

- **30-second answer:** `[x*x for x in data]` builds the whole list in memory immediately; `(x*x for x in data)` creates a lazy generator that produces one value at a time. Use a generator when you only iterate once or the data is huge (`sum(x*x for x in data)`).
- **Go deeper:** A generator object is ~200 bytes regardless of how many items it will produce, but it can be consumed only once and doesn't support `len()` or indexing. Lists are better when you need random access or several passes.
- **❌ Common wrong answer:** "They're the same, just different brackets" — or "a generator is always faster" (it saves memory; per-item speed is similar or a bit slower).

</details>

### 💻 Coding

**Q9. Remove duplicates from a list while preserving the original order.**

<details><summary>Show answer</summary>

- **30-second answer:** `list(dict.fromkeys(items))` — dict keys are unique and keep insertion order, O(n). Or loop with a `seen` set and append items not seen yet.
- **Go deeper:** `list(set(items))` also removes duplicates but loses the order. If items aren't hashable (e.g. lists), fall back to an O(n²) check or convert them to tuples first.
- **❌ Common wrong answer:** `list(set(items))` when order matters, or a nested loop with `if x not in result` (O(n²)).

</details>

**Q10. Given a list of words, return the k most frequent.**

<details><summary>Show answer</summary>

- **30-second answer:** Count with a dict (or `collections.Counter`), then `sorted(counts.items(), key=lambda kv: (-kv[1], kv[0]))[:k]` — O(n + u log u) for u unique words. `Counter(words).most_common(k)` does it in one call.
- **Go deeper:** For huge vocabularies and small k, use a heap: `heapq.nlargest(k, counts.items(), key=...)` is O(u log k). State the tie-breaking rule explicitly — interviewers often ask about ties.
- **❌ Common wrong answer:** Calling `words.count(w)` inside a loop over `words` — that's O(n²).

</details>

**Q11. Flatten a list of lists, e.g. `[[1, 2], [3], []]` → `[1, 2, 3]`.**

<details><summary>Show answer</summary>

- **30-second answer:** `[x for sub in nested for x in sub]` — read the `for`s left to right like nested loops. `itertools.chain.from_iterable(nested)` is the lazy equivalent.
- **Go deeper:** For *arbitrarily* deep nesting you need recursion (or an explicit stack): if an item is a list, flatten it, otherwise yield it. `sum(nested, [])` works but is O(n²) because each `+` copies.
- **❌ Common wrong answer:** Writing the comprehension with the `for`s in the wrong order (`[x for x in sub for sub in nested]` → `NameError`).

</details>

### 🐛 Debugging Scenarios

**Q12. A function `def log_event(event, history=[])` keeps "remembering" events from previous calls. Why?**

<details><summary>Show answer</summary>

- **30-second answer:** Default values are evaluated **once**, when `def` runs, so every call that doesn't pass `history` appends to the same list object. Use `history=None` and create `history = []` inside the function.
- **Go deeper:** You can see the shared object in `log_event.__defaults__`. The same trap applies to `{}`, `set()`, and objects like `datetime.now()` as defaults (evaluated once, so the timestamp never changes).
- **❌ Common wrong answer:** "Python has a bug with lists" or "the list is global."

</details>

**Q13. `grid = [[0] * 3] * 3; grid[0][0] = 1` sets the first element of *every* row. What happened?**

<details><summary>Show answer</summary>

- **30-second answer:** `* 3` on the outer list repeats the **reference** to one inner list three times — all rows are the same object. Build independent rows with `[[0] * 3 for _ in range(3)]`.
- **Go deeper:** `[0] * 3` on the inner list is fine because ints are immutable. Verify with `grid[0] is grid[1]`. For numeric grids, a NumPy array (`np.zeros((3, 3))`) avoids the issue entirely.
- **❌ Common wrong answer:** "Use `copy.copy(grid)` to fix it" — a shallow copy still contains the same shared row.

</details>

**Q14. Code wrapped in `try: ... except: pass` "works", but results are sometimes silently wrong. What would you change?**

<details><summary>Show answer</summary>

- **30-second answer:** A bare `except:` catches *everything* — including typos (`NameError`), `KeyboardInterrupt`, and `SystemExit` — and `pass` throws the information away. Catch only the specific exceptions you expect and can handle, log them, and let everything else propagate.
- **Go deeper:** Use `else` for code that should run only on success and `finally` (or `with`) for cleanup. When converting errors, use `raise NewError(...) from err` to keep the original traceback. Keep the `try` block small so you know which line can fail.
- **❌ Common wrong answer:** "Replace it with `except Exception: pass`" — still hides bugs, just not Ctrl+C.

</details>
